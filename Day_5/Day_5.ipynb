```python
def stats(numbers):
    low = min(numbers)
    high = max(numbers)
    avg = sum(numbers) / len(numbers)

    return low, high, avg


numbers = [4, 9, 1, 7, 15]

minimum, maximum, average = stats(numbers)

print(f"Minimum: {minimum}")
print(f"Maximum: {maximum}")
print(f"Average: {average}")
```

```python
def vowel_counts(sentence):
    words = sentence.lower().split()
    counts = {}

    for word in words:
        vowel_count = (
            word.count("a")
            + word.count("e")
            + word.count("i")
            + word.count("o")
            + word.count("u")
        )
        counts[word] = vowel_count

    return counts


result = vowel_counts("The Quick Brown Fox Jumps")
print(result)
```

```python
def grade_summary(roster):
    result = []

    for student in roster:
        average = sum(student["scores"]) / len(student["scores"])

        if average >= 90:
            grade = "A"
        elif average >= 80:
            grade = "B"
        elif average >= 70:
            grade = "C"
        elif average >= 60:
            grade = "D"
        else:
            grade = "F"

        result.append({
            "name": student["name"],
            "average": round(average, 1),
            "grade": grade
        })

    return result


roster = [
    {"name": "Reni", "scores": [91, 85, 88]},
    {"name": "Amit", "scores": [78, 82, 90]},
    {"name": "Tara", "scores": [95, 89, 92]}
]

summary = grade_summary(roster)

print(summary)
```

```python
def unique_words(text):
    words = text.lower().split()
    unique = set(words)
    return sorted(unique)


sentence = "Python is easy and Python is powerful"

result = unique_words(sentence)

print(result)
```

```python
def fizzbuzz_map(n):
    result = {}

    for i in range(1, n + 1):
        if i % 15 == 0:
            result[i] = "FizzBuzz"
        elif i % 3 == 0:
            result[i] = "Fizz"
        elif i % 5 == 0:
            result[i] = "Buzz"
        else:
            result[i] = str(i)

    return result


fizzbuzz = fizzbuzz_map(20)

print(fizzbuzz)
```

```python
def password_report(passwords):
    result = {
        "weak": [],
        "medium": [],
        "strong": []
    }

    for pw in passwords:
        has_digit = any(ch.isdigit() for ch in pw)

        if len(pw) >= 10 and has_digit:
            category = "strong"
        elif len(pw) >= 6:
            category = "medium"
        else:
            category = "weak"

        result[category].append(pw)

    return result

passwords = [
    "cat",
    "hello",
    "python",
    "Python",
    "Python1234",
    "Coding2026",
    "abc123"
]

report = password_report(passwords)

print(report)
```
