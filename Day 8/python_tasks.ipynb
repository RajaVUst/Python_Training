```python
# Task 1.1 - Inventory Tracker
items = ["pen", "notebook", "pen", "eraser", "notebook", "pen", "stapler"]
arrival_list = items.copy()  # List is best for preserving arrival order.
first_batch = tuple(items[:3])  # Tuple is best for an immutable first-batch record.
unique_items = set(items)  # Set is best for finding unique item names.
counts = {}
for item in items:
    counts[item] = counts.get(item, 0) + 1  # Dict is best for storing item-to-count mappings.
print(arrival_list)
print(first_batch)
print(unique_items)
print(counts)
```

```python
# Task 1.2 - Comprehensions Rewrite
squares = [n * n for n in range(1, 21) if n % 2 == 0]
word_lengths = {word: len(word) for word in ["python", "java", "c", "kotlin"]}
unique_vowels = {ch for ch in "the quick brown fox jumps over the lazy dog" if ch in "aeiou"}
print(squares)
print(word_lengths)
print(unique_vowels)
```

```python
# Task 1.3 - Nested Comprehension Challenge
sentences = ["the sky is blue", "python is fun", "list comprehensions are handy"]
words = [word for sentence in sentences for word in sentence.split() if len(word) > 2]
unique_words = {word.lower() for sentence in sentences for word in sentence.split()}
print(words)
print(unique_words)
# A nested comprehension becomes less readable when it has multiple conditions or several nested loops; for example, three nested loops are usually clearer as a plain loop.
```

```python
# Task 2.1
students = {
    "alice": [88, 92, 79],
    "bob": [65, 70, 68],
    "carol": [95, 91, 89],
    "dave": []
}
averages = {
    name: round(sum(grades) / len(grades), 1) if grades else 0
    for name, grades in students.items()
}
above_80 = {
    name for name, grades in students.items()
    if grades and all(grade > 80 for grade in grades)
}
sorted_students = sorted(
    averages.items(),
    key=lambda x: x[1],
    reverse=True
)
print(averages)
print(above_80)
print(sorted_students)
```

```python
# Task 2.2
team_a = {"figma", "jira", "slack", "github"}
team_b = {"jira", "slack", "notion", "vscode"}
common = team_a & team_b
only_a = team_a - team_b
combined = team_a | team_b
different = team_a ^ team_b
print(common)
print(only_a)
print(combined)
print(different)
def access_report(team_a, team_b):
    return {
        "common": team_a & team_b,
        "only_a": team_a - team_b,
        "combined": team_a | team_b,
        "different": team_a ^ team_b
    }
print(access_report(team_a, team_b))
```

```python
# Task 3.2
import csv
employees = [
    ["Amit", "Engineering", 60000],
    ["Reni", "HR", 50000],
    ["Tara", "Engineering", 70000],
    ["Sam", "Finance", 55000],
    ["John", "Engineering", 65000]
]
with open("employees.csv", "w", newline="") as f:
    writer = csv.writer(f)
    writer.writerow(["name", "department", "salary"])
    writer.writerows(employees)
data = []
with open("employees.csv", "r", newline="") as f:
    reader = csv.DictReader(f)
    for row in reader:
        row["salary"] = int(row["salary"])
        data.append(row)
        print(row)
dept_salary = {}
for emp in data:
    dept = emp["department"]
    dept_salary[dept] = dept_salary.get(dept, 0) + emp["salary"]
print(dept_salary)
for emp in data:
    if emp["department"] == "Engineering":
        emp["salary"] = round(emp["salary"] * 1.10)
with open("employees_updated.csv", "w", newline="") as f:
    writer = csv.DictWriter(f, fieldnames=["name", "department", "salary"])
    writer.writeheader()
    writer.writerows(data)
```

```python
# Task 4.1
def safe_divide(a, b):
    try:
        return a / b
    except ZeroDivisionError:
        print("Cannot divide by zero.")
        return None
    except TypeError:
        print("Invalid input type.")
        return None
    finally:
        print("Operation attempted")
print(safe_divide(10, 2))
print(safe_divide(10, 0))
print(safe_divide("10", 2))
```

```python
# Task 4.3
import csv
import json
orders = [
    ["O101", "Amit", "2500.50", "Completed"],
    ["O102", "Reni", "1800", "Pending"],
    ["O103", "Tara", "3200.75", "Completed"],
    ["O104", "Amit", "invalid", "Cancelled"],
    ["O105", "Sam", "950", "Pending"],
    ["O106", "Reni", "", "Completed"],
    ["O107", "Tara", "4100", "Completed"],
    ["O108", "Sam", "1250.25", "Pending"]
]
with open("orders.csv", "w", newline="") as f:
    writer = csv.writer(f)
    writer.writerow(["order_id", "customer", "amount", "status"])
    writer.writerows(orders)
try:
    totals = {}
    statuses = set()
    with open("orders.csv", "r", newline="") as f:
        reader = csv.DictReader(f)
        for i, row in enumerate(reader, 1):
            try:
                amount = float(row["amount"])
            except (ValueError, TypeError):
                print(f"Warning: invalid amount on row {i}, customer {row['customer']}")
                continue
            customer = row["customer"]
            totals[customer] = totals.get(customer, 0) + amount
            statuses.add(row["status"])
    summary = {
        "total_per_customer": totals,
        "statuses": sorted(statuses)
    }
    with open("orders_summary.json", "w") as f:
        json.dump(summary, f, indent=2)
    print(summary)
except FileNotFoundError:
    print("orders.csv was not found.")
except Exception as e:
    print(f"Conversion error: {e}")
finally:
    print("Conversion attempt finished")
```
