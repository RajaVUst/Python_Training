```python
# Task 1.1 - Inventory Tracker

items = ["pen", "notebook", "pen", "eraser", "notebook", "pen", "stapler"]
arrival_list = items.copy()
first_batch = tuple(items[:3])
unique_items = set(items)
item_counts = {}

for item in items:
    item_counts[item] = item_counts.get(item, 0) + 1

print("Arrival List:", arrival_list)
print("First Batch:", first_batch)
print("Unique Items:", unique_items)
print("Item Counts:", item_counts)
```

```python
# Task 1.2 - Comprehensions Rewrite

# List comprehension
squares = [n * n for n in range(1, 21) if n % 2 == 0]
print(squares)

# Dictionary comprehension
word_lengths = {
    word: len(word)
    for word in ["python", "java", "c", "kotlin"]
}
print(word_lengths)

# Set comprehension
unique_vowels = {
    ch
    for ch in "the quick brown fox jumps over the lazy dog"
    if ch in "aeiou"
}
print(unique_vowels)
```

```python
#  Task 1.3 - Nested comprehension challenge

sentences = [
    "the sky is blue",
    "python is fun",
    "list comprehensions are handy"
]

# flat list of words longer than 2 characters
long_words = [
    word
    for sentence in sentences
    for word in sentence.split()
    if len(word) > 2
]
print(long_words)

# set of unique words - case insensitive
unique_words = {
    word.lower()
    for sentence in sentences
    for word in sentence.split()
}
print(unique_words)
```

```python
# Task 2.1 - Student Grade Book

students = {
    "alice": [88, 92, 79],
    "bob": [65, 70, 68],
    "carol": [95, 91, 89],
    "david": []  
}

averages = {
    name: sum(grades) / len(grades) if grades else 0
    for name, grades in students.items()
}
print("Average Grades:", averages)

above_80 = {
    name
    for name, grades in students.items()
    if grades and all(grade > 80 for grade in grades)
}
print("Students with every grade above 80:", above_80)

sorted_students = sorted(
    averages.items(),
    key=lambda item: item[1],
    reverse=True
)
print("Students sorted by average:", sorted_students)
```

```python
# TAsk 2.2 - Set operations : Access Control

team_a = {"figma", "jira", "slack", "github"}
team_b = {"jira", "slack", "notion", "vscode"}

common_tools = team_a & team_b
print("Common tools:", common_tools)

team_a_only = team_a - team_b
print("Team A only:", team_a_only)

combined_tools = team_a | team_b
print("Combined tools:", combined_tools)

symmetric_difference = team_a ^ team_b
print("Tools only one team has:", symmetric_difference)

def access_report(team_a, team_b):
    return {
        "common": team_a & team_b,
        "team_a_only": team_a - team_b,
        "combined": team_a | team_b,
        "symmetric_difference": team_a ^ team_b
    }

report = access_report(team_a, team_b)
print(report)
```

```python
# Task 2.3 - Mutability Trap

def add_item(item, basket=[]):
    basket.append(item)
    return basket

print(add_item("apple"))
print(add_item("banana"))

# The default list is created only once, when the function is defined.
#It is not recreated every time the function is called.

# Correct way to fix it

def add_item(item, basket=None):
    if basket is None:
        basket = []

    basket.append(item)
    return basket

print(add_item("apple"))
print(add_item("banana"))

# another mutability example
original = ["apple", "banana"]
copy = original
copy.append("orange")

print("Original:", original)
print("Copy:", copy)
```

```python
# Task 3.2 - CSV: Employee records

import csv

# 1. Create employee records
employees = [
    ["Chaila", "Engineering", 75000],
    ["Shree", "HR", 60000],
    ["Dhilip", "Engineering", 80000],
    ["Asmitha", "Sales", 65000],
    ["Lekhya", "HR", 55000]
]

# 2. Write records to employees.csv
with open(r"C:\Users\308333\Documents\Python_Training\Day 8\employees.csv", "w", newline="") as file:
    writer = csv.writer(file)

    # Header
    writer.writerow(["name", "department", "salary"])

    # Employee records
    writer.writerows(employees)

print("employees.csv created successfully.")

# 3. Read using csv.DictReader
employee_records = []

with open(r"C:\Users\308333\Documents\Python_Training\Day 8\employees.csv", "r", newline="") as file:
    reader = csv.DictReader(file)

    for employee in reader:
        print(employee)
        employee_records.append(employee)


# 4. Compute total salary per department
department_salary = {}

for employee in employee_records:
    department = employee["department"]
    salary = float(employee["salary"])

    if department not in department_salary:
        department_salary[department] = 0

    department_salary[department] += salary

print("\nTotal salary per department:")

for department, total in department_salary.items():
    print(f"{department}: {total}")


# 5. Add 10% raise to Engineering employees
updated_employees = []

for employee in employee_records:
    salary = float(employee["salary"])

    if employee["department"] == "Engineering":
        salary = salary * 1.10

    employee["salary"] = salary
    updated_employees.append(employee)


# 6. Write updated records to employees_updated.csv
with open(r"C:\Users\308333\Documents\Python_Training\Day 8\employees_updated.csv", "w", newline="") as file:
    fieldnames = ["name", "department", "salary"]

    writer = csv.DictWriter(file, fieldnames=fieldnames)

    writer.writeheader()
    writer.writerows(updated_employees)

print("\nemployees_updated.csv created successfully.")
```

```python
# Task 4.1 - try - except - finally Basics

# Function to safely divide two values
def safe_divide(a, b):

    try:
        # Try to divide a by b
        return a / b

    except ZeroDivisionError:
        # Handle division by zero
        print("Warning: Cannot divide by zero.")
        return None

    except TypeError:
        # Handle incorrect data types
        print("Warning: Invalid data type. Please provide numbers.")
        return None

    finally:
        # This will always execute
        print("Operation attempted")


# Test 1: Normal division
print("Test 1:")
result = safe_divide(10, 2)
print("Result:", result)


# Test 2: Division by zero
print("\nTest 2:")
result = safe_divide(10, 0)
print("Result:", result)


# Test 3: TypeError
print("\nTest 3:")
result = safe_divide("10", 2)
print("Result:", result)
```

```python
# Task 4.2 - Concept check

# 1. Specific exception vs bare except
     # A specific exception such as ZeroDivisionError catches only the error we expect and can handle. This makes the program easier to understand and debug.
     # A bare except: catches almost every exception, including unexpected errors. It can hide real programming problems and make debugging difficult. Therefore, specific exceptions should normally be preferred.


# 2. Purpose of finally
    # finally contains code that should execute whether an exception occurs or not.
    # For example, if a program manually opens a database connection, finally can be used to close the connection even when an error occurs.

# 3. Why catch file exceptions separately?
    #Different exceptions indicate different problems.
    #FileNotFoundError → the requested file doesn't exist.
    #PermissionError → the program doesn't have permission to access the file.
    #IsADirectoryError → a directory was supplied where a file was expected.
    #Catching them separately allows the program to give a more useful message and take the appropriate action for each problem. A generic Exception does not provide this level of specific handling.
```

```python
# Task 4.3 -  Robust CSV -> JSON Convertor

import csv
import json

customer_totals = {}
statuses = set()

try:
    with open(r"C:\Users\308333\Documents\Python_Training\Day 8\orders.csv", "r", newline="") as file:
        reader = csv.DictReader(file)

        for row in reader:

            try:
                amount = float(row["amount"])

            except (ValueError, TypeError):
                print(
                    f"Warning: Invalid amount for row {row['order_id']} "
                    f"customer {row['customer']}. Skipping row."
                )
                continue

            # Add amount to customer total
            if row["customer"] not in customer_totals:
                customer_totals[row["customer"]] = 0

            customer_totals[row["customer"]] += amount

            # Add status to set
            statuses.add(row["status"])

    # Create final summary
    summary = {
        "customer_totals": customer_totals,
        "statuses": sorted(statuses)
    }

    # Write summary to JSON
    with open(r"C:\Users\308333\Documents\Python_Training\Day 8\orders_summary.json", "w") as file:
        json.dump(summary, file, indent=2)

    print("Conversion completed successfully.")

except FileNotFoundError:
    print("Error: orders.csv was not found.")

except Exception as e:
    print(f"Unexpected error: {e}")

finally:
    print("Conversion attempt finished")
```
