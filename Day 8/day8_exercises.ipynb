```python
## Task 1.1 — Inventory Tracker
items = ["pen", "notebook", "pen", "eraser", "notebook", "pen", "stapler"]

# List: preserves the original arrival order and allows duplicates.
copy_list = items.copy()

# Tuple: stores the first 3 items as an immutable first-batch record.
first_batch = tuple(items[:3])

# Set: stores only unique item names, removing duplicates.
unique_items = set(items)

# Dictionary: maps each item name to the number of times it was received.
items_count = {}
for i in items:
    items_count[i] = items_count.get(i,0) + 1

print(copy_list)
print(first_batch)
print(unique_items)
print(items_count)

```

```python
## Task 1.3 — Nested Comprehension Challenge
sentences = ["the sky is blue", "python is fun", "list comprehensions are handy"]

# Flat list of words longer than 2 characters from all sentences
long_words = [word for sentence in sentences for word in sentence.split() if len(word) > 2]

# Set of all unique words, ignoring case
unique_words = {word.lower() for sentence in sentences for word in sentence.split()}

print("Long words:", long_words)
print("Unique words:", unique_words)

# Checkpoint: In my opinion, a nested comprehension becomes less readable when it has
# multiple loops and several conditions
```

```python
## Task 2.1 — Student Grade Book
students = {
    "alice": [88, 92, 79],
    "bob": [65, 70, 68],
    "carol": [95, 91, 89],
}

# Dict comprehension: calculate the average for each student.
# If the grade list is empty, use 0 instead of dividing by zero.
averages = {
    name: sum(grades) / len(grades) if grades else 0
    for name, grades in students.items()
}

# Set comprehension: find students whose every grade is above 80.
above_80 = {
    name
    for name, grades in students.items()
    if grades and all(grade > 80 for grade in grades)
}

# Sort by average grade in descending order and store as (name, average) tuples.
sorted_students = sorted(
    averages.items(),
    key=lambda x: x[1],
    reverse=True
)

print("Averages:", averages)
print("Students with every grade above 80:", above_80)
print("Sorted students:", sorted_students)
```

```python
## Task 2.2 — Set Operations: Access Control
team_a = {"figma", "jira", "slack", "github"}
team_b = {"jira", "slack", "notion", "vscode"}

# Tools common to both teams
common_tools = team_a & team_b

# Tools only team A has
team_a_only = team_a - team_b

# Combined toolset of both teams
combined_tools = team_a | team_b

# Tools that only one of the teams has
symmetric_diff = team_a ^ team_b

print("Common tools:", common_tools)
print("Team A only:", team_a_only)
print("Combined tools:", combined_tools)
print("Symmetric difference:", symmetric_diff)
```

```python
## Task 2.3 — Mutability Trap

# The default list is created only once, when the function is defined—not every time the function is called.
#That's the mutable default argument trap.

def add_item(item, basket=None):
    if basket is None:
        basket = []

    basket.append(item)
    return basket

print(add_item("apple"))
print(add_item("banana"))
```

```python
## ## Task 2.3 — Mutability Trap
# Second example - make a copy 
def add_student(student, students):
    new_students = students.copy()
    new_students.append(student)
    return new_students

original = ["Alice", "Bob"]

new_list = add_student("Carol", original)

print("Original:", original)
print("New:", new_list)
```

```python
## Task 3.1 — Basic Read/Write
# 1. Write 5 log lines to log.txt using "w" and a with block

log_lines = [
    "2026-08-21 09:00 INFO Application started\n",
    "2026-08-21 09:05 INFO User logged in\n",
    "2026-08-21 09:10 WARNING High memory usage\n",
    "2026-08-21 09:15 ERROR Database connection failed\n",
    "2026-08-21 09:20 INFO Application stopped\n"
]

with open("log.txt", "w") as file:
    for line in log_lines:
        file.write(line)


# 2. Read the file line by line and print line numbers

with open("log.txt", "r") as file:
    for line_number, line in enumerate(file, start=1):
        print(line_number, line.strip())


# 3. Manual open/close - deliberately forgetting to close

file = open("log.txt", "r")

for line in file:
    print(line.strip())

# file.close()  # Forgot to close the file!
# Risk: The file handle may remain open, causing a file handle/resource leak.
# Unflushed buffers can also cause data not to be written properly when writing.


# 4. Correct approach using "with" again

with open("log.txt", "r") as file:
    for line in file:
        print(line.strip())


# 5. Append mode - adds new content without overwriting existing content

with open("log.txt", "a") as file:
    file.write("2026-08-21 09:25 INFO New log entry\n")
```

```python
## Task 3.2 — CSV: Employee Records
import csv

# 1. Create employees.csv with 5 employees
employees = [
    ["Alice", "Engineering", 80000],
    ["Bob", "HR", 60000],
    ["Carol", "Engineering", 90000],
    ["David", "Sales", 70000],
    ["Eve", "HR", 65000]
]

with open("employees.csv", "w", newline="") as file:
    writer = csv.writer(file)

    # Write column headers
    writer.writerow(["name", "department", "salary"])

    # Write employee records
    writer.writerows(employees)


# 2. Read the CSV using DictReader
employee_records = []

with open("employees.csv", "r", newline="") as file:
    reader = csv.DictReader(file)

    for employee in reader:
        print(employee)
        employee_records.append(employee)


# 3. Compute total salary per department
department_salary = {}

for employee in employee_records:
    department = employee["department"]
    salary = float(employee["salary"])

    department_salary[department] = (
        department_salary.get(department, 0) + salary
    )

print("\nTotal salary per department:")
print(department_salary)


# 4. Add 10% raise to Engineering employees
updated_employees = []

for employee in employee_records:
    updated_employee = employee.copy()

    if updated_employee["department"] == "Engineering":
        salary = float(updated_employee["salary"])
        updated_employee["salary"] = salary * 1.10

    updated_employees.append(updated_employee)


# 5. Write updated employees to employees_updated.csv
with open("employees_updated.csv", "w", newline="") as file:
    fieldnames = ["name", "department", "salary"]

    writer = csv.DictWriter(file, fieldnames=fieldnames)

    writer.writeheader()
    writer.writerows(updated_employees)

print("\nUpdated employee records written to employees_updated.csv")
```

```python
## Task 3.3 — JSON: Config File  
import json

config = {
    "app_name": "InventoryApp",
    "version": 1.2,
    "features": ["export", "search", "notifications"],
    "limits": {
        "max_users": 100,
        "max_items": 5000
    }
}


# 1. Write config to config.json with pretty formatting

with open("config.json", "w") as file:
    json.dump(config, file, indent=2)


# 2. Read config back and access max_users

with open("config.json", "r") as file:
    config = json.load(file)

print("Maximum users:", config["limits"]["max_users"])


# 3. Add "dark_mode" to features and save the updated config

config["features"].append("dark_mode")

with open("config.json", "w") as file:
    json.dump(config, file, indent=2)

print("Updated features:", config["features"])


# 4. Deliberately corrupt the JSON file

with open("config.json", "w") as file:
    file.write('''
{
  "app_name": "InventoryApp",
  "version": 1.2,
  "features": ["export", "search", "notifications", "dark_mode"],
  "limits": {
    "max_users": 100,
    "max_items": 5000
  }
''')  # Missing closing }


# 5. Try to load the corrupted JSON

try:
    with open("config.json", "r") as file:
        config = json.load(file)

except json.JSONDecodeError as e:
    print("Invalid JSON:", e)
```

```python
## Task 4.1 — try/except/finally Basics
def safe_divide(a, b):
    try:
        return a / b

    except ZeroDivisionError:
        print("Warning: Cannot divide by zero.")
        return None

    except TypeError:
        print("Warning: Both values must be numbers.")
        return None

    finally:
        print("Operation attempted")


# Normal division
print(safe_divide(10, 2))

# ZeroDivisionError
print(safe_divide(10, 0))

# TypeError
print(safe_divide("10", 2))
```

```python
## Task 4.3 — Capstone: Robust CSV → JSON Converter
import csv
import json


# Create sample orders.csv with 8 rows.
# Two rows contain invalid/missing amounts to test exception handling.
orders = [
    ["order_id", "customer", "amount", "status"],
    ["ORD001", "Alice", "250.50", "completed"],
    ["ORD002", "Bob", "120.00", "pending"],
    ["ORD003", "Alice", "300.75", "completed"],
    ["ORD004", "Carol", "invalid", "cancelled"],
    ["ORD005", "Bob", "450.00", "completed"],
    ["ORD006", "David", "", "pending"],
    ["ORD007", "Carol", "175.25", "completed"],
    ["ORD008", "Alice", "100.00", "pending"],
]

with open("orders.csv", "w", newline="") as file:
    writer = csv.writer(file)
    writer.writerows(orders)


try:
    customer_totals = {}
    statuses = set()

    # Read orders.csv using DictReader inside a with block.
    with open("orders.csv", "r", newline="") as file:
        reader = csv.DictReader(file)

        for row in reader:
            customer = row["customer"]

            try:
                amount = float(row["amount"])
            except (ValueError, TypeError):
                print(
                    f"Warning: Invalid amount for "
                    f"order {row['order_id']} / customer {customer}. Skipping row."
                )
                continue

            # Add the status to the set of distinct statuses.
            statuses.add(row["status"])

            # Add the amount to the customer's total.
            customer_totals[customer] = (
                customer_totals.get(customer, 0) + amount
            )

    # Build the final summary.
    summary = {
        "customer_totals": customer_totals,
        "statuses": sorted(statuses),
    }

    # Write summary to JSON.
    with open("orders_summary.json", "w") as file:
        json.dump(summary, file, indent=2)

    print("\nConversion successful!")
    print("Customer totals:", customer_totals)
    print("Statuses:", statuses)

except FileNotFoundError:
    print("Error: orders.csv was not found.")

except Exception as e:
    print(f"Unexpected error: {e}")

finally:
    print("Conversion attempt finished")
```
