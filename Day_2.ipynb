```python
# Exercise 1: Name Tag Formatter

first_name = input("Enter your first name: ")
last_name = input("Enter your last name: ")

first_name = first_name.title()
last_name = last_name.title()

print(f"Hello, {first_name} {last_name}!")
```

```python
# Exercise 2: Coupon Code Slicer

code = "PYTHON2026SALE"

print("First 6 characters:", code[:6])
print("Last 4 characters:", code[-4:])
print("Reversed code:", code[::-1])
```

```python
# Exercise 3: Palindrome Checker

word = "level"

is_palindrome = (word == word[::-1])

print(is_palindrome)
```

```python
# Exercise 4: Email Splitter

email = "ayappabandari.2003@gmail.com"

username, domain = email.split("@")

print("Username:", username)
print("Domain:", domain)
```

```python
# Exercise 5: Slug Normalizer

title = " My First Python Project!! "

clean = title.strip()
print("After strip:", clean)

clean = clean.replace("!!", "")
print("After removing !!:", clean)

clean = clean.lower()
print("After lowercase:", clean)

clean = clean.replace(" ", "_")
print("Final slug:", clean)
```

```python
# Exercise 6: Sentence Dashboard

sentence = input("Enter a sentence: ")

character_count = len(sentence)
word_count = len(sentence.split())

vowel_count = (
    sentence.count("a")
    + sentence.count("e")
    + sentence.count("i")
    + sentence.count("o")
    + sentence.count("u")
)

longer_than_30 = len(sentence) > 30

print("Character count:", character_count)
print("Word count:", word_count)
print("Vowel count:", vowel_count)
print("Longer than 30 characters:", longer_than_30)
```

```python
# Exercise 7: Initials Generator

full_name = "Mani Kanta"

name_parts = full_name.split()

initials = name_parts[0][0] + name_parts[1][0]

print("Initials:", initials)
```

```python
# Exercise 8: Loading Bar

full_bar = "=" * 20
partial_bar = "=" * 8 + "-" * 12

print("Full:", full_bar)
print("Partial:", partial_bar)
```

```python
# Exercise 9: File Extension Checker

filename = "day2_notes.docx"

is_docx = filename.endswith(".docx")
is_pdf = filename.endswith(".pdf")

print("Ends with .docx:", is_docx)
print("Ends with .pdf:", is_pdf)
```

```python
# Exercise 10: Find and Replace Report

text = "the quick brown fox"

index = text.find("brown")
new_text = text.replace("brown", "red")

print("Original text:", text)
print("Index of 'brown':", index)
print("New text:", new_text)
```

```python
# Exercise 11: Username Validator

username = "reni_k99"

is_alphanumeric = username.isalnum()
username_length = len(username)
is_at_least_6 = username_length >= 6

print("Is alphanumeric:", is_alphanumeric)
print("Length:", username_length)
print("At least 6 characters:", is_at_least_6)
```

```python
# Exercise 12: Title Case vs Capitalize

phrase = "the quick brown fox"

print("Title:", phrase.title())
print("Capitalize:", phrase.capitalize())
```

```python
# Exercise 13: Receipt Line Formatter

item = "Notebook"
price = 149.0

print(f"{item:<15}|{price:>8.2f}")
```
