```python
#Task 1.1 — Inventory Tracker

items = ["pen", "notebook", "pen", "eraser", "notebook", "pen", "stapler"]
# List is used because we want to preserve the arrival order.
item_list = items
# Tuple is used because the first batch should be immutable.
first_batch = tuple(items[:3])
# Set is used because we only need unique item names.
unique_items = set(items)
# Dictionary is used to store each item and its count.
item_counts = {}
for item in items:
    item_counts[item] = item_counts.get(item, 0) + 1
print("List:", item_list)
print("Tuple:", first_batch)
print("Set:", unique_items)
print("Dictionary:", item_counts)
```

```python
#Task 1.2 — Comprehensions Rewrite

# List comprehension
squares = [n * n for n in range(1, 21) if n % 2 == 0]
# Dictionary comprehension
word_lengths = {
    word: len(word)
    for word in ["python", "java", "c", "kotlin"]
}
# Set comprehension
unique_vowels = {
    ch
    for ch in "the quick brown fox jumps over the lazy dog"
    if ch in "aeiou"
}
print("Squares:", squares)
print("Word lengths:", word_lengths)
print("Unique vowels:", unique_vowels)
```

```python
#Task 1.3 — Nested Comprehension Challenge

#1. Words longer than 2 characters
sentences = [
    "the sky is blue",
    "python is fun",
    "list comprehensions are handy"
]
long_words = [
    word
    for sentence in sentences
    for word in sentence.split()
    if len(word) > 2
]
print("Words longer than 2 characters:")
print(long_words)

#2. Unique words — case insensitive
unique_words = {
    word.lower()
    for sentence in sentences
    for word in sentence.split()
}
print("Unique words:")
print(unique_words)
```

```python
#Task 2.1 — Student Grade Book

students = {
    "alice": [88, 92, 79],
    "bob": [65, 70, 68],
    "carol": [95, 91, 89],
    "david": []
}
# Calculate average safely
averages = {
    name: sum(grades) / len(grades) if grades else 0
    for name, grades in students.items()
}
print("Average grades:")
print(averages)
# Students whose every grade is above 80
above_80 = {
    name
    for name, grades in students.items()
    if grades and all(grade > 80 for grade in grades)
}
print("\nStudents with every grade above 80:")
print(above_80)
# Sort by average in descending order
sorted_students = sorted(
    averages.items(),
    key=lambda item: item[1],
    reverse=True
)
print("\nStudents sorted by average:")
print(sorted_students)
```

```python
#Task 3.2 — CSV Employee Records

import csv
employees = [
    ["Alice", "Engineering", 75000],
    ["Bob", "HR", 55000],
    ["Carol", "Engineering", 82000],
    ["David", "Finance", 65000],
    ["Eva", "Engineering", 70000]
]

# Create employees.csv
with open("employees.csv", "w", newline="") as file:
    writer = csv.writer(file)
    writer.writerow(["name", "department", "salary"])
    writer.writerows(employees)

# Read using DictReader
employee_records = []
with open("employees.csv", "r", newline="") as file:
    reader = csv.DictReader(file)
    for employee in reader:
        employee["salary"] = float(employee["salary"])
        employee_records.append(employee)
        print(employee)

# Calculate total salary per department
department_salary = {}
for employee in employee_records:
    department = employee["department"]
    salary = employee["salary"]

    department_salary[department] = (
        department_salary.get(department, 0) + salary
    )

print("\nTotal salary per department:")
for department, total in department_salary.items():
    print(department, total)

# Give Engineering employees a 10% raise
for employee in employee_records:
    if employee["department"] == "Engineering":
        employee["salary"] *= 1.10

# Write updated employees
with open("employees_updated.csv", "w", newline="") as file:
    writer = csv.DictWriter(
        file,
        fieldnames=["name", "department", "salary"]
    )
    writer.writeheader()
    writer.writerows(employee_records)
print("\nUpdated employee records saved.")
```

```python
#Task 4.1 — try / except / finally

def safe_divide(a, b):
    try:
        result = a / b
        return result
    except ZeroDivisionError:
        print("Warning: Cannot divide by zero.")
        return None
    except TypeError:
        print("Warning: Invalid data type.")
        return None
    finally:
        print("Operation attempted")
print(safe_divide(10, 2))
print()
print(safe_divide(10, 0))
print()
print(safe_divide("10", 2))
```

```python
#Task 4.2 — Exception Handling Concept Check

1. Specific exception vs bare except
A specific exception, such as ZeroDivisionError, handles only the expected type of error. A bare except: catches almost all exceptions. Bare except: is usually considered bad practice because it can hide unexpected errors and make debugging difficult.

2. Purpose of finally
finally is used for code that must execute whether an exception occurs or not. For example, it can be used to close a file or database connection after completing an operation.

3. Why catch file exceptions separately
Different exceptions indicate different problems. FileNotFoundError means the file does not exist, PermissionError means access is denied, and IsADirectoryError means the given path is a directory. Catching them separately allows the program to provide the correct error message and handle each problem appropriately.
```

```python
#Task 4.3 — Capstone: Robust CSV → JSON Converter

#Step 1 — Create orders.csv

import csv
orders = [
    ["order_id", "customer", "amount", "status"],
    [1001, "Alice", "250.50", "Completed"],
    [1002, "Bob", "120.00", "Pending"],
    [1003, "Alice", "300.00", "Completed"],
    [1004, "Carol", "invalid", "Cancelled"],
    [1005, "David", "450.75", "Completed"],
    [1006, "Bob", "", "Pending"],
    [1007, "Carol", "200.00", "Completed"],
    [1008, "Alice", "150.25", "Pending"],
    [1009, "David", "500.00", "Completed"]
]
with open("orders.csv", "w", newline="") as file:
    writer = csv.writer(file)
    writer.writerows(orders)
print("orders.csv created.")

#Step 2 — convert_orders.py

import csv
import json
def convert_orders():
    customer_totals = {}
    statuses = set()
    try:
        with open("orders.csv", "r", newline="") as file:
            reader = csv.DictReader(file)
            for row in reader:
# Store every status
                statuses.add(row["status"])
                try:
                    amount = float(row["amount"])
                except (ValueError, TypeError):
                    print(
                        f"Warning: Invalid amount for "
                        f"order {row['order_id']} / "
                        f"customer {row['customer']}. Skipping row."
                    )
                    continue
                customer = row["customer"]
# Add amount to customer's total
                customer_totals[customer] = (
                    customer_totals.get(customer, 0) + amount
                )
# Create final summary
        summary = {
            "customer_totals": customer_totals,
            "statuses": sorted(statuses)
        }
# Write summary to JSON
        with open("orders_summary.json", "w") as file:
            json.dump(summary, file, indent=2)
        print("\nConversion successful.")
        print("Customer totals:", customer_totals)
        print("Statuses:", statuses)
    except FileNotFoundError:
        print("Error: orders.csv was not found.")
    except Exception as error:
        print(f"Unexpected error: {error}")
    finally:
        print("Conversion attempt finished")
convert_orders()

#Generated orders_summary.json

{
  "customer_totals": {
    "Alice": 700.75,
    "Bob": 120.0,
    "David": 950.75,
    "Carol": 200.0
  },
  "statuses": [
    "Cancelled",
    "Completed",
    "Pending"
  ]
}
```
