# Collections Fundamentals

```python
# Task 1.1 — Inventory Tracker
items = ["pen", "notebook", "pen", "eraser", "notebook", "pen", "stapler"]

# List is appropriate because it preserves every item and its arrival order.
arrival_order = list(items)

# Tuple is appropriate because the first batch should be immutable.
first_batch = tuple(items[:3])

# Set is appropriate because we only need unique item names.
unique_items = set(items)

# Dict is appropriate because it maps each item name to its number of occurrences.
item_counts = {}

for item in items:
    item_counts[item] = item_counts.get(item, 0) + 1
print("Arrival order:", arrival_order)
print("First batch:", first_batch)
print("Unique items:", unique_items)
print("Item counts:", item_counts)
```

```python
#Task 1.2 — Comprehensions Rewrite
# List comprehension
squares = [n * n for n in range(1, 21) if n % 2 == 0]

# Dict comprehension
word_lengths = {word: len(word) for word in ["python", "java", "c", "kotlin"]}

# Set comprehension
unique_vowels = {ch for ch in "the quick brown fox jumps over the lazy dog" if ch in "aeiou"}
print("Squares:", squares)
print("Word lengths:", word_lengths)
print("Unique vowels:", unique_vowels)
```

```python
#Task 1.3 — Nested Comprehension Challenge

sentences = ["the sky is blue", "python is fun", "list comprehensions are handy"]

# Flat list
long_words = [word for sentence in sentences for word in sentence.split() if len(word) > 2]

# Set of all unique words
unique_words = {word.lower() for sentence in sentences for word in sentence.split()}

# Checkpoint: A nested comprehension becomes less readable when it contains several loops, conditions, or complicated transformations
# For example, three nested loops with multiple conditions would be clearer as ordinary loops with meaningful variable names.
print("Words longer than 2 characters:", long_words)
print("Unique words:", unique_words)
```

# Collections in Practice

```python
#Task 2.1 — Student Grade Book

students = {
    "a": [88, 92, 79],
    "b": [65, 70, 68],
    "c": [95, 91, 89],
}
# Dict comprehension
averages = {
    name: sum(grades) / len(grades) if grades else 0
    for name, grades in students.items()
}
# Set: find students every grade above 80.
above_80 = {
    name
    for name, grades in students.items()
    if grades and all(grade > 80 for grade in grades)
}
# Sort students by average grade in descending order.
sorted_students = sorted(
    averages.items(),
    key=lambda item: item[1],
    reverse=True
)
print("Averages:", averages)
print("Students with all grades above 80:", above_80)
print("Sorted students:", sorted_students)
```

```python
#Task 2.2 — Set Operations: Access Control

team_a = {"figma", "jira", "slack", "github"}
team_b = {"jira", "slack", "notion", "vscode"}
common_tools = team_a & team_b
team_a_only = team_a - team_b
combined_tools = team_a | team_b
symmetric_difference = team_a ^ team_b
print("Common tools:", common_tools)
print("Only Team A:", team_a_only)
print("Combined tools:", combined_tools)
print("Symmetric difference:", symmetric_difference)
```

```python
#Task 2.3 — Mutability Trap
# Output:
# ['apple']
# ['apple', 'banana']
# The default list is created only once.
# Both function calls use the same list.
# Correct solution
def add_item_fixed(item, basket=None):
    if basket is None:
        basket = []
    basket.append(item)
    return basket
print(add_item_fixed("apple"))
print(add_item_fixed("banana"))
```

# File Handling: Text, CSV, JSON

```python
#Task 3.1 — Basic Read/Write
lines = [
    "Application started",
    "User logged in",
    "User opened dashboard",
    "Report generated",
    "Application closed"
]
# Write to file
with open("log.txt", "w") as file:
    for line in lines:
        file.write(line + "\n")

# Read file
with open("log.txt", "r") as file:
    number = 1

    for line in file:
        print(number, line.strip())
        number += 1

# Manual open and close
file = open("manual.txt", "w")
file.write("Hello")
file.close()
# If we forget close(), the file may remain open.
# This can cause a file handle/resource leak.
# Append instead of overwrite
with open("log.txt", "a") as file:
    file.write("New log entry\n")

print("New line added.")
```

```python
#Task 3.2 — CSV: Employee Recordsimport csv
import csv
employees = [
    ["Anu", "Engineering", 85000],
    ["Ritika", "Sales", 60000],
    ["Ardra", "Engineering", 90000],
    ["Keerthi", "HR", 55000],
    ["Lekhya", "Sales", 65000]
]

# Create CSV file
with open("employees.csv", "w", newline="") as file:
    writer = csv.writer(file)

    writer.writerow(["name", "department", "salary"])

    for employee in employees:
        writer.writerow(employee)

# Read CSV
employee_list = []
with open("employees.csv", "r") as file:
    reader = csv.DictReader(file)
    for employee in reader:
        print(employee)
        employee_list.append(employee)

# Total salary by department
department_salary = {}
for employee in employee_list:
    department = employee["department"]
    salary = float(employee["salary"])
    if department in department_salary:
        department_salary[department] += salary
    else:
        department_salary[department] = salary
print("Department salaries:")
print(department_salary)

# Give Engineering employees a 10% raise
for employee in employee_list:
    if employee["department"] == "Engineering":
        salary = float(employee["salary"])
        employee["salary"] = salary * 1.10

# Write updated file
with open("employees_updated.csv", "w", newline="") as file:
    writer = csv.DictWriter(
        file,
        fieldnames=["name", "department", "salary"]
    )
    writer.writeheader()
    for employee in employee_list:
        writer.writerow(employee)
print("Updated file created.")

```

```python
#Task 3.3 — JSON: Config File
import json
config = {
    "app_name": "InventoryApp",
    "version": 1.2,
    "features": ["export", "search", "notifications"],
    "limits": {
        "max_users": 100,
        "max_items": 5000
    }
}

# Write JSON
with open("config.json", "w") as file:
    json.dump(config, file, indent=2)

# Read JSON
with open("config.json", "r") as file:
    data = json.load(file)
print("Maximum users:", data["limits"]["max_users"])

# Add dark mode
data["features"].append("dark_mode")

# Save again
with open("config.json", "w") as file:
    json.dump(data, file, indent=2)
print("Updated features:", data["features"])

# Test invalid JSON
try:
    with open("config.json", "w") as file:
        file.write('{"name": "InventoryApp"')
    with open("config.json", "r") as file:
        json.load(file)
except json.JSONDecodeError:
    print("Invalid JSON file!")
```

# Exceptions + Capstone

```python
#Task 4.1 — try/except/finally Basics

def safe_divide(a, b):
    try:
        return a / b
    except ZeroDivisionError:
        print("Warning: Cannot divide by zero.")
        return None
    except TypeError:
        print("Warning: Invalid data type.")
        return None
    finally:
        print("Operation attempted")
print(safe_divide(10, 2))
print()
print(safe_divide(10, 0))
print()
print(safe_divide("10", 2))
```

```python
#Task 4.3 — Capstone: Robust CSV → JSON Converter
import csv
import json
try:
    totals = {}
    statuses = set()

    # Read CSV
    with open("orders.csv", "r") as file:
        reader = csv.DictReader(file)
        for row_number, row in enumerate(reader, start=2):
            customer = row["customer"]
            try:
                amount = float(row["amount"])
            except (ValueError, TypeError):
                print(
                    "Warning: invalid amount on row",
                    row_number,
                    "for",
                    customer
                )
                continue

            # Add amount to customer total
            if customer in totals:
                totals[customer] += amount
            else:
                totals[customer] = amount

            # Add status to set
            statuses.add(row["status"])


    # Create final result
    summary = {
        "totals_by_customer": totals,
        "statuses": list(statuses)
    }


    # Write JSON
    with open("orders_summary.json", "w") as file:
        json.dump(summary, file, indent=2)

    print("Conversion successful.")
    print(summary)
except FileNotFoundError:
    print("orders.csv was not found.")

except Exception as error:
    print("Something went wrong:", error)

finally:
    print("Conversion attempt finished")
```
