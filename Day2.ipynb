```python
title = "  My First   Python Project!!  "

title = title.strip()
print("After strip:", title)
 
title = title.replace("!!", "")
print("After removing !!:", title)
 
title = title.lower()
print("After lowercase:", title)
 
slug = title.replace(" ", "_")
print("Final slug:", slug)
```
