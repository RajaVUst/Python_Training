```python
# Exercise 12: Compute an average from CSV data
import csv

scores = []

with open("scores.csv", "r") as f:
    reader = csv.DictReader(f)

    for row in reader:
        if row["score"] != "":
            scores.append(int(row["score"]))

average = sum(scores) / len(scores)

print("Scores:", scores)
print("Average:", average)
```

```python
# Exercise 25: Catch your custom exception
class InvalidScoreError(Exception):
    pass


def validate_score(score):
    if score < 0 or score > 100:
        raise InvalidScoreError("Score must be between 0 and 100.")
    return score


test_scores = [95, 105, 78, -10, 100]

for score in test_scores:
    try:
        valid_score = validate_score(score)
        print(f"Valid score: {valid_score}")

    except InvalidScoreError as e:
        print(f"Invalid score ({score}): {e}")
```

```python
# Exercise 26: Safe score loader (function)
import csv

def load_scores(path):
    scores = {}

    try:
        with open(path, "r") as f:
            reader = csv.DictReader(f)

            for row in reader:
                try:
                    score = int(row["score"])
                    scores[row["name"]] = score

                except ValueError:
                    print(f"Skipping invalid score for {row['name']}")

    except FileNotFoundError:
        print(f"File '{path}' not found.")
        return {}

    return scores

scores = load_scores("scores.csv")
print(scores)
```

```python
# Exercise 27: Validate and summarise
def summarise(scores):
    if not scores:
        return {
            "count": 0,
            "average": 0,
            "highest": None,
            "lowest": None
        }

    values = scores.values()

    return {
        "count": len(values),
        "average": sum(values) / len(values),
        "highest": max(values),
        "lowest": min(values)
    }

scores = {
    "Alice": 95,
    "Bob": 88,
    "Charlie": 72
}

summary = summarise(scores)
print(summary)
```

```python
# Exercise 28: Word frequency counter
def word_frequencies(path):
    frequencies = {}

    with open(path, "r") as f:
        for line in f:
            words = line.lower().split()

            for word in words:
                frequencies[word] = frequencies.get(word, 0) + 1

    return frequencies

result = word_frequencies("notes.txt")
print(result)
```

```python
# Exercise 29: JSON config with validation
import json

class MissingConfigKeyError(Exception):
    pass


def load_config(path):
    with open(path, "r") as f:
        data = json.load(f)

    if "batch_size" not in data:
        raise MissingConfigKeyError("Missing key: batch_size")

    if "learning_rate" not in data:
        raise MissingConfigKeyError("Missing key: learning_rate")

    return data
```

```python
import json

config = {
    "batch_size": 32,
    "learning_rate": 0.001
}

with open("config.json", "w") as f:
    json.dump(config, f, indent=2)
try:
    config = load_config("config.json")
    print("Config loaded successfully!")
    print(config)

except MissingConfigKeyError as e:
    print("Configuration Error:", e)

except FileNotFoundError:
    print("Config file not found.")
```

```python
# Exercise 30: Draft a mini log parser
# Step 1: Create access.log
with open("access.log", "w") as f:
    f.write("2026-03-23_10:00 INFO Server started\n")
    f.write("2026-03-23_10:01 INFO User login successful\n")
    f.write("2026-03-23_10:02 ERROR Connection timeout\n")
    f.write("2026-03-23_10:03 WARNING Disk space low\n")
    f.write("MALFORMED LINE\n")  # malformed
    f.write("2026-03-23_10:04 INFO Request processed\n")
    f.write("2026-03-23_10:05 ERROR Database unavailable\n")
    f.write("2026-03-23_10:06 INFO Server stopped\n")

print("access.log created!")
```

```python
# Step 2: Define parse_log(path)
def parse_log(path):
    entries = []
    skipped = 0

    try:
        with open(path, "r") as f:
            for line in f:
                line = line.strip()

                parts = line.split(maxsplit=2)

                if len(parts) != 3:
                    skipped += 1
                    continue

                entry = {
                    "timestamp": parts[0],
                    "level": parts[1],
                    "message": parts[2]
                }

                entries.append(entry)

    except FileNotFoundError:
        print(f"File '{path}' not found.")
        return [], 0

    return entries, skipped
```

```python
# Step 3: Call the Function
logs, skipped = parse_log("access.log")

print("Parsed successfully:", len(logs))
print("Skipped:", skipped)

print("\nParsed entries:")

for log in logs:
    print(log)
```
