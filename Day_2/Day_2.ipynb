```python
title = "  My First   Python Project!!  "

title = title.strip()
print(f"After strip: {title}")

title = title.replace("!!", "")
print(f"After removing !!: {title}")

title = title.lower()
print(f"After lowercase: {title}")

title = title.replace(" ", "_")
print(f"Final slug: {title}")
```

```python
sentence = "Python is a very powerful programming language"

character_count = len(sentence)
word_count = len(sentence.split())

vowel_count = (sentence.count("a")+ sentence.count("e")+ sentence.count("i")+ sentence.count("o")+ sentence.count("u"))

longer_than_30 = len(sentence) > 30

print(f"Character count: {character_count}")
print(f"Word count: {word_count}")
print(f"Vowel count: {vowel_count}")
print(f"Longer than 30 characters: {longer_than_30}")
```

```python
#PART-1------------------String Manipulation Script-------------------
sentence=input("Enter the sentence:")
print(len(sentence))
print(sentence.upper())
print(sentence.lower())
print(sentence[::-1])
print(len(sentence.split(" ")))
words=sentence.split(" ")
print(f"First word: {words[0]},Last word: {words[-1]}")
sentence=sentence.replace(" ","")
sentence=sentence.lower()
print("Old sentence is: ",sentence)
reverse=sentence[::-1]
print("rev sentence is: ",reverse)
print(sentence==reverse)
```
