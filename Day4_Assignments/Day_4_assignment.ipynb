```python
#3.	Task 3: Write a function show_square(number) that prints the square of the given number. Call it with three different numbers, including a negative one.
def show_square(number):
    print(f"Square of {number} is {number**2}")
show_square(12)
show_square(-6)
show_square(3)
```

```python
#Part B: Return Values vs. print()  
#4.	Task 4: Write a function multiply(a, b) that RETURNS the product of a and b (do not print inside the function). Store the result in a variable and print that variable.
def multiply(a,b):
    return a*b
a,b = map(int,input("Enter a and b :").split())
res= multiply(a,b)
print(f"Multiplication value is {res} ")
```

```python
#5.	Task 5: Write two versions of a function that checks if a number is even: is_even_print(n) which prints "Even"/"Odd", and is_even_return(n) which returns True/False. 
def is_even_print(n):
    if n%2==0:
        print(f"{n} is even")
    else:
        print(f"{n} is odd")
def is_even_return(n):
     if n%2==0:
         return True
     else:
         return False
n = int(input("Enter n:"))
res= is_even_return(n)
print(res)
is_even_print(n)
```

```python
#6.	Task 6: Write a function celsius_to_fahrenheit(celsius) that returns the Fahrenheit equivalent using the formula F = C * 9/5 + 32. Use it to convert 0, 37, and 100 degrees Celsius, printing each result.
def celsius_to_fahrenheit(celsius):
    f = celsius*(9/5) + 32
    print(f"{f:.2f}")
 
celsius_to_fahrenheit(0)
celsius_to_fahrenheit(37)
celsius_to_fahrenheit(100)
```
