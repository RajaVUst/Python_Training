```python
#15.	Task 15: Write a recursive function countdown(n) that prints n, n-1, n-2, ... down to 1, and then prints "Liftoff!". Make sure it has a correct base case.
def countdown(n):
    if n==0:
        print("Liftoff!")
        return
    print(n)
    countdown(n-1)
n = int(input("Enter n value:"))
countdown(n)
```

```python
#16.	Task 16: Write a recursive function sum_upto(n) that returns the sum of all integers from 1 to n (e.g. sum_upto(5) returns 15). Identify the base case and the recursive case in a comment above your code.
def sum_upto(n):
    if n==0:
        return 0
    return n+sum_upto(n-1)
print(sum_upto(5))
    
    
```

```python
#Part E: Cumulative Challenges 
#17.	Task 17: Write a function is_palindrome(word) that returns True if a string reads the same forwards and backwards, and False otherwise. Use string slicing (from Day 2) inside the function. Test it with "level", "python", and "Madam" (think about whether case should matter, and handle it).
def is_palindrome(word):
    if word.lower() == word.lower()[::-1]:
        return True
    return False
print(is_palindrome("level"))
print(is_palindrome("python"))
print(is_palindrome("Madam"))
```

```python
#18.	Task 18: Write a function grade_from_score(score) that uses if/elif/else (from Day 3) to RETURN a letter grade: 90+ -> 'A', 80-89 -> 'B', 70-79 -> 'C', 60-69 -> 'D', below 60 -> 'F'. Call it in a loop for the scores [95, 82, 61, 40, 100] and print each result using an f-string.
def grade_from_score(score):
    if score >= 90:
        return 'A'
    elif score >= 80:
        return 'B'
    elif score >= 70:
        return 'C'
    elif score >= 60:
        return 'D'
    else:
        return 'F'
scores = [95, 82, 61, 40, 100]
for score in scores:
    print(f"{score} -> {grade_from_score(score)}")
```

```python
#19.	Task 19: Write a function count_vowels(text) that loops through a string (Day 2 + Day 3 for-loop) and returns how many vowels (a, e, i, o, u — case-insensitive) it contains. Test it on "Python Bootcamp" and on a sentence of your choice.
def count_vowels(text):
    text=text.lower()
    cnt=0
    for i in text:
        if i in "aeiou":
            cnt+=1
    return cnt
print(count_vowels("Python Bootcamp"))
print(count_vowels("Vinayasri Sakinala"))
```

```python
#20.	Task 20: Write a function fizzbuzz_range(start, end) that loops from start to end (inclusive) using range() and, for each number, prints "Fizz" if divisible by 3, "Buzz" if divisible by 5, "FizzBuzz" if divisible by both, and the number otherwise. This reuses your Day 3 FizzBuzz logic but now wrapped in a reusable function with parameters.
def fizzbuzz_range(start, end):
    for num in range(start, end + 1): 
        if num % 3 == 0 and num % 5 == 0:
            print("FizzBuzz")
        elif num % 3 == 0:
            print("Fizz")
        elif num % 5 == 0:
            print("Buzz")
        else:
            print(num)
fizzbuzz_range(2,15)
```

```python
#21.	Task 21: Write a function is_prime(n) that returns True if n is a prime number and False otherwise, using a for loop and the % operator. Then write a second function primes_up_to(limit) that uses is_prime() inside a loop to return a list of all prime numbers up to limit. Test primes_up_to(30).
def is_prime(n):
    if n<2:
        return False
    for i in range(2,n):
        if n%i==0:
            return False
    return True
def primes_upto(limit):
    for i in range(2,limit+1):
        if is_prime(i):
            print(i,end=" ")
primes_upto(30)
```

```python
#22.	Task 22: Write a function password_strength(password) that returns "Weak", "Medium", or "Strong" based on simple rules you define yourself (e.g. length, presence of digits, presence of uppercase letters) using string methods from Day 2 (such as .isdigit(), .isupper()) combined with conditionals. Test it on at least three different passwords.
def password_strength(password):
    has_upper = False
    has_digit = False
    for ch in password:
        if ch.isupper():
            has_upper = True
        if ch.isdigit():
            has_digit = True
    if len(password) < 6:
        return "Weak"
    elif len(password) >= 8 and has_upper and has_digit:
        return "Strong"
    else:
        return "Medium"
print(password_strength("Hello"))
print(password_strength("Hello123"))
print(password_strength("password"))
print(password_strength("Pass123"))
```

```python
#23.	Task 23: Capstone: Write a function build_utilities(operation, *values) that behaves like a mini calculator: if operation is "sum", return the total of all values; if "max", return the largest value; if "min", return the smallest value; if "average", return the mean. Use *args to accept any number of values, and if/elif/else to decide what to do. Test it with build_utilities("sum", 4, 8, 15) and build_utilities("average", 2, 4, 6, 8).
def build_utilities(operation,*values):
    if operation =="sum":
        return sum(values)
    elif operation=="max":
        return max(values)
    elif operation=="min":
        return min(values)
    elif operation=="average":
        return ((sum(values))/len(values))
print(build_utilities("sum",4,8,15))
print(build_utilities("max",2,3,4))
print(build_utilities("min",2,3,4))
print(build_utilities("average",2,4,6,8))
```
