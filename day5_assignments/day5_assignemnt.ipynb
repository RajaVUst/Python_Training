```python
# Exercise 7: Multiple Return Values
# Write a function stats(numbers) that returns a tuple of (minimum, maximum, average) for a list of numbers.
# Call it on [4, 9, 1, 7, 15] and unpack the result into three separate variables.

def stats(numbers):
    return (min(numbers), max(numbers), sum(numbers)/len(numbers))

stats([4, 9, 1, 7, 15])
```

```python
# Exercise 17: Dictionary From Two Lists
# Given names = ["Amit", "Reni", "Tara"] and scores = [78, 91, 65], use a dict comprehension with zip() to build a dictionary 
# mapping each name to their score.
# Hint: zip(names, scores) pairs up items positionally: {name: score for name, score in zip(names, scores)}

names = ["Amit", "Reni", "Tara"]
scores = [78, 91, 65]

dictionary = {name: score for name, score in zip(names, scores)}
print(dictionary)
```

```python
# Exercise 18: Filtering With a Dict Comprehension
# Reusing the name-to-score dictionary from the previous exercise, use a dict comprehension to build a second dictionary 
# containing only entries where the score is 70 or above.
# Hint: {name: score for name, score in scores_dict.items() if score >= 70}

names = ["Amit", "Reni", "Tara"]
scores = [78, 91, 65]

scores_dict = {name: score for name, score in zip(names, scores)}
dictionary = {name: score for name, score in scores_dict.items() if score >= 70}
print(dictionary)

```

```python
# Part F — Cumulative Challenges
# Exercise 19: Vowel Counter by Word
# Write a function vowel_counts(sentence) that splits a sentence into words and returns a dictionary mapping each word to its number of vowels (a, e, i, o, u — case-insensitive). Test it on "The Quick Brown Fox Jumps".
# Hint: Combine string methods (.lower(), .split()) from Day 2 with a dictionary comprehension or loop, inside a function.

def vowel_counts(sentence):
    sentence = sentence.lower().split()

    res_dict = { word: sum(1 for ch in word if ch in 'aeiou') for word in sentence}
    return res_dict

vowel_counts("Varshith is a happy")
```

```python
# Exercise 20: Grade Book Summary
# Write a function grade_summary(roster) that takes a list of dictionaries (each with name and scores, like Exercise 11) and 
# returns a new list of dictionaries, each with name, average, and a letter grade computed using if/elif/else. Print the result for a roster 
# of at least 3 students.
# Hint: Build a new list with a for-loop; inside the loop use conditionals from Day 3 to decide the letter grade, and append a new dict to the result list for each student.

def grade_summary(roster):
    res_list = []
    for dict1 in roster:
        res_dict={}
        res_dict['name'] = dict1['name']
        res_dict['avg_score'] = sum(dict1['scores'])/len(dict1['scores'])
        
        if res_dict['avg_score'] > 90:
            res_dict['grade'] = 'A'
        elif res_dict['avg_score'] > 80:
            res_dict['grade'] = 'B'
        elif res_dict['avg_score'] > 70:
            res_dict['grade'] = 'C'
        elif res_dict['avg_score'] > 60:
            res_dict['grade'] = 'D'
        else:
            res_dict['grade'] = 'F'

        res_list.append(res_dict)

    return res_list


grade_summary([ {'name': 'Varshith', 'scores': [1,2,3] }, {'name': 'Eren', 'scores': [11,22,33]}, {'name': 'Yeager', 'scores': [21,12,43]}])
```

```python
# Exercise 21: Unique Word Finder
# Write a function unique_words(text) that takes a string, splits it into words, lowercases them, and returns a sorted list of 
# the unique words (no duplicates) using a set internally. Test it on a sentence of your choice that has at least one repeated word.
# Hint: Build a set from the lowercased words, then wrap sorted() around it to return a sorted list.

def unique_words(text):
    text = text.lower().split()
    text = sorted(set(text))

    return text

unique_words("Varshith is happy man, he is good")
```
