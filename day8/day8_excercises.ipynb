# Collections Fundamentals

```python
# Task 1.1 — Inventory Tracker
items = ["pen", "notebook", "pen", "eraser", "notebook", "pen", "stapler"]

# List is ordered and allows duplicates.
item_list = items

# Tuple is immutable, so it is suitable for first batch.
first_batch = tuple(items[:3])

# Set stores only unique item names.
unique_items = set(items)

# Dictionary stores each item as a key and its number of occurrences as the value.
item_counts = {}
for i in items:
    item_counts[i] = item_counts.get(i, 0) + 1

print("List:", item_list)
print("First batch tuple:", first_batch)
print("Unique items:", unique_items)
print("Item counts:", item_counts)
```

```python
#Task 1.2 — Comprehensions Rewrite

# List comprehension
squares = [n * n for n in range(1, 21) if n % 2 == 0]

# Dict comprehension
word_lengths = {word: len(word) for word in ["python", "java", "c", "kotlin"]}

# Set comprehension
unique_vowels = {ch for ch in "the quick brown fox jumps over the lazy dog" if ch in "aeiou"}

print("Squares:", squares)
print("Word lengths:", word_lengths)
print("Unique vowels:", unique_vowels)
```

```python
#Task 1.3 — Nested Comprehension Challenge

sentences = ["the sky is blue", "python is fun", "list comprehensions are handy"]

# Flat list
long_words = [word for sentence in sentences for word in sentence.split() if len(word) > 2]

# Set of all unique words
unique_words = {word.lower() for sentence in sentences for word in sentence.split()}

# Checkpoint: In my opinion, nested comprehensions become less readable when there are
# multiple conditions or complex logic. For example, if we need several if/else conditions
# or multiple calculations inside the comprehension, a normal loop is easier to understand.

print("Words longer than 2 characters:", long_words)
print("Unique words:", unique_words)
```

# Collections in Practice

```python
#Task 2.1 — Student Grade Book

students = {
    "a": [88, 92, 79],
    "b": [65, 70, 68],
    "c": [95, 91, 89],
}

# Dict comprehension
averages = {
    name: sum(grades) / len(grades) if grades else 0
    for name, grades in students.items()
}

# Set: find students every grade above 80.
above_80 = {
    name
    for name, grades in students.items()
    if grades and all(grade > 80 for grade in grades)
}

# Sort students by average grade in descending order.
sorted_students = sorted(
    averages.items(),
    key=lambda item: item[1],
    reverse=True
)

print("Averages:", averages)
print("Students with all grades above 80:", above_80)
print("Sorted students:", sorted_students)
```

```python
#Task 2.2 — Set Operations: Access Control

team_a = {"figma", "jira", "slack", "github"}
team_b = {"jira", "slack", "notion", "vscode"}

common_tools = team_a & team_b
team_a_only = team_a - team_b
combined_tools = team_a | team_b
symmetric_difference = team_a ^ team_b

print("Common tools:", common_tools)
print("Only Team A:", team_a_only)
print("Combined tools:", combined_tools)
print("Symmetric difference:", symmetric_difference)
```

# File Handling: Text, CSV, JSON

```python
#Task 3.1 — Basic Read/Write

# 1. Write 5 log lines
log_lines = [
    "INFO: Application started",
    "INFO: User logged in",
    "INFO: Data loaded",
    "WARNING: Low disk space",
    "INFO: Application closed"
]

with open("log.txt", "w") as file:
    for line in log_lines:
        file.write(line + "\n")


# 2. Read the file line by line and print line numbers
with open("log.txt", "r") as file:
    for number, line in enumerate(file, start=1):
        print(number, line.strip())


# 3. Manual open() and close()
# If we forget to call close(), the file handle may remain open.
# This can cause a file handle leak and may leave buffered data unflushed.
file = open("log.txt", "r")
for line in file:
    print(line.strip())

# file.close()  # Deliberately forgotten!


# 4. Correct version using with again
with open("log.txt", "r") as file:
    for line in file:
        print(line.strip())


# 5. Append new log lines instead of overwriting the file
with open("log.txt", "a") as file:
    file.write("INFO: New session started\n")
    file.write("INFO: New data added\n")
```

```python
#Task 3.2 — CSV: Employee Records

# 1. Create employees.csv with 5 employee records
import csv
employees = [
    ["Alice", "Engineering", 60000],
    ["Bob", "HR", 50000],
    ["Carol", "Engineering", 70000],
    ["David", "Sales", 55000],
    ["Eve", "HR", 52000]
]
with open("employees.csv", "w", newline="") as file:
    writer = csv.writer(file)
    writer.writerow(["name", "department", "salary"])
    writer.writerows(employees)


# 2. Read the CSV using DictReader

employee_records = []

with open("employees.csv", "r", newline="") as file:
    reader = csv.DictReader(file)

    for employee in reader:
        print(employee)
        employee_records.append(employee)


# 3. Compute total salary per department

department_salaries = {}

for employee in employee_records:
    department = employee["department"]
    salary = float(employee["salary"])

    department_salaries[department] = (
        department_salaries.get(department, 0) + salary
    )

print("\nTotal salary per department:")
print(department_salaries)


# 4. Add a 10% raise to Engineering employees and write the results to employees_updated.csv

with open("employees_updated.csv", "w", newline="") as file:
    fieldnames = ["name", "department", "salary"]

    writer = csv.DictWriter(file, fieldnames=fieldnames)
    writer.writeheader()

    for employee in employee_records:
        salary = float(employee["salary"])

        if employee["department"] == "Engineering":
            salary = salary * 1.10

        writer.writerow({
            "name": employee["name"],
            "department": employee["department"],
            "salary": salary
        })
```

# Exceptions + Capstone

```python
#Task 4.1 — try/except/finally Basics

def safe_divide(a, b):
    try:
        return a / b

    except ZeroDivisionError:
        print("Warning: Cannot divide by zero.")
        return None

    except TypeError:
        print("Warning: Invalid data type.")
        return None

    finally:
        print("Operation attempted")


print(safe_divide(10, 2))
print()
print(safe_divide(10, 0))
print()
print(safe_divide("10", 2))
```

### Task 4.2 — Exception Handling: Concept Check
1. A specific exception catches a particular type of error. For example, ZeroDivisionError catches an error caused by dividing by zero. A bare except: catches almost all errors. It is usually not recommended because it can hide the actual problem and make debugging harder. It is better to catch the specific errors we expect.

2. The finally block is used for something that should always happen, whether there is an error or not. For example, if we open a file manually, we can use finally to make sure the file is closed even if an error happens while reading it.

3. Different file errors tell us what went wrong. FileNotFoundError means the file does not exist, PermissionError means we do not have permission to access it, and IsADirectoryError means we tried to use a directory as a file. Catching them separately helps us understand the problem and give the correct response instead of just saying that some error occurred.



```python
#Task 4.3 — Capstone: Robust CSV → JSON Converter

import csv
import json

try:
    customer_totals = {}
    statuses = set()

    with open("orders.csv", "r", newline="") as file:
        reader = csv.DictReader(file)

        for row in reader:
            try:
                amount = float(row["amount"])
            except (ValueError, TypeError):
                print(
                    f"Warning: Invalid amount for row {row['order_id']} "
                    f"(customer: {row['customer']}). Skipping row."
                )
                continue

            customer = row["customer"]
            status = row["status"]

            customer_totals[customer] = customer_totals.get(customer, 0) + amount
            statuses.add(status)

    summary = {
        "customer_totals": customer_totals,
        "statuses": list(statuses)
    }

    with open("orders_summary.json", "w") as file:
        json.dump(summary, file, indent=2)

    print("Summary written to orders_summary.json")

except FileNotFoundError:
    print("Error: orders.csv was not found.")

except Exception as e:
    print("An unexpected error occurred:", e)

finally:
    print("Conversion attempt finished")
```
