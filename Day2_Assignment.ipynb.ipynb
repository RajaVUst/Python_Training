```python
# PART 1 - STRING MANIPULATION

sentence = input("Enter a sentence: ")

character_count = len(sentence)
uppercase_sentence = sentence.upper()
lowercase_sentence = sentence.lower()
reversed_sentence = sentence[::-1]

words = sentence.split()
word_count = len(words)

first_word = words[0]
last_word = words[-1]

cleaned_sentence = sentence.replace(" ", "").lower()
is_palindrome = cleaned_sentence == cleaned_sentence[::-1]

print()
print("========== STRING MANIPULATION ==========")

print("Character Count :", character_count)
print("Uppercase       :", uppercase_sentence)
print("Lowercase       :", lowercase_sentence)
print("Reversed        :", reversed_sentence)
print("Word Count      :", word_count)
print("First Word      :", first_word)
print("Last Word       :", last_word)
print("Is Palindrome?  :", is_palindrome)
```

```python
# PART 2 - SIMPLE CALCULATOR

number1 = float(input("Enter first number: "))
number2 = float(input("Enter second number: "))

addition = number1 + number2
subtraction = number1 - number2
multiplication = number1 * number2
division = number1 / number2
floor_division = number1 // number2
modulus = number1 % number2
power = number1 ** number2

print()
print("============== CALCULATOR ==============")

print(f"{'Operation':<15}{'Operator':^8}{'Result':>12}")
print("-" * 35)

print(f"{'Addition':<15}{'+':^8}{addition:>12.2f}")
print(f"{'Subtraction':<15}{'-':^8}{subtraction:>12.2f}")
print(f"{'Multiplication':<15}{'*':^8}{multiplication:>12.2f}")
print(f"{'Division':<15}{'/':^8}{division:>12.2f}")
print(f"{'Floor Division':<15}{'//':^8}{floor_division:>12.2f}")
print(f"{'Modulus':<15}{'%':^8}{modulus:>12.2f}")
print(f"{'Power':<15}{'**':^8}{power:>12.2f}")
```
