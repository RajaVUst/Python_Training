  Exercise 5  ·  Slug Normalizer  

```python
title = "  My First   Python Project!!  "
print(title)
title = title.strip()
print(title)
title = title.lower()
print(title)
title = title.replace("!!","")
print(title)
title = title.replace(" ","_")
print(title)
```

  Exercise 6  ·  Sentence Dashboard  

```python
sentence = input("Enter a sentence: ").strip().lower()
formatted_sentence = sentence.split(" ")

total_word_count = len(formatted_sentence)
formatted_sentence = sentence.replace(" ", "")
total_char_count = len(formatted_sentence)
is_char_count = total_char_count>30
total_vowel_count = formatted_sentence.count('a') + formatted_sentence.count('e') + formatted_sentence.count('i') + formatted_sentence.count('o') + formatted_sentence.count('u')

print(f"Total word count: {total_word_count}")
print(f"Total character count: {total_char_count}")
print(f"Total vowel count: {total_vowel_count}")
```
