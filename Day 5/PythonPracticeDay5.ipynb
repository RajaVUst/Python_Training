```python
# Task 7
def stats(numbers):
    low = min(numbers)
    high = max(numbers)
    avg = sum(numbers) / len(numbers)
    return low, high, avg
low, high, avg = stats([4, 9, 1, 7, 15])
print(low, high, avg)
```

```python
# Part-F
# Task 19
def vowel_counts(sentence):
    result = {}
    for word in sentence.split():
        result[word] = sum(ch in "aeiou" for ch in word.lower())
    return result
print(vowel_counts("The Quick Brown Fox Jumps"))
```

```python
# Task 20
def grade_summary(roster):
    result = []
    for student in roster:
        avg = sum(student["scores"]) / len(student["scores"])
        if avg >= 90:
            grade = "A"
        elif avg >= 80:
            grade = "B"
        elif avg >= 70:
            grade = "C"
        elif avg >= 60:
            grade = "D"
        else:
            grade = "F"
        result.append({
            "name": student["name"],
            "average": round(avg, 1),
            "grade": grade
        })
    return result
roster = [
    {"name": "Amit", "scores": [85, 90, 88]},
    {"name": "Reni", "scores": [92, 89, 95]},
    {"name": "Tara", "scores": [65, 70, 68]}
]
print(grade_summary(roster))
```

```python
# Task 21
def unique_words(text):
    words = set(text.lower().split())
    return sorted(words)
print(unique_words("Python is easy and Python is powerful"))
```

```python
# Task 22
def fizzbuzz_map(n):
    result = {}
    for i in range(1, n + 1):
        if i % 15 == 0:
            result[i] = "FizzBuzz"
        elif i % 3 == 0:
            result[i] = "Fizz"
        elif i % 5 == 0:
            result[i] = "Buzz"
        else:
            result[i] = str(i)
    return result
print(fizzbuzz_map(20))
```

```python
# Task 23
def password_report(passwords):
    result = {
        "weak": [],
        "medium": [],
        "strong": []
    }
    for pw in passwords:
        has_digit = any(ch.isdigit() for ch in pw)
        if len(pw) >= 10 and has_digit:
            result["strong"].append(pw)
        elif len(pw) >= 6:
            result["medium"].append(pw)
        else:
            result["weak"].append(pw)
    return result
passwords = [
    "abc",
    "hello1",
    "python",
    "welcome",
    "Secure1234",
    "myPassword99"
]
print(password_report(passwords))
```
