```python
# Clean it up
sentence = "   Python IS Fun to Learn!!   "
clean = sentence.strip()
print(clean)
```

```python
#Print a lowercase and an uppercase version
print(clean.lower())
print(clean.upper())
```

```python
# indexing and slicing 
print(clean[0])
print(clean[-1])
print(clean[:5])
```

```python
# Split clean into a list of words
words = clean.split()
print(words)
print(len(words))
```

```python
# replace() 
print(clean.replace("Fun", "Powerful"))
```

```python
# join()
print("-".join(words))
```

```python
# word count and the reversed sentence
print(f"Words: {len(words)} | Reversed: {clean[::-1]}")
```

```python
# Part C — Mini-Exercises
# Exercise 1  ·  Name Tag Formatter  
first = input('Enter your First Name : ')
last = input ('Enter your Last Name : ')
print(f'Hello, {first.title()} {last.title()}!')
```

```python
# Exercise 5  ·  Slug Normalizer  
title = "  My First   Python Project!!  "
clean = title.replace('!',' ')
clean = clean.strip()
clean = clean.replace(' ','_')
print(clean.lower())
```

```python
# Exercise 6  ·  Sentence Dashboard  
text = input('Enter Your Text Here : ')
chars = text.replace(' ','')
words = text.split(' ')
text = text.lower()
long = len(chars)>30
vowels = text.count('a') + text.count('e') + text.count('i') + text.count('o') + text.count('u')
print(f'Character Count : {len(chars)}\nWord Count : {len(words)}\nVowel Count : {vowels}\nLonger than 30 Character : {long}')
```
