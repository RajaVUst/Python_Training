```python
# Part 1 — String Manipulation Script
text = input('Enter the Text here : ')
chars = text.replace(' ','')
words = text.split(' ')
print(f'Character Count : {len(chars)}\nUppercase : {text.upper()}\nLowercase : {text.lower()}\nReversed : {text[::-1]}\nWord Count : {len(words)}\nFirst Word : {words[0]}\nLast Word : {words[len(words)-1]}\nPalindrome : {chars[::-1].lower()==chars.lower()}')
```

```python
# Part 2 — Simple Calculator, Revisited
num1 = input('Enter the First Number : ')
num2 = input('Enter the Second Number : ')
num1,num2 = float(num1),float(num2)
print(f"{'Label':<18}{'Operator':^8}{'Result':>10}")
print(f"{'Addition':<18}{'+':^8}{num1+num2:>10.2f}")
print(f"{'Subraction':<18}{'-':^8}{num1-num2:>10.2f}")
print(f"{'Mulitiplication':<18}{'*':^8}{num1*num2:>10.2f}")
print(f"{'Division':<18}{'/':^8}{num1/num2:>10.2f}")
print(f"{'Floor Division':<18}{'//':^8}{num1//num2:>10.2f}")
print(f"{'Modulus':<18}{'%':^8}{num1%num2:>10.2f}")
print(f"{'Power':<18}{'**':^8}{num1**num2:>10.2f}")
```
