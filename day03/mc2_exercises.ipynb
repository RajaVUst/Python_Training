```python
# Exercise 11: Sum of first N natural numbers 
n = int(input("Enter a number : "))
total = 0
for i in range(1,n+1):
    total += i
print(total)
```

```python
# Exercise 12: Sum of digits 
n = int(input("Enter a number : "))
total = 0
while n>0:
    digit = n % 10
    total = total + digit
    n = n // 10
print(f"Sum of digits: {total}")
```

```python
# Exercise 13: Countdown with a while loop 
n = int(input("Enter a number : "))
while n>0:
    print(n)
    n = n -1
print("Liftoff!")
```

```python
# Exercise 14: Right-angled triangle pattern 
n = int(input("Enter number of rows: "))
for i in range(1, n + 1):
    for j in range(i):
        print("*", end="")
    print()
```

```python
# Exercise 15: Vowel counter 
text = input("Enter a word or sentence: ")
count = 0
for i in text:
    if i=="a" or i=="e" or i=="i" or i=="o" or i=="u":
        count +=1
print(f"Number of vowels: {count}")
```

# Part E: Combined Logic — Classic Problems 


```python
# Exercise 16: FizzBuzz 
for i in range(1,51):
    if i % 15 == 0:
        print("FizzBuzz")
    elif i % 3 == 0:
        print("Fizz")
    elif i % 5 == 0:
        print("Buzz")
    else:
        print(i)
```

```python
# Exercise 17: Reverse an integer 
n = int(input("Enter a positive whole number: "))
reversed_n = 0
while n > 0:
    digit = n % 10
    reversed_n = reversed_n * 10 + digit
    n = n // 10
print(f"Reversed number: {reversed_n}")
```

```python
# Exercise 18: Palindrome check for an integer 
n = int(input("Enter a positive whole number: "))
original = n
reversed_n = 0
while n > 0:
    digit = n % 10
    reversed_n = reversed_n * 10 + digit
    n = n // 10
if original == reversed_n:
    print("is a palindrome")
else:
    print("Not a palindrome")
```

```python
# Exercise 19: Prime number check 
n = int(input("Enter a positive whole number: "))
is_prime = True
for i in range(2, n):
    if n % i == 0:
        is_prime = False
        break
if is_prime:
    print(f"{n} is a prime number")
else:
    print(f"{n} is not a prime number")
```

```python
# Exercise 20: Print primes in a range 
for n in range(2, 51):
    is_prime = True
    for i in range(2, n):
        if n % i == 0:
            is_prime = False
            break
    if is_prime:
        print(n)
```

```python
# Exercise 21: Factorial of a number 
n = int(input("Enter a whole number: "))
factorial = 1
for i in range(1, n + 1):
    factorial = factorial * i
print(f"{n}! = {factorial}")
```

```python
# Exercise 22: Fibonacci series 
a = 0
b = 1
for i in range(15):
    print(a)
    next_num = a + b
    a = b
    b = next_num
```
