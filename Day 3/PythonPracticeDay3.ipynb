```python
# Exercise 12: Sum of digits
num  = int(input('Enter a Number : '))
tot = 0
rem = 0
while num > 0:
    rem = num % 10
    tot = tot + rem
    num = num // 10
print(tot)
```

```python
# Exercise 17: Reverse an integer
n = int(input('Enter a number : '))
rev = 0
while n > 0:
    rev = rev * 10 + (n % 10)
    n = n // 10
print(f'Reversed number : {rev}') 
```

```python
# Exercise 18: Palindrome check for an integer
n = int(input('Enter a positive whole number: '))
temp = n
rev = 0
while n > 0:
    rev = rev * 10 + (n % 10)
    n = n // 10
if temp == rev:
    print('Palindrome')
else:
    print('Not a palindrome')
```

```python
# Exercise 19: Prime number check
n = int(input('Enter a number : '))
is_prime = True
for i in range(2, n):
    if n % i == 0:
        is_prime = False
        break
if is_prime:
    print('Prime Number')
else:
    print('Not Prime Number')
```

```python
# Exercise 20: Print primes in a range
for n in range(2, 51):
    is_prime = True
    for i in range(2, n):
        if n % i == 0:
            is_prime = False
            break
    if is_prime:
        print(n)
```

```python
# Exercise 23: Simple currency converter

amt = float(input('Enter amount : '))
cur = input('Enter target currency (USD, EUR, GBP): ').upper()

if cur == 'USD':
    conv = amt / 90
    print('Amount in USD:', round(conv, 2))

elif cur == 'EUR':
    conv = amt / 105
    print('Amount in EUR:', round(conv, 2))

elif cur == 'GBP':
    conv = amt / 120
    print('Amount in GBP:', round(conv, 2))

else:
    print('Invalid currency')
```

```python
# Stretch Exercises (Optional)
# Exercise 24: Number guessing game (fixed target)
num = 25
opt = [10, 30, 20, 25, 40]
for guess in opt:
    if guess > num:
        print(guess, '- Too high')
    elif guess < num:
        print(guess, '- Too low')
    else:
        print(guess, '- Correct!')
        break
```

```python
# Exercise 25: Pyramid of numbers
n = int(input('Enter number of rows: '))
for row in range(1, n + 1):
    for num in range(1, row + 1):
        print(num, end=' ')
    print()
```
