```python
# exercise 17 - Reverse an integer
```

```python
n = int(input("Enter a positive whole number: "))

reversed_n = 0

while n > 0:
    digit = n % 10
    reversed_n = reversed_n * 10 + digit
    n = n // 10

print(f"Reversed number: {reversed_n}")
```

```python
# exercise 18 - Palindrome check for an integer
```

```python
n = int(input("Enter a positive whole number: "))

original = n
reversed_n = 0

while n > 0:
    digit = n % 10
    reversed_n = reversed_n * 10 + digit
    n = n // 10

if original == reversed_n:
    print("Palindrome")
else:
    print("Not a palindrome")
```

```python
# exercise 19 - Prime number check
```

```python
n = int(input("Enter a number greater than 1: "))

is_prime = True

for i in range(2, n):
    if n % i == 0:
        is_prime = False
        break

if is_prime:
    print("Prime")
else:
    print("Not prime")
```

```python
# exercise 20 - Print primes in a range
```

```python
for n in range(2, 51):
    is_prime = True

    for i in range(2, n):
        if n % i == 0:
            is_prime = False
            break

    if is_prime:
        print(n)
```

```python
# exercise 23 - Simple currency converter
```

```python
amount = float(input("Enter amount in INR: "))
currency = input("Enter target currency (USD/EUR/GBP): ").upper()

if currency == "USD":
    converted = amount * 0.0119
    print(f"Converted amount: {round(converted, 2)} USD")

elif currency == "EUR":
    converted = amount * 0.0101
    print(f"Converted amount: {round(converted, 2)} EUR")

elif currency == "GBP":
    converted = amount * 0.0088
    print(f"Converted amount: {round(converted, 2)} GBP")

else:
    print("Invalid currency")
```

```python
# exercise 24 - Number guessing game
```

```python
secret_number = 25

guesses = [10, 30, 15, 25, 40]

for guess in guesses:
    if guess > secret_number:
        print(f"{guess}: Too high")
    elif guess < secret_number:
        print(f"{guess}: Too low")
    else:
        print(f"{guess}: Correct!")
        break
```

```python
# exercise 25 - Pyramid of numbers
```

```python
n = int(input("Enter the number of rows: "))

for row in range(1, n + 1):
    for number in range(1, row + 1):
        print(number, end=" ")
    print()
```
