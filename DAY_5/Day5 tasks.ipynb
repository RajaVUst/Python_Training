```python
##exercise 19

def in_loops():
    sentence = "The Quick Brown Fox Jumps"
    lower = sentence.lower()
    words = lower.split()
    print(words)
    vowels = ['a','e','i','o','u']
    result = {}
    for word in words:
        count = 0
        for ch in word:
            if ch in vowels:
                count += 1
        result[word] = count
    print("With loops : ",result)

def in_comprehension():
    sentence = "The Quick Brown Fox Jumps"
    lower = sentence.lower()
    words = lower.split()
    vowels = ['a','e','i','o','u']
    dic = {word : sum(1 for ch in word if ch in vowels )for word in words}
    print("With comprehension : ",dic)
in_loops()
in_comprehension()
```

```python
#exercise 20
def grade_summary(roster):
    result=[]
    for student in roster:
        average=sum(student["scores"])/len(student["scores"])
        if average>=90:
            grade="A"
        elif average>=80:
            grade="B"
        elif average>=70:
            grade="C"
        elif average>=60:
            grade="D"
        else:
            grade="F"
        result.append({"name":student['name'],
        "average:":f"{average:.2f}",
        "Grade":grade})
    return result            
 
 
roster=[
    {'name':"pavan",'scores':[56,76,98]},
    {'name':"prem",'scores':[87,98,65]},
    {'name':"ram",'scores':[98,65,43]},
]        
 
print(grade_summary(roster))
```

```python
#21
def unique_words(text):
    senetence=text.lower().split()
    result=set(senetence)
    return sorted(result)
text=input("Enter the text:")
print(unique_words(text))    
```

```python
def fizzbuzz_map(n):
    result={}
    for i in range(1, n+1):
        if i%3==0 and i%5==0:
            result[i]="FizzBuzz"
        elif i%3==0:
            result[i]="Fizz"
        elif i%5==0:
            result[i]="Buzz"
        else:
            result[i]=str(i)
    return result                
 
n=int(input("Entehr the number"))    
print(fizzbuzz_map(n))
```
