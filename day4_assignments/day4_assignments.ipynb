```python
# 15.	Task 15: Write a recursive function countdown(n) that prints n, n-1, n-2, ... down to 1, and 
# then prints "Liftoff!". Make sure it has a correct base case.

def countdown(n):
    if n == 0:
        return
    print(n)
    countdown(n-1)

countdown(5)
```

```python
# 16.	Task 16: Write a recursive function sum_upto(n) that returns the sum of all integers from 1 to n 
# (e.g. sum_upto(5) returns 15). Identify the base case and the recursive case in a comment above your code.
# Expected: sum_upto(5) -> 15, sum_upto(1) -> 1

def sum_upto(n):
    if n == 0:
        return 0
    return n + sum_upto(n-1)

sum_upto(5)
```

```python
# Part E: Cumulative Challenges 
# Task 17: Write a function is_palindrome(word) that returns True if a string reads the same forwards and backwards, 
# and False otherwise. Use string slicing (from Day 2) inside the function. Test it with "level", "python", 
# and "Madam" (think about whether case should matter, and handle it).

def is_palindrome(word):
    word = word.lower()
    print(word == word[::-1])

is_palindrome("level")
is_palindrome("python")
is_palindrome("Madam")
```

```python
# Task 18: Write a function grade_from_score(score) that uses if/elif/else (from Day 3) to RETURN a letter grade:
# 90+ -> 'A', 80-89 -> 'B', 70-79 -> 'C', 60-69 -> 'D', below 60 -> 'F'. Call it in a loop for the scores [95, 82, 61, 40, 100] and print each result using an f-string.
# Expected: 95 -> A, 82 -> B, 61 -> D, 40 -> F, 100 -> A

def grade_from_score(score):
    if score>=90:
        return 'A'
    elif score>=80 and score<=89:
        return 'B'
    elif score>=70 and score<=79:
        return 'C'
    elif score>=60 and score<=69:
        return 'D'
    else:
        return 'F'

l = [95, 82, 61, 40, 100]
for i in l:
    print(grade_from_score(i))
```

```python
# Task 19: Write a function count_vowels(text) that loops through a string (Day 2 + Day 3 for-loop) and returns how many vowels
# (a, e, i, o, u — case-insensitive) it contains. Test it on "Python Bootcamp" and on a sentence of your choice.

def count_vowels(text):
    c = 0
    text = text.lower()
    for ch in text:
        if ch == 'a' or ch == 'e' or ch == 'i' or ch == 'o' or ch == 'u' :
            c+=1

    print(c)

count_vowels("Python Bootcamp")
count_vowels("Varshith is happy")
```

```python
# Task 21: Write a function is_prime(n) that returns True if n is a prime number and False otherwise, using a for loop
# and the % operator. Then write a second function primes_up_to(limit) that uses is_prime() inside a loop to return a 
# list of all prime numbers up to limit. Test primes_up_to(30).

def is_prime(n):
    for i in range(2, n):
        if n%i == 0:
            print("Not Prime")
            break
    else:
        print("Prime")

def primes_up_to(limit):
    for n in range(2, limit):
        if n == 2:
            print(f"Prime")
            continue
        is_prime(n)

primes_up_to(30)
```

```python
# Task 22: Write a function password_strength(password) that returns "Weak", "Medium", or "Strong" based on simple rules you 
# define yourself (e.g. length, presence of digits, presence of uppercase letters) using string methods from Day 2 
# (such as .isdigit(), .isupper()) combined with conditionals. Test it on at least three different passwords.

def password_strength(password):
    has_upper = False
    has_digit = False

    for char in password:
        if char.isupper():
            has_upper = True
        if char.isdigit():
            has_digit = True

    if len(password) < 8:
        return "Weak"
    elif has_upper and has_digit:
        return "Strong"
    else:
        return "Medium"


print(password_strength("hello"))        
print(password_strength("helloworld"))   
print(password_strength("Hello123")) 
```
