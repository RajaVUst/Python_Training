```python
# Part B: Return Values vs. print()  

# Task 4: Write a function multiply(a, b) that RETURNS the product of a and b (do not print inside the function). 
# Store the result in a variable and print that variable.

def multiply(a,b):
    res = a*b
    return res

multiply(2,3)
multiply(4,5)
```

```python
# Task 5: Write two versions of a function that checks if a number is even: is_even_print(n) which prints "Even"/"Odd",
# and is_even_return(n) which returns True/False. Call both with the number 7 and observe the difference in what you can do with each result.

def is_even_print(n):
    if n%2 == 0:
        print("Even")
    else:
        print("Odd")

def is_even_return(n):
    if n%2 == 0:
        return True
    else:
        return False

is_even_print(7)
is_even_return(7)
```

```python
# Task 6: Write a function celsius_to_fahrenheit(celsius) that returns the Fahrenheit equivalent using the formula F = C * 9/5 + 32.
# Use it to convert 0, 37, and 100 degrees Celsius, printing each result.

def celsius_to_fahrenheit(celsius):
    f = celsius*(9/5) + 32
    print(f"{f:.2f}")

celsius_to_fahrenheit(0)
celsius_to_fahrenheit(37)
celsius_to_fahrenheit(100)
```

```python
# Part C: Positional, Default & Keyword Arguments  

# Task 7: Write a function order_summary(item, quantity=1) where quantity defaults to 1 if not provided. 
# Call it once with only 'item', and once with both arguments.

def order_summary(item, quantity=1):
    print(f'{item} : {quantity}')

order_summary("Bucket")
order_summary("Brush", 3)
```

```python
# Task 8: Write a function book_ticket(passenger, seat_type="Economy", meal="Veg") with two default parameters. 
# Call it: (a) with only the passenger name, (b) overriding just meal using a keyword argument, (c) overriding both defaults using 
# keyword arguments in a different order than they were defined.

def book_ticket(passenger, seat_type="Economy", meal="Veg"):
    print(passenger, seat_type, meal)

book_ticket("Varshith")
book_ticket("Varshith", meal="Non veg")
book_ticket("Varshith", meal="Non Non veg", seat_type="First Class")
```

```python
# Task 9: Write a function total_cost(*prices) using *args that returns the sum of any number of prices passed in. 
# Test it with 2 prices, then with 5 prices.

def total_cost(*prices):
    print(sum(prices))

total_cost(10,30,50)
total_cost(20,40,60)
```

```python
# Task 10: Write a function print_student_info(**details) using **kwargs that prints each key-value pair on its own line, formatted as "key: value".
# Call it with at least four keyword arguments of your choice (e.g. name, age, course, city).

def print_student_info(**details):
    print(details)

print_student_info(name="Varshith", age=22, course="Gen AI", city="Trivandrum")
```
