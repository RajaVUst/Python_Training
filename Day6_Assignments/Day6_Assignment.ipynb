```python
#Part G: Cumulative — Files, Exceptions, Functions & Data Structures
#Exercise 26: Safe score loader (function)
def load_scores(path):
    try:
        scores = {}
        with open(path,"r") as f:
            for row in f:
                try:
                    name,score = row.strip().split(",")
                    scores[name]=int(score)
                except ValueError:
                    continue
    except FileNotFoundError:
        return {}
    return scores
print(load_scores("scores.csv"))
```

```python
#Exercise 27: Validate and summarise
def summarise(scores):
    values = list(scores.values())
    
    
    summarised_scores = {}
    summarised_scores["highest"]= max(values)
    summarised_scores["count"] = len(values)
    summarised_scores["lowest"] = min(values)
    summarised_scores["average"] = sum(values)/(len(values))
    return summarised_scores
print(summarise({"Vinaya":99, "Varshith":95,"Prasanna" : 97, "Aishuuu" : 90, "Ajay" : 98}))
            
```

```python
#Exercise 28: Word frequency counter
def word_frequencies(path):
    frequencies = {}
    try:
        with open(path,"r") as f:
            text = f.read().lower()
            words = text.split()
            for word in words:
                frequencies[word] = frequencies.get(word,0)+1
    except FileNotFoundError:
        print("FileNot found")
    return frequencies
print(word_frequencies("notes.txt"))
        
```

```python
#Exercise 29: JSON config with validation
import json
class MissingConfigKeyError(Exception):
    pass

def load_config(path):
    with open(path, "r") as file:
        data = json.load(file)
        for config in data:
            if "batch_size" not in config:
                raise MissingConfigKeyError("Missing required key: batch_size")

            if "learning_rate" not in config:
                raise MissingConfigKeyError("Missing required key: learning_rate")

    return data
try:
    config = load_config("config.json")
    print("Config loaded successfully:")
    print(config)

except FileNotFoundError:
    print("Error: Config file not found.")

except MissingConfigKeyError as e:
    print(f"Configuration Error: {e}")
    
    
```

```python
#Exercise 30: Draft a mini log parser
def parse_log(path):
    logs = []
    skipped = 0
    try:
        with open(path, "r") as file:
            for line in file:
                line = line.strip()
                try:
                    date, time, level, message = line.split(" ", 3)
                    logs.append({
                        "timestamp": f"{date} {time}",
                        "level": level,
                        "message": message
                    })
                except ValueError:
                    skipped += 1

    except FileNotFoundError:
        print("Log file not found.")
        return []
    print(f"Parsed successfully: {len(logs)}")
    print(f"Skipped: {skipped}")
    return logs
parsed_logs = parse_log("access.log")
for log in parsed_logs:
    print(log)
```
