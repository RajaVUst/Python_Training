```python
# 10
# Exercise 10: Read a CSV with csv.reader
# Prompt: Open scores.csv with csv.reader, print the header separately from the data rows, and print each data row as a list.
# Hint: Use next(reader) to pull off the header row before looping over the rest.

import csv 
with open("scores.csv", "r") as f:
    reader = csv.reader(f)
    header = next(reader)
    print(header)

    for i in reader:
        print(list(i))
```

```python
# Exercise 11: Read a CSV with DictReader
# Prompt: Re-read scores.csv using csv.DictReader and print each row as "<name>: <score>" using dictionary access.
# Hint: Each row from DictReader behaves like a dict, e.g. row["name"].

import csv 
with open("scores.csv", "r") as f:
    data = csv.DictReader(f)
    for i in data:
        print(i['name'], i['score'])
```

```python
# Exercise 12: Compute an average from CSV data
import csv 
with open("scores.csv", "r") as f:
    data = csv.DictReader(f)
    res_sum = 0
    for i in data:
        if i['score']!="":
            res_sum += int(i['score'])
    print(res_sum)
```

```python
# Part D: Working with .json Files

# Exercise 14: Read a JSON file
# Prompt: Load profile.json with json.load() and print the value of "name" and the length of "completed_days".
# Hint: json.load(f) returns a Python dict; access it like any other dict.

import json
with open("profile.json", "r") as f:
    data = json.load(f)
    print(data['name'], len(data['completed_days']))
```

```python
# Exercise 15: Modify and re-save JSON
# Prompt: Load profile.json, append 6 to the "completed_days" list, then write the updated dict back to profile.json with json.dump() (use indent=2).
# Hint: You'll need two separate with blocks -- one to read (and close), one to write.

import json
with open("profile.json", "r") as f:
    data = json.load(f)
    data['completed_days'].append(6)

f.close()
with open("profile.json", "w") as f:
    json.dump(data, f, indent=2)
```

```python
# Exercise 16: String <-> JSON with loads/dumps
# Prompt: Build a small dict describing today's session (topic, duration_minutes, format), convert it to a JSON string with json.dumps(),
# print the string, then convert it back to a dict with json.loads() and print the type of the result to confirm it's a dict again.
# Hint: dumps() takes a Python object and returns a str; loads() takes a str and returns a Python object.

session = {'topic': 'Files', 'duration_minutes': '120',  'format': 'online'}

import json
json_data = json.dumps(session)
print(json_data)
dict_data = json.loads(json_data)
print(dict_data)
```

```python
# Exercise 17: CSV to JSON
# Prompt: Read all rows from scores.csv (Part C) with DictReader, collect them into a list of dicts, then write that list to scores.json
# using json.dump().
# Hint: DictReader rows are dict-like -- wrap each one in dict(row) if you want a plain dict before storing it in a list.
import csv
with open('scores.csv', 'r') as f:
    data = list(csv.DictReader(f))

import json
with open('scores.json', 'w') as f:
    json_data = json.dump(data, f, indent=2)
```
