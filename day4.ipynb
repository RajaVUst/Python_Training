## Part B: Return Values vs. print()

```python
#Task 4:
# Define the function
def multiply(a, b):
    return a * b

# Store the result in a variable
result = multiply(6, 4)

# Print the variable
print(result)
```

```python
#Task 5:
# Function that prints the result
def is_even_print(n):
    if n % 2 == 0:
        print("Even")
    else:
        print("Odd")

# Function that returns the result
def is_even_return(n):
    return n % 2 == 0

# Call the print version
is_even_print(7)

# Call the return version
result = is_even_return(7)
print(result)

# Using the returned value in a condition
if is_even_return(7):
    print("7 is even")
else:
    print("7 is odd")
```

```python
#Task 11:
# Function 1
def show_square(number):
    """Prints the square of the given number."""
    print(f"The square of {number} is {number ** 2}")

# Function 2
def celsius_to_fahrenheit(celsius):
    """Converts Celsius to Fahrenheit and returns the result."""
    return celsius * 9 / 5 + 32

# Display help information
help(show_square)
help(celsius_to_fahrenheit)
```

```python
#Task 12:
# Define the function
def reset_score():
    score = 0  # Local variable
    print("Score inside function:", score)

# Call the function
reset_score()

# Try to access score outside the function
print(score)  # This will cause an error

# Explanation
# The variable 'score' is local to the function reset_score().
# Local variables exist only while the function is running.
# After the function ends, 'score' is no longer accessible,
# so trying to print(score) outside the function raises a NameError.
```

```python
#Task 13:
# Global variable
total_attempts = 0

# Function to increment the global variable
def log_attempt():
    global total_attempts
    total_attempts += 1

# Call the function 3 times
log_attempt()
log_attempt()
log_attempt()

# Print the result
print("Total attempts:", total_attempts)
```

```python
#Task 15:
# Recursive function
def countdown(n):
    if n <= 0:  # Base case
        print("Liftoff!")
    else:
        print(n)
        countdown(n - 1)

# Test the function
countdown(5)
```

```python
#Task 16:
# Base case:
# If n is 1, return 1.

# Recursive case:
# Return n plus the sum of all integers from 1 to (n - 1).

def sum_upto(n):
    if n == 1:  # Base case
        return 1
    else:       # Recursive case
        return n + sum_upto(n - 1)

# Test the function
print(sum_upto(5))
```

## Part E: Cumulative Challenges

```python
#Task 17:
def is_palindrome(word):
    # Convert to lowercase so that case does not matter
    word = word.lower()
    return word == word[::-1]

# Test the function
print(is_palindrome("level"))
print(is_palindrome("python"))
print(is_palindrome("Madam"))
```

```python
#Task 18:
def grade_from_score(score):
    if score >= 90:
        return 'A'
    elif score >= 80:
        return 'B'
    elif score >= 70:
        return 'C'
    elif score >= 60:
        return 'D'
    else:
        return 'F'

# List of scores
scores = [95, 82, 61, 40, 100]

# Print grades using an f-string
for score in scores:
    print(f"{score} -> {grade_from_score(score)}")
```

```python
#Task 19:
def count_vowels(text):
    vowels = "aeiou"
    count = 0

    for char in text.lower():
        if char in vowels:
            count += 1

    return count

# Test the function
print(count_vowels("Python Bootcamp"))
print(count_vowels("Hello, how are you today?"))
```

```python
#Task 20:
def fizzbuzz_range(start, end):
    for num in range(start, end + 1):
        if num % 3 == 0 and num % 5 == 0:
            print("FizzBuzz")
        elif num % 3 == 0:
            print("Fizz")
        elif num % 5 == 0:
            print("Buzz")
        else:
            print(num)

# Test the function
fizzbuzz_range(1, 15)
```

```python
#Task 21:
def is_prime(n):
    if n < 2:
        return False

    for i in range(2, n):
        if n % i == 0:
            return False

    return True


def primes_up_to(limit):
    primes = []

    for num in range(2, limit + 1):
        if is_prime(num):
            primes.append(num)

    return primes


# Test the function
print(primes_up_to(30))
```

```python
#Task 22:
def password_strength(password):
    has_digit = False
    has_upper = False

    for char in password:
        if char.isdigit():
            has_digit = True
        if char.isupper():
            has_upper = True

    if len(password) >= 8 and has_digit and has_upper:
        return "Strong"
    elif len(password) >= 6:
        return "Medium"
    else:
        return "Weak"


# Test the function
print("abc ->", password_strength("abc"))
print("python1 ->", password_strength("python1"))
print("Python123 ->", password_strength("Python123"))
```

```python
#Task 23:
def build_utilities(operation, *values):
    if operation == "sum":
        return sum(values)

    elif operation == "max":
        return max(values)

    elif operation == "min":
        return min(values)

    elif operation == "average":
        return sum(values) / len(values)

    else:
        return "Invalid operation"


# Test cases
print(build_utilities("sum", 4, 8, 15))
print(build_utilities("average", 2, 4, 6, 8))
```
