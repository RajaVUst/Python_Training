```python
def countdown(n):
    if n <= 0:
        print("Liftoff!")
    else:
        print(n)
        countdown(n - 1)

countdown(5)
```

```python
def sum_upto(n):
    if n == 1:
        return 1
    else:
        return n + sum_upto(n - 1)

print(sum_upto(5)) 
print(sum_upto(1))
```

```python
def grade_from_score(score):
    if score >= 90:
        return 'A'
    elif score >= 80:
        return 'B'
    elif score >= 70:
        return 'C'
    elif score >= 60:
        return 'D'
    else:
        return 'F'
scores = [95, 82, 61, 40, 100]
for score in scores:
    print(f"{score} -> {grade_from_score(score)}")
```

```python
def count_vowels(text):
    count = 0
    for char in text:
        if char.lower() in "aeiou":
            count += 1
    return count
text1 = "Python Bootcamp"
print(f'"{text1}" has {count_vowels(text1)} vowels.')
text2 = "I love learning Python."
print(f'"{text2}" has {count_vowels(text2)} vowels.')
```

```python
def fizzbuzz_range(start, end):
    for number in range(start, end + 1):
        if number % 3 == 0 and number % 5 == 0:
            print("FizzBuzz")
        elif number % 3 == 0:
            print("Fizz")
        elif number % 5 == 0:
            print("Buzz")
        else:
            print(number)
fizzbuzz_range(1, 15)
```

```python
import math as m
# num=int(input("Enter a Number: "))
a=[]
def is_prime(n):
    if n<=1:
        return False
    s=int(m.sqrt(n))+1
    for i in range(2,s):
        if n%i==0:
            return False
    return True
def primes_up_to(limit):
    for i in range(limit):
        if is_prime(i):
            a.append(i)
    return a
primes_up_to(30)
```

```python
def password_strength(password):
    has_digit = False
    has_uppercase = False
    for char in password:
        if char.isdigit():
            has_digit = True
        if char.isupper():
            has_uppercase = True
    if len(password) >= 8 and has_digit and has_uppercase:
        return "Strong"
    elif len(password) >= 6 and (has_digit or has_uppercase):
        return "Medium"
    else:
        return "Weak"
print(password_strength("hello"))  
print(password_strength("Hello12"))     
print(password_strength("Hello1234"))   
```

```python
def build_utilities(operation, *values):
    if operation == "sum":
        return sum(values)
    elif operation == "max":
        return max(values)
    elif operation == "min":
        return min(values)
    elif operation == "average":
        return sum(values) / len(values)
    else:
        return "Invalid operation"
print(build_utilities("sum", 4, 8, 15))
print(build_utilities("average", 2, 4, 6, 8))
```
