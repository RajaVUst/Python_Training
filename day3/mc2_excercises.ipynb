# Part E: Combined Logic — Classic Problems 

```python
#Exercise 16: FizzBuzz 
n = 20
for i in range (1, n + 1):
    if i % 15 == 0:
        print("FizzBuzz")
    elif i % 3 ==0:
        print("Fizz")
    elif i % 5 == 0:
        print("Buzz")
    else:
        print(i)
```

```python
#Exercise 17: Reverse an integer 
number = 12321
reverse = 0
while number > 0:
    reverse *= 10
    reverse += number % 10
    number //= 10
print (reverse)
```

```python
#Exercise 18: Palindrome check for an integer 
number = 12321
n = number
reverse = 0
while number > 0:
    reverse *= 10
    reverse += number % 10
    number //= 10
if reverse == n:
    print("palindrome")
else:
    print("not palindrome")
```

```python
#Exercise 19: Prime number check 
n = 2
flag = 0
for i in range (2, n):
    if n % i == 0:
        flag = 1
        print("not prime")
        break
if flag == 0:
    print("prime")
```

```python
#Exercise 20: Print primes in a range 
for j in range (2, 51):
    flag = 0
    for i in range (2, j):
        if j % i == 0:
            flag = 1
            break
    if flag == 0:
        print(j)
    
```

```python
#Exercise 24: Number guessing game (fixed target) 
secret_number = 7
guesses = [1, 3, 9, 7, 5, 10]
for i in guesses:
    if i == secret_number:
        print("Correct!")
        break
    if i < secret_number:
        print("too low")
    else:
        print("too high")
```
