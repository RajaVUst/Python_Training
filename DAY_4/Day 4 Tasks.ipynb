```python
##exercise 17

def palindrome(str):
    reverse = str[::-1]
    if str == reverse:
        print(f"{str} is Palindrome")
    else:
        print(f"{str} Not a palindrome")
palindrome('level')
palindrome('python')
palindrome('madam')
```

```python
##exercise 18
def grade_from_score(score):
    if score>=90:
        print(f"score is {score} Your grade is A")
    elif 90>score>=75:
        print(f"score is {score} Your grade is B")
    elif 75>score>=60:
        print(f"score is {score} Your grade is C")
    elif 60>score>=40:
        print(f"score is {score} Your grade is D")
    else:
        print(f"score is {score} Your grade is F")
scores = [95, 82, 61, 40, 100]
for s in scores:
    grade_from_score(s)
```

```python
##exercise 20

def FizzBuzz(start , end):
    for i in range(start,end+1):
        if i%3 == 0 and i%5 == 0:
            print(f"{i } is FIZZBUZZ")
        elif i%3 ==0:
            print(f"{i } is FIZZ")
        elif i%5 == 0:
            print(f"{i } is BUZZ")
        else:
            print(f"{i } is Not divisible by 3 and 5")
FizzBuzz(2,9)
```

```python
##exercise 21

def is_prime(n):
    count = 0
    if n==0 or n==1:
        return False
    else:
        for i in range(2,n):
            if n%i == 0:
                return False
        else:
            return True
def prime_range(num):
    primee = []
    for i in range(num+1):
        if is_prime(i):
            primee.append(i)
    print("Prime number within the range : ",primee)


Prime = int(input("Enter range for prime nambers : "))
prime_range(Prime)
```

```python
##exercise 22
def passwordcheck(password):
    length = len(password)
    has_upper = any(ch.isupper() for ch in password)
    has_digit = any(ch.isdigit() for ch in password)
    if length>6 and has_upper and has_digit:
        print("Strong password")
    elif length>6 and (has_upper or has_digit):
        print("Medium password")
    else:
        print("Weak password")
passwordcheck('SJVP3233')
passwordcheck('wcbwcw212')
passwordcheck('wedxSZA')
```

```python
##exercise 23

def build_utilities(operation,*values):
    if operation == 'sum':
        s = sum(values)
        print(f"Sum of all numbers : {s}")
    elif operation == 'average':
        s = sum(values)
        c = len(values)
        avg = s/c
        print(f"Average of all numbers : {avg}")
    elif operation == 'max' :
        m = max(values)
        print(f"Maximum of all numbers : {m}")
    elif operation == 'min' :
        mi = min(values)
        print(f"Minimum of all numbers : {mi}")
    else:
        print("Invalid operation please enter valid operation from sum,average,min,max")
build_utilities("sum", 4, 8, 15)
build_utilities("average", 2, 4, 6, 8)
```
