```python
def is_palindrome(word):
    word=word.lower()
    if word==word[::-1]:
        return True
    else:
        return False
print(is_palindrome("level"))
print(is_palindrome("python"))
print(is_palindrome("Madam"))
```

```python
def grade(score):
    if score>=90:
        return "A"
    elif score>=80:
        return "B"
    elif score>=70:
        return "c"
    elif score>=60:
        return "d" 
    else:
        return "FAil"               
scores=list(map(int,input("ENter the scores").split()))
for num in scores:
    result=grade(num) 
    print(f"{num}->{result}", end=" ") 
```

```python
def count_vowels(word):
    count=0
    vowel='aeiou'
    vowel=vowel.lower()
    for w in word:
        if w in vowel:
            count+=1
    return count   
    

word=input("Enter the sentence")
print(count_vowels(word))
```

```python
def fizzbuz_range(start, end):
    for i in range(start, end+1):
        if i%3==0 and i %5==0:
            print("FIzzBuzz")
        elif i%3==0:
            print("FIZZ")
        elif   i%5==0:
            print("Buzz")
        else:
            print(f"{i}  is not divisble wiith 3 or 5")
start, end=map(int,input().split())
fizzbuz_range(start, end) 
```

```python
def is_prime(n):
    if n<2:
        return False
    for i in range(2,n):
        if n%i==0:
            return False
    return True


def prme_limit(limit):
    primes=[]
    non_primes=[]
    for num in range(2, limit+1):
        if is_prime(num):
            primes.append(num)
    return primes      
   

print(prme_limit(30))
```

```python

def is_password(password):
    digit=False
    upper=False
    for ch in password:
        if ch.isdigit():
            digit= True

        if ch.isupper():
            upper=True

    if len(password)<6:
        return "weak"
    elif digit and upper and len(password)>8:
        return "strong"
    else:
        return "medium"        
print(is_password("123"))    
print(is_password("123iytirytU"))    
print(is_password("123jgkjk"))   
```

```python
def build_utilities(operation, *values):
    if operation=="sum":
        return f"sum of the sumbers:{sum(values)}"
    elif operation=="max":
        return f"maximum nuber:{max(values)}"
    elif operation=="min":
        return f"maximum nuber:{min(values)}"
    elif operation=="average":
        return f"average of numbers:{sum(values)/len(values)}"
    else:
        return "Invalid operation"
result=build_utilities("sum",1, 2,3)        
print(f"Build_utilites:sum:->{result}")   
```
