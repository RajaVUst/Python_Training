  Exercise 6  ·  Restaurant Bill Splitter  

```python
tot_bill = float(input("Enter total bill amount: "))
no_of_people = int(input("Enter the number of people splitting it: "))
wants_tip = input("whether the group wants to add a 10% tip: ") == "yes"
tip = (wants_tip == True  and (tot_bill*10)/100 or 0)
share = ((tot_bill+tip)/no_of_people)
print(f"{share:.1f}")
```

```python
a = 5
b = 10

temp = a
a = b
b = temp
print(a, b)

a, b = b, a
print(a,b)
```

  Exercise 8  ·  Rectangle Calculator  

```python
length = float(input("Enter rectangle's length: "))
width = float(input("Enter rectangle's width: "))

area = length * width
perimeter = 2*(length + width)

print(f"Area of Rectangle: {area}")
print(f"Perimeter of Rectangle: {perimeter}")
```
