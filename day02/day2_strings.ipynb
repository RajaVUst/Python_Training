```python
# 1.Create a variable holding a messy sentence
sentence = "   Python IS Fun to Learn!!   "
```

```python
# 2. Removing the leading/trailing whitespace 
clean = sentence.strip()
print(clean)
```

```python
# 3. printing lowercase and an uppercase version of the cleaned sentence
print(clean.lower())
print(clean.upper())
```

```python
# 4. Indexing and slicing
print(clean[0])
print(clean[-1])
print(clean[:5])
```

```python
# 5. split the sentence into words and count them
words = clean.split()
print(words)
print(len(words))
```

```python
# 6. Swap the words fun for powerful
print(clean.replace("Fun", "Powerful"))
```

```python
# 7. Rebulding the sentence with - 
print("-".join(words))
```

```python
# 8. printing the Length of the sentence and reversing of the sentence
print(f"Words: {len(words)} | Reversed: {clean[::-1]}")
```

```python
#   Exercise 1  ·  Name Tag Formatter  
first_name = input("Enter you first name :")
last_name =  input("Enter you last name :")
print(f" Hello, {first_name.title()} {last_name.title()}!")
```

```python
# Exercise 3  ·  Palindrome Checker  
word = "level"
is_palindrome = (word == word[::-1])
print(is_palindrome)
```

```python
# Exercise 4  ·  Email Splitter  
email = "reni.k@company.com"
username, domain = email.split("@")
print(f"Usernname: {username}")
print(f"Domain: {domain}")
```

```python
#   Exercise 5  ·  Slug Normalizer  
title = "  My First   Python Project!!  "
title = title.strip()
print(title)

title = title.replace("!!", "")
print(title)

title = title.lower()
print(title)

title = title.replace(" ", "_")
print(title)
```

```python
#   Exercise 6  ·  Sentence Dashboard  
sentence = input("Enter a sentence:")

character_count = len(sentence)
word_count = len(sentence.split())

vowel_count = (
    sentence.count("a")
    + sentence.count("e")
    + sentence.count("i")
    + sentence.count("o")
    + sentence.count("u")
)

longer_than_30 = len(sentence) > 30

print(f"Character count: {character_count}")
print(f"Word count: {word_count}")
print(f"Vowel count: {vowel_count}")
print(f"Longer than 30 characters: {longer_than_30}")
```
