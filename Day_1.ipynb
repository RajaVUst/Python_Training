```python
celsius = float(input("Enter temperature in Celsius: "))
fahrenheit = celsius * 9 / 5 + 32
print(f"Celsius: {celsius:.1f}°C, Fahrenheit: {fahrenheit:.1f}°F")
```

```python
bill = float(input("Enter the total bill amount: "))
people = int(input("Enter the number of people: "))
wants_tip = (input("Do you want to add a 10% tip? (yes/no): ").lower() == "yes")

if wants_tip:
    bill = bill * 1.10
share = bill / people

print(f"Each person's share: {share:.2f}")
```

```python
a = 5
b = 10

print(f"Before swap: a = {a}, b = {b}")

temp = a
a = b
b = temp

print(f"method 1: a = {a}, b = {b}")

a = 5
b = 10

a, b = b, a

print(f"method 2: a = {a}, b = {b}")
```

```python
n = 15

div_by_3 = (n % 3 == 0)
div_by_5 = (n % 5 == 0)
div_by_both = div_by_3 and div_by_5

print(f"Divisible by 3: {div_by_3}")
print(f"Divisible by 5: {div_by_5}")
print(f"Divisible by both: {div_by_both}")
```

```python
score1 = float(input("Enter first score: "))
score2 = float(input("Enter second score: "))
score3 = float(input("Enter third score: "))

average = (score1 + score2 + score3) / 3

has_passed = average >= 40

print(f"Average: {average:.1f} | Passed: {has_passed}")
```
