```python
# Exercise 5: Celsius to Fahrenheit Converter

celsius = float(input("Enter temperature in Celsius: "))

fahrenheit = celsius * 9 / 5 + 32

print(f"Celsius: {celsius:.1f}°C")
print(f"Fahrenheit: {fahrenheit:.1f}°F")
```

```python
# Exercise 6: Restaurant Bill Splitter

total_bill = float(input("Enter total bill amount: "))
number_of_people = int(input("Enter number of people: "))

wants_tip = input("Do you want to add a 10% tip? (yes/no): ").lower() == "yes"

if wants_tip:
    total_amount = total_bill * 1.10
else:
    total_amount = total_bill

share = total_amount / number_of_people

print(f"Total amount: ₹{total_amount:.2f}")
print(f"Each person's share: ₹{share:.2f}")
```

```python
# Exercise 7: Variable Swap

# Original values
a = 5
b = 10

print("Before swapping:")
print(f"a = {a}")
print(f"b = {b}")

# Method 1: Using a temporary variable
temp = a
a = b
b = temp

print("\nAfter swapping using temporary variable:")
print(f"a = {a}")
print(f"b = {b}")

# Reset values
a = 5
b = 10

# Method 2: Python multiple assignment
a, b = b, a

print("\nAfter swapping using multiple assignment:")
print(f"a = {a}")
print(f"b = {b}")
```
