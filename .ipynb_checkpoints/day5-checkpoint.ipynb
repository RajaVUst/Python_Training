# Day5 - Hands-On Exercises (Collections)

## Part A — Lists

```python
# Exercise 1: Shopping List Manager
# Create a list with three grocery items
cart = ["milk", "bread", "eggs"]

# Append two more items
cart.append("apples")
cart.append("rice")

# Remove one item
cart.remove("bread")

# Print the original list
print("Original cart:", cart)

# Print the list sorted alphabetically without changing the original
print("Sorted cart:", sorted(cart))
```

```python
# Exercise 2: Rotate the Queue
# Given queue
queue = ["Amit", "Reni", "Tara", "Sam"]

# Print the original queue
print("Before rotation:", queue)

# Remove the first person and add them to the end
person = queue.pop(0)
queue.append(person)

# Print the updated queue
print("After rotation:", queue)
```

```python
# Exercise 3: Running Total With Slicing
# Given list of readings
readings = [12, 15, 9, 22, 30, 4, 18]

# First three readings
print("First three readings:", readings[:3])

# Last two readings
print("Last two readings:", readings[-2:])

# Every second reading starting from index 0
print("Every second reading:", readings[::2])
```

```python
# Exercise 4: Copy vs. Reference
# Create the original list
original = [1, 2, 3]

# Create an alias (reference) and a copy
alias = original
safe_copy = original.copy()

# Modify both lists
alias.append(100)
safe_copy.append(200)

# Print the results
print("original:", original)
print("alias:", alias)
print("safe_copy:", safe_copy)

# alias points to the same list as original, so changes through alias affect original; safe_copy is a separate list.
```

```python
# Exercise 6: Unpacking a Tuple
# Store the student's record in a tuple
record = ("Reni", 91, "Cohort 2")

# Unpack the tuple into variables
name, score, cohort = record

# Print a formatted sentence using an f-string
print(f"{name} scored {score} marks and belongs to {cohort}.")
```

```python
# Exercise 7: Multiple Return Values
def stats(numbers):
    low = min(numbers)
    high = max(numbers)
    avg = sum(numbers) / len(numbers)
    return low, high, avg

# Call the function
minimum, maximum, average = stats([4, 9, 1, 7, 15])

# Print the results
print(f"Minimum: {minimum}")
print(f"Maximum: {maximum}")
print(f"Average: {average}")
```

## Part C — Sets

```python
# Exercise 8: De-duplicating Attendance
# Given attendance list
attendance = ["amit", "reni", "amit", "tara", "reni", "sam"]

# Convert to a set to remove duplicates
unique_attendees = set(attendance)

# Print the unique attendees
print("Unique attendees:", unique_attendees)

# Print the count of unique attendees
print("Number of unique attendees:", len(unique_attendees))
```

```python
# Exercise 9: Common Interests
# Given sets
cohort_a = {"python", "sql", "excel"}
cohort_b = {"python", "tableau", "excel"}

# Skills common to both cohorts
print("Common skills:", cohort_a & cohort_b)

# Skills unique to cohort_a
print("Skills only in cohort_a:", cohort_a - cohort_b)

# All skills across both cohorts
print("All skills:", cohort_a | cohort_b)
```

```python
# Exercise 10: Fast Membership Check
# Create a set of 1000 numbers
seen = set(range(1000))

# Check whether 999 is in the set
if 999 in seen:
    print("999 is in the set.")
else:
    print("999 is not in the set.")

# Sets are better than lists for repeated membership checks because set lookups are much faster on large collections.
```

## Part D — Dictionaries & Nested Structures

```python
# Exercise 11: Contact Card
# Create a dictionary
contact = {
    "name": "Reni",
    "email": "reni@example.com",
    "phone": "1234567890"
}

# Print each value by key
print("Name:", contact["name"])
print("Email:", contact["email"])
print("Phone:", contact["phone"])

# Add a new key
contact["city"] = "Trivandrum"

# Update the phone number
contact["phone"] = "9876543210"

# Remove the email key
contact.pop("email")

# Print the updated dictionary
print("Updated contact:", contact)
```

```python
# Exercise 12: Word Frequency Counter
# Given sentence
text = "the quick brown fox jumps over the lazy dog the fox runs"

# Create an empty dictionary
word_counts = {}

# Count occurrences of each word
for word in text.split():
    word_counts[word] = word_counts.get(word, 0) + 1

# Print the dictionary
print(word_counts)
```

```python
# Exercise 13: Class Roster (Nested Structure)
# List of dictionaries
roster = [
    {"name": "Amit", "scores": [85, 90, 88]},
    {"name": "Reni", "scores": [92, 95, 89]},
    {"name": "Tara", "scores": [78, 82, 80]}
]

# Print each student's average score
for student in roster:
    average = sum(student["scores"]) / len(student["scores"])
    print(f"{student['name']}: {round(average, 1)}")
```

```python
# Exercise 14: Lookup Table
# Grade thresholds
grade_lookup = {
    "A": 90,
    "B": 80,
    "C": 70,
    "D": 60,
    "F": 0
}

score = 84

# Check thresholds from highest to lowest
for grade, threshold in sorted(
        grade_lookup.items(),
        key=lambda item: item[1],
        reverse=True):
    if score >= threshold:
        print(f"Score {score} receives grade {grade}")
        break
```

## Part E — Comprehensions

```python
# Exercise 15: Squares and Cubes
# List of cubes from 1 to 10
cubes = [n ** 3 for n in range(1, 11)]

# List of even cubes only
even_cubes = [n ** 3 for n in range(1, 11) if n % 2 == 0]

print("Cubes:", cubes)
print("Even cubes:", even_cubes)
```

```python
# Exercise 16: Label the Temperatures
temps = [15, 22, 31, 8, 27, 19]

labels = [
    "hot" if t >= 25 else ("mild" if t >= 15 else "cold")
    for t in temps
]

print(labels)
```

```python
# Exercise 17: Dictionary From Two Lists
names = ["Amit", "Reni", "Tara"]
scores = [78, 91, 65]

scores_dict = {
    name: score
    for name, score in zip(names, scores)
}

print(scores_dict)
```

```python
# Exercise 18: Filtering With a Dict Comprehension
names = ["Amit", "Reni", "Tara"]
scores = [78, 91, 65]

scores_dict = {
    name: score
    for name, score in zip(names, scores)
}

passing_scores = {
    name: score
    for name, score in scores_dict.items()
    if score >= 70
}

print("Original:", scores_dict)
print("70 and above:", passing_scores)
```

## Part F — Cumulative Challenges

```python
# Exercise 19: Vowel Counter by Word
def vowel_counts(sentence):
    vowels = "aeiou"

    return {
        word: sum(1 for ch in word.lower() if ch in vowels)
        for word in sentence.split()
    }

# Test
result = vowel_counts("The Quick Brown Fox Jumps")
print(result)
```

```python
# Exercise 20: Grade Book Summary
def grade_summary(roster):
    summary = []

    for student in roster:
        avg = sum(student["scores"]) / len(student["scores"])

        if avg >= 90:
            grade = "A"
        elif avg >= 80:
            grade = "B"
        elif avg >= 70:
            grade = "C"
        elif avg >= 60:
            grade = "D"
        else:
            grade = "F"

        summary.append({
            "name": student["name"],
            "average": round(avg, 1),
            "grade": grade
        })

    return summary


# Test
roster = [
    {"name": "Amit", "scores": [78, 82, 80]},
    {"name": "Reni", "scores": [95, 91, 93]},
    {"name": "Tara", "scores": [65, 70, 68]}
]

print(grade_summary(roster))
```

```python
# Exercise 21: Unique Word Finder
def unique_words(text):
    words = text.lower().split()
    return sorted(set(words))

# Test
sentence = "Python is fun and Python is powerful"
print(unique_words(sentence))
```

```python
# Exercise 22: FizzBuzz, Structured
def fizzbuzz_map(n):
    result = {}

    for i in range(1, n + 1):
        if i % 3 == 0 and i % 5 == 0:
            result[i] = "FizzBuzz"
        elif i % 3 == 0:
            result[i] = "Fizz"
        elif i % 5 == 0:
            result[i] = "Buzz"
        else:
            result[i] = str(i)

    return result

# Test
print(fizzbuzz_map(20))
```

```python
# Exercise 23: Password Strength Report (Capstone)
def password_report(passwords):
    report = {
        "weak": [],
        "medium": [],
        "strong": []
    }

    for pw in passwords:
        has_digit = any(ch.isdigit() for ch in pw)

        if len(pw) >= 10 and has_digit:
            report["strong"].append(pw)
        elif len(pw) >= 6:
            report["medium"].append(pw)
        else:
            report["weak"].append(pw)

    return report


# Test
passwords = [
    "cat",
    "secret",
    "welcome1",
    "mypassword123",
    "hello",
    "python2024"
]

print(password_report(passwords))
```
