```python
# Part E: Combined Logic — Classic Problems

# Exercise 16: FizzBuzz
for i in range(1, 51):
    if i%15 == 0:
        print("FizzBuzz")
    elif i%3 == 0:
        print("Fizz")
    elif i%5 == 0:
        print("Buzz")
    else:
        print(i)
```

```python
# Exercise 17: Reverse an integer
n = 1234
rev = 0

while n>0:
    r = n%10
    rev = rev*10 + r
    n = n//10

print(rev)
```

```python
# Exercise 18: Palindrome check for an integer
n = 1222
t = n
rev = 0

while n>0:
    r = n%10
    rev = rev*10 + r
    n = n//10

if t == rev:
    print("Palindrome")
else:
    print("Not a Palindrome")
```

```python
# Exercise 19: Prime number check
n = 89

for i in range(2, n):
    if n%i == 0:
        print("Not Prime")
        break
else:
    print("Prime")
    
```

```python
# Exercise 20: Print primes in a range

for n in range(2, 51):
    if n == 2:
        print(f"{n} Prime")
        continue
    for i in range(2, n):
        if n%i == 0:
            print(f"{n} Not Prime")
            break
    else:
        print("Prime")
```

```python
# Exercise 22: Fibonacci series
n = 15
s1 = 0
s2 = 1
print(0)
print(0)

for i in range(n-1):
    print(s1+s2)
    t = s1
    s1 = s2
    s2 = t+s2
```
