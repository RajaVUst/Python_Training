```python
# Task 1.1 — Inventory Tracker
# Scenario: Track warehouse items received today using appropriate Python data structures.

items = ["pen", "notebook", "pen", "eraser", "notebook", "pen", "stapler"]

# List: preserves arrival order and allows duplicates, making it ideal for tracking arrivals.
arrival_list = items

# Tuple: immutable record of the first batch received.
first_batch = tuple(items[:3])

# Set: stores only unique item names.
unique_items = set(items)

# Dictionary: maps each item to the number of times it was received.
item_count = {}

for item in items:
    item_count[item] = item_count.get(item, 0) + 1

print("List (arrival order):", arrival_list)
print("Tuple (first 3 items):", first_batch)
print("Set (unique items):", unique_items)
print("Dictionary (item counts):", item_count)
```

```python
# Task 1.2 — Comprehensions Rewrite
# Scenario: Rewrite loop-based code using list, dictionary, and set comprehensions.

# Original data
words = ["python", "java", "c", "kotlin"]
sentence = "the quick brown fox jumps over the lazy dog"

# List Comprehension
# Creates a list of squares for even numbers from 1 to 20.
squares = [n * n for n in range(1, 21) if n % 2 == 0]

# Dictionary Comprehension
# Creates a dictionary with words as keys and their lengths as values.
word_lengths = {word: len(word) for word in words}

# Set Comprehension
# Creates a set of unique vowels found in the sentence.
unique_vowels = {ch for ch in sentence if ch in "aeiou"}

print("Squares:", squares)
print("Word Lengths:", word_lengths)
print("Unique Vowels:", unique_vowels)
```

```python
# Task 1.3 — Nested Comprehension Challenge
# Scenario: Work with multiple sentences using nested comprehensions.

sentences = [
    "the sky is blue",
    "python is fun",
    "list comprehensions are handy"
]

# Nested list comprehension:
# Creates a flat list of all words longer than 2 characters from all sentences.
long_words = [
    word
    for sentence in sentences
    for word in sentence.split()
    if len(word) > 2
]

# Set comprehension:
# Creates a case-insensitive set of all unique words across all sentences.
unique_words = {
    word.lower()
    for sentence in sentences
    for word in sentence.split()
}

print("Words longer than 2 characters:", long_words)
print("Unique words (case-insensitive):", unique_words)

# Checkpoint Answer:
# In my opinion, a nested comprehension becomes less readable when it contains
# multiple loops and complex conditions. For example, a comprehension with
# three nested loops and several if-else conditions is harder to understand
# than a plain loop. In such cases, using regular loops improves readability
# and makes debugging easier.
```

```python
# Task 2.1 — Student Grade Book
# Scenario: Build a small grade book system using dictionaries, sets, and sorting.

students = {
    "alice": [88, 92, 79],
    "bob": [65, 70, 68],
    "carol": [95, 91, 89],
    "david": []  # Empty grade list for testing
}

# Dictionary comprehension:
# Maps each student to their average grade.
# If the grade list is empty, assign 0 to avoid ZeroDivisionError.
averages = {
    name: (sum(grades) / len(grades) if grades else 0)
    for name, grades in students.items()
}

# Set comprehension:
# Finds students whose every grade is above 80.
students_above_80 = {
    name
    for name, grades in students.items()
    if grades and all(grade > 80 for grade in grades)
}

# Sort students by average grade in descending order.
sorted_students = sorted(
    averages.items(),
    key=lambda item: item[1],
    reverse=True
)

print("Average Grades:", averages)
print("Students with all grades above 80:", students_above_80)
print("Sorted by Average:", sorted_students)
```

```python
# Task 2.2 — Set Operations: Access Control
# Scenario: Compare tool access between two teams using set operators.

team_a = {"figma", "jira", "slack", "github"}
team_b = {"jira", "slack", "notion", "vscode"}

# Common tools
common_tools = team_a & team_b

# Tools only Team A has
unique_to_a = team_a - team_b

# Combined toolset
all_tools = team_a | team_b

# Tools only one team has
symmetric_difference = team_a ^ team_b

print("Common Tools:", common_tools)
print("Unique to Team A:", unique_to_a)
print("Combined Toolset:", all_tools)
print("Symmetric Difference:", symmetric_difference)
```

```python
# Task 4.1 — try/except/finally Basics
# Scenario: Safely divide two values while handling common exceptions.

def safe_divide(a, b):
    try:
        return a / b

    except ZeroDivisionError:
        print("Warning: Cannot divide by zero.")
        return None

    except TypeError:
        print("Warning: Invalid data type provided.")
        return None

    finally:
        print("Operation attempted")


# Example calls
print(safe_divide(10, 2))
print(safe_divide(10, 0))
print(safe_divide("10", 2))
```

```python
1. Difference Between a Specific Exception and a Bare except:

Catching a specific exception such as ZeroDivisionError means the program
handles only the error you expect. Unexpected errors are allowed to
surface, making debugging easier.

A bare except: catches every exception type, including programming bugs,
keyboard interrupts, and system-exit events. This can hide problems and
make debugging difficult.

Because it masks errors and can make programs harder to maintain, a bare
except: is generally considered bad practice. It is usually better to
catch only the exceptions you expect.

------------------------------------------------------------

2. Purpose of finally:

The finally block runs whether an exception occurs or not. It is used
for cleanup actions that must always happen.

Real-world example:

Suppose you manually open a file using open() instead of a with block.
Even if an exception occurs while processing the file, the finally block
can ensure file.close() is executed so the file is properly released.

Similarly, finally is often used to close database connections, network
sockets, or release other system resources.

------------------------------------------------------------

3. Why Catch File Errors Separately?

Different exceptions indicate different problems:

• FileNotFoundError
  The file does not exist.

• PermissionError
  The user does not have permission to access the file.

• IsADirectoryError
  A directory was supplied where a file was expected.

Handling them separately allows the program to provide clear messages
and appropriate recovery actions for each situation.

If all file errors are caught using a generic Exception, the user may
receive vague error messages and it becomes harder to diagnose the real
problem.
```

```python
# Task 4.3 — Capstone: Robust CSV → JSON Converter
# File: convert_orders.py

import csv
import json

# --------------------------------------------------
# Step 1: Create Sample orders.csv
# --------------------------------------------------

sample_orders = [
    ["1001", "Alice", "250.50", "Completed"],
    ["1002", "Bob", "175.25", "Pending"],
    ["1003", "Alice", "320.00", "Completed"],
    ["1004", "Carol", "", "Pending"],           # Missing amount
    ["1005", "David", "invalid", "Cancelled"], # Invalid amount
    ["1006", "Bob", "95.00", "Completed"],
    ["1007", "Eve", "410.75", "Pending"],
    ["1008", "Alice", "120.00", "Completed"]
]

with open("orders.csv", "w", newline="") as file:
    writer = csv.writer(file)
    writer.writerow(["order_id", "customer", "amount", "status"])
    writer.writerows(sample_orders)

# --------------------------------------------------
# Step 2: Process CSV and Generate Summary
# --------------------------------------------------

customer_totals = {}
statuses = set()

try:

    with open("orders.csv", "r", newline="") as file:

        reader = csv.DictReader(file)

        for row in reader:

            try:
                amount = float(row["amount"])

            except (ValueError, TypeError):
                print(
                    f"Warning: Invalid amount for customer "
                    f"{row['customer']} "
                    f"(order {row['order_id']}). Row skipped."
                )
                continue

            customer = row["customer"]

            customer_totals[customer] = (
                customer_totals.get(customer, 0) + amount
            )

            statuses.add(row["status"])

    summary = {
        "customer_totals": customer_totals,
        "statuses": sorted(list(statuses))
    }

    with open("orders_summary.json", "w") as file:
        json.dump(summary, file, indent=2)

    print("orders_summary.json created successfully.")

except FileNotFoundError:
    print("Error: orders.csv was not found.")

except Exception as e:
    print("Unexpected error:", e)

finally:
    print("Conversion attempt finished")
```
