```python
items = ["pen", "notebook", "pen", "eraser", "notebook", "pen", "stapler"]

# A list is suitable because it preserves the exact arrival order.
item_list = items.copy()

# A tuple is suitable because the first batch should be immutable.
first_batch = tuple(items[:3])

# A set is suitable because it stores only unique item names.
unique_items = set(items)

# A dictionary is suitable because it maps each item to its number of occurrences.
item_counts = {}

for item in items:
    item_counts[item] = item_counts.get(item, 0) + 1

print("List:", item_list)
print("Tuple:", first_batch)
print("Set:", unique_items)
print("Dictionary:", item_counts)
```

```python
sentences = [
    "the sky is blue",
    "python is fun",
    "list comprehensions are handy"
]

# Flat list of words longer than 2 characters
long_words = [word for sentence in sentences for word in sentence.split() if len(word) > 2]

# Set of all unique words, case-insensitive
unique_words = {word.lower() for sentence in sentences for word in sentence.split()}

print("Words longer than 2 characters:", long_words)
print("Unique words:", unique_words)

# A nested comprehension becomes less readable when there are several
# levels of nesting or many conditions. For example, if I need three or
# four for/if parts, I would prefer a normal loop because it is easier to follow.
```

```python
students = {
    "alice": [88, 92, 79],
    "bob": [65, 70, 68],
    "carol": [95, 91, 89],
    "dave": []   # Testing an empty grade list
}

# 1. Create a dictionary of student -> average grade.
# If grades is empty, use 0 instead of dividing by zero.
averages = {
    name: (sum(grades) / len(grades) if grades else 0)
    for name, grades in students.items()
}

print("Average grades:", averages)


# 2. Find students who have every grade above 80.
above_80 = {
    name
    for name, grades in students.items()
    if grades and all(grade > 80 for grade in grades)
}

print("Students with every grade above 80:", above_80)


# 3. Sort students by average grade from highest to lowest.
sorted_students = sorted(
    averages.items(),
    key=lambda item: item[1],
    reverse=True
)

print("Students sorted by average:", sorted_students)
```

```python
team_a = {"figma", "jira", "slack", "github"}
team_b = {"jira", "slack", "notion", "vscode"}

# Tools common to both teams
common = team_a & team_b

# Tools only team A has
only_team_a = team_a - team_b

# All tools available to either team
combined = team_a | team_b

# Tools that only one of the two teams has
symmetric_difference = team_a ^ team_b

print("Common tools:", common)
print("Only team A:", only_team_a)
print("Combined tools:", combined)
print("Symmetric difference:", symmetric_difference)

# The symmetric difference answers:
# "Which tools are available to one team but not the other?"


# Bonus: function that returns all four results
def access_report(team_a, team_b):
    return {
        "common": team_a & team_b,
        "only_team_a": team_a - team_b,
        "combined": team_a | team_b,
        "symmetric_difference": team_a ^ team_b
    }


report = access_report(team_a, team_b)
print("Access report:", report)
```

```python
import csv

employees = [
    ["Amit", "Engineering", 60000],
    ["Reni", "Marketing", 50000],
    ["Tara", "Engineering", 70000],
    ["Sam", "Sales", 45000],
    ["Priya", "Engineering", 65000]
]

with open("employees.csv", "w", newline="") as f:
    writer = csv.writer(f)

    # Write the header
    writer.writerow(["name", "department", "salary"])

    # Write all employee records
    writer.writerows(employees)

print("employees.csv created successfully.")

print("\nEmployee records:")

with open("employees.csv", "r", newline="") as f:
    reader = csv.DictReader(f)

    for row in reader:
        print(dict(row))


department_totals = {}

with open("employees.csv", "r", newline="") as f:
    reader = csv.DictReader(f)

    for row in reader:
        department = row["department"]
        salary = float(row["salary"])

        department_totals[department] = (
            department_totals.get(department, 0) + salary
        )

print("\nTotal salary per department:")
for department, total in department_totals.items():
    print(f"{department}: {total:.2f}")


updated_employees = []

with open("employees.csv", "r", newline="") as f:
    reader = csv.DictReader(f)

    for row in reader:
        salary = float(row["salary"])

        if row["department"] == "Engineering":
            salary = salary * 1.10

        updated_employees.append({
            "name": row["name"],
            "department": row["department"],
            "salary": salary
        })


with open("employees_updated.csv", "w", newline="") as f:
    fieldnames = ["name", "department", "salary"]

    writer = csv.DictWriter(f, fieldnames=fieldnames)

    writer.writeheader()

    for employee in updated_employees:
        writer.writerow({
            "name": employee["name"],
            "department": employee["department"],
            "salary": f"{employee['salary']:.2f}"
        })

print("\nemployees_updated.csv created successfully.")

print("\nUpdated employee records:")

with open("employees_updated.csv", "r", newline="") as f:
    reader = csv.DictReader(f)

    for row in reader:
        print(dict(row))
```

```python
def safe_divide(a, b):
    try:
        return a / b

    except ZeroDivisionError:
        print("Warning: Cannot divide by zero.")
        return None

    except TypeError:
        print("Warning: Invalid types for division.")
        return None

    finally:
        print("Operation attempted")


print(safe_divide(10, 2))
print(safe_divide(10, 0))
print(safe_divide("10", 2))
```

```python
import csv
import json

orders = [
    ["order_id", "customer", "amount", "status"],
    ["1001", "Amit", "250.50", "shipped"],
    ["1002", "Reni", "120.00", "pending"],
    ["1003", "Tara", "99.99", "shipped"],
    ["1004", "Amit", "bad_amount", "cancelled"],  # invalid amount
    ["1005", "Sam", "450.00", "delivered"],
    ["1006", "Reni", "", "pending"],              # missing amount
    ["1007", "Priya", "75.25", "delivered"],
    ["1008", "Tara", "310.00", "shipped"]
]

with open("orders.csv", "w", newline="") as f:
    writer = csv.writer(f)
    writer.writerows(orders)

print("orders.csv created.")


try:
    total_by_customer = {}
    statuses = set()
    skipped_rows = 0

    with open("orders.csv", "r", newline="") as f:
        reader = csv.DictReader(f)

        for row_number, row in enumerate(reader, start=2):
            customer = row["customer"]

            try:
                amount = float(row["amount"])

            except (ValueError, TypeError):
                print(
                    f"Warning: skipping row {row_number} "
                    f"for customer {customer} because amount is invalid."
                )
                skipped_rows += 1
                continue

            # Add the amount to the customer's running total
            total_by_customer[customer] = (
                total_by_customer.get(customer, 0) + amount
            )

            # Add the status to the set
            statuses.add(row["status"])


    summary = {
        "total_by_customer": {
            customer: round(total, 2)
            for customer, total in total_by_customer.items()
        },
        "statuses": sorted(statuses),
        "skipped_rows": skipped_rows
    }


    with open("orders_summary.json", "w") as f:
        json.dump(summary, f, indent=2)

    print("orders_summary.json created.")
    print("\nFinal summary:")
    print(summary)


except FileNotFoundError:
    print("Error: orders.csv was not found.")

except Exception as e:
    print("Unexpected error:", e)

finally:
    print("Conversion attempt finished")
```
