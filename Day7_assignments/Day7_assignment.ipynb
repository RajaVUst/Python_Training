```python
# Collections in Practice
# Task 2.1 — Student Grade Book
students = {
    "alice": [88, 92, 79],
    "bob":   [65, 70, 68],
    "carol": [95, 91, 89],
}
res_dict = {i:sum(j)/len(j) if j else 0 for i,j in students.items() }
print(res_dict)
res_set = set()
for i,j in students.items():
    f = True
    for s in j:
        if s<=80:
            f = False
            break
    if f:
        res_set.add(i)
print(res_set)
l = sorted(res_dict.items(), key=lambda x: x[1], reverse=True)
print(l)
```

```python
import csv
employees = [
    ["Vinaya", "Engineering", 70000],
    ["Bob", "HR", 50000],
    ["Varshith", "Engineering", 80000],
    ["Aishuu", "Finance", 60000],
    ["Prasana", "HR", 55000]
]

with open("employees.csv", "w", newline="") as file:
    writer = csv.writer(file)
    writer.writerow(["name", "department", "salary"])
    writer.writerows(employees)

print("employees.csv created.\n")
print("Employee Records:")
employee_data = []

with open("employees.csv", "r") as file:
    reader = csv.DictReader(file)
    for row in reader:
        row["salary"] = float(row["salary"])
        employee_data.append(row)
        print(row)

department_totals = {}

for emp in employee_data:
    dept = emp["department"]
    salary = emp["salary"]

    department_totals[dept] = department_totals.get(dept, 0) + salary

print("\nTotal Salary Per Department:")
for dept, total in department_totals.items():
    print(f"{dept}: {total}")

for emp in employee_data:
    if emp["department"] == "Engineering":
        emp["salary"] *= 1.10

with open("employees_updated.csv", "w", newline="") as file:
    fieldnames = ["name", "department", "salary"]
    writer = csv.DictWriter(file, fieldnames=fieldnames)

    writer.writeheader()

    for emp in employee_data:
        writer.writerow({
            "name": emp["name"],
            "department": emp["department"],
            "salary": round(emp["salary"], 2)
        })

print("\nemployees_updated.csv created with Engineering salaries increased by 10%.")
```

```python
#Task 4.1 — try/except/finally Basics
def safe_divide(a, b):
    try:
        return a / b
    except ZeroDivisionError:
        print("Warning: Cannot divide by zero.")
        return None

    except TypeError:
        print("Warning: Invalid data type. Both arguments must be numbers.")
        return None

    finally:
        print("Operation attempted")

print(safe_divide(10, 2))
print(safe_divide(10, 0))
print(safe_divide("10", 2))
```
