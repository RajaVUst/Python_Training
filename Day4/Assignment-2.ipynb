```python
#Exercise -16

# Base case: when n == 1, return 1
# Recursive case: return n + sum_upto(n - 1)
 
def sum_upto(n):
    if n == 1:
        return 1
    return n + sum_upto(n - 1)
print(sum_upto(5))
print(sum_upto(1))
```

```python
#Exercise -17

def is_palindrome(word):
    word = word.lower()
    if word == word[::-1]:
        return True
    else:
        return False 
print(is_palindrome("level"))
print(is_palindrome("python"))
print(is_palindrome("Madam"))
```

```python
#Exercise -18

def grade_from_score(score):
    if score >= 90:
        return "A"
    elif score >= 80:
        return "B"
    elif score >= 70:
        return "C"
    elif score >= 60:
        return "D"
    else:
        return "F"
scores = [95, 82, 61, 40, 100] 
for score in scores:
    print(f"{score} -> {grade_from_score(score)}")
```

```python
#Exercise -19

def count_vowels(text):
    count = 0
    for ch in text:
        if ch.lower() in "aeiou":
            count += 1
    return count
print(count_vowels("Python Bootcamp"))
print(count_vowels("I am learning Python language"))
```

```python
#Exercise -20

def fizzbuzz_range(start, end):
    for i in range(start, end + 1):
        if i % 15 == 0:
            print("FizzBuzz")
        elif i % 3 == 0:
            print("Fizz")
        elif i % 5 == 0:
            print("Buzz")
        else:
            print(i)
fizzbuzz_range(1, 20)
```

```python
#Exercise -21

def is_prime(n):
    if n <= 1:
        return False
    for i in range(2, n):
        if n % i == 0:
            return False
    return True
res=is_prime(3)
print(res)
```

```python
#Exercise -22

def password_strength(password):
    if len(password) < 8:
        return "Weak"
    has_digit = False
    has_upper = False
    for ch in password:
        if ch.isdigit():
            has_digit = True
        if ch.isupper():
            has_upper = True
    if has_digit and has_upper:
        return "Strong"
    elif has_digit:
        return "Medium"
    else:
        return "Weak" 
print(password_strength("abc"))
print(password_strength("python123"))
print(password_strength("Python123"))
 
```

```python
#Exercise -23

def build_utilities(operation, *values):
    if operation == "sum":
        return sum(values)
    elif operation == "max":
        return max(values)
    elif operation == "min":
        return min(values)
    elif operation == "average":
        return sum(values) / len(values)
    else:
        return "Invalid operation" 
print(build_utilities("sum", 4, 8, 15))
print(build_utilities("average", 2, 4, 6, 8))
```
