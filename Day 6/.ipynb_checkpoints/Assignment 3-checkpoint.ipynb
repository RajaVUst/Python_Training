```python
## Exercise 30: Draft a mini log parser
def parse_log(path):
    parsed_logs = []
    skipped = 0

    with open(path, "r") as f:
        for line in f:
            parts = line.strip().split(maxsplit=2)

            if len(parts) != 3:
                skipped += 1
                continue

            timestamp = parts[0] + " " + parts[1]
            level = parts[2].split(maxsplit=1)[0]
            message = parts[2].split(maxsplit=1)[1]

            parsed_logs.append({
                "timestamp": timestamp,
                "level": level,
                "message": message
            })

    return parsed_logs, skipped
logs, skipped = parse_log("access.log")

print("Successfully parsed:", len(logs))
print("Skipped:", skipped)

for log in logs:
    print(log)
```
