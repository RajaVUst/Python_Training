```python
# exercise 23

def check_age(age):
    if age < 0:
        raise ValueError("Age cannot be negative.")
    
    return age

# Test with a valid value
try:
    age = check_age(5)
    print("Valid age:", age)

except ValueError as e:
    print("Error:", e)

# Test with an invalid value
try:
    age = check_age(-5)
    print("Valid age:", age)

except ValueError as e:
    print("Error:", e)
```

```python
# exercise 24

class InvalidScoreError(Exception):
    pass


def validate_score(score):
    if score < 0 or score > 100:
        raise InvalidScoreError("Score must be between 0 and 100.")
    
    return score

# Test with a valid score
try:
    score = validate_score(85)
    print("Valid score:", score)

except InvalidScoreError as e:
    print("Error:", e)


# Test with an invalid score
try:
    score = validate_score(120)
    print("Valid score:", score)

except InvalidScoreError as e:
    print("Error:", e)
```

```python
# exercise 25

scores = [85, 120, 72, -5, 95]

for score in scores:
    try:
        valid_score = validate_score(score)
        print("Valid score:", valid_score)

    except InvalidScoreError as e:
        print("Invalid score:", score, "-", e)
```

```python
# exercise 26

import csv


def load_scores(path):
    scores = {}

    try:
        with open(path, "r", newline="") as f:
            reader = csv.DictReader(f)

            for row in reader:
                try:
                    score = int(row["score"])
                    scores[row["name"]] = score

                except ValueError:
                    continue

    except FileNotFoundError:
        return {}

    return scores

path = r"C:\Users\308333\Documents\Python_Training\Day 6\scores.csv"
scores = load_scores(path)
print(scores)
```

```python
# exercise 27

def summarise(scores):
    values = scores.values()

    return {
        "count": len(scores),
        "average": sum(values) / len(scores),
        "highest": max(values),
        "lowest": min(values)
    }
    
summary = summarise(scores)
print(summary)
```

```python
# exercise 28

def word_frequencies(path):
    frequencies = {}

    try:
        with open(path, "r") as f:
            for line in f:
                words = line.lower().split()

                for word in words:
                    frequencies[word] = frequencies.get(word, 0) + 1

    except FileNotFoundError:
        return {}

    return frequencies

path = r"C:\Users\308333\Documents\Python_Training\Day 6\notes.txt"
frequencies = word_frequencies(path)
print(frequencies)
```

```python
# exercise 29

class MissingConfigKeyError(Exception):
    pass


def load_config(path):
    with open(path, "r") as f:
        data = json.load(f)

    if "batch_size" not in data:
        raise MissingConfigKeyError("Missing required key: batch_size")

    if "learning_rate" not in data:
        raise MissingConfigKeyError("Missing required key: learning_rate")

    return data

config_path = r"C:\Users\308333\Documents\Python_Training\Day 6\config.json"

try:
    config = load_config(config_path)
    print("\nConfiguration:")
    print(config)

except MissingConfigKeyError as e:
    print("\nConfiguration error:", e)
```

```python
# exercise 30

def parse_log(path):
    parsed_logs = []
    skipped = 0

    try:
        with open(path, "r") as f:

            for line in f:
                clean_line = line.strip()

                parts = clean_line.split(maxsplit=3)

                if len(parts) != 4:
                    skipped += 1
                    continue

                timestamp = parts[0] + " " + parts[1]
                level = parts[2]
                message = parts[3]

                parsed_logs.append({
                    "timestamp": timestamp,
                    "level": level,
                    "message": message
                })

    except FileNotFoundError:
        print("Log file not found.")
        return [], 0

    return parsed_logs, skipped


log_path = r"C:\Users\308333\Documents\Python_Training\Day 6\access.log"

logs, skipped = parse_log(log_path)

print("Successfully parsed:", len(logs))
print("Skipped:", skipped)

for log in logs:
    print(log)
```
