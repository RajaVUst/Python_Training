```python
#Part A — Lists
#Exercise 1: Shopping List Manager
cart = ['Sugar','Tea Powder', 'rava']
cart.append('chocolate')
cart.append('toothpaste')
print(cart)
cart.remove('rava')
print(cart)
print(sorted(cart))

```

```python
#Exercise 2: Rotate the Queue
queue = ["Amit", "Reni", "Tara", "Sam"]
item = queue.pop(0)
print(queue)

queue.append(item)
print(queue)

```

```python
#Exercise 3: Running Total With Slicing
readings = [12, 15, 9, 22, 30, 4, 18]
first_three_readings = readings[:3]
lst_two_reading = readings[-2:]
every_second_reading = readings[::2]
print(first_three_readings)
print(lst_two_reading)
print(every_second_reading)

```

```python
#Exercise 4: Copy vs. Reference
original = [1,2,3]
alias = original
safe_copy = original.copy()
alias.append(100)
safe_copy.append(200)
print(original)
print(alias)
print(safe_copy)
```

```python
#Part B — Tuples
#Exercise 5: Coordinates
location = (12.97, 77.59) 
location[0]=13.78
print(location)
```

```python
#Exercise 6: Unpacking a Tuple
record = ("Reni", 91, "Cohort 2")
name, score, cohort = record
print(f"Name -> {name} | Cohort -> {cohort} | score -> {score}")
```

```python
#Exercise 7: Multiple Return Values
def stats(numbers):
    return (min(numbers), max(numbers), ((sum(numbers))/len(numbers)))
print(stats([4, 9, 1, 7, 15] ))
```

```python
#Part C — Sets
#Exercise 8: De-duplicating Attendance
attendance = ["amit", "reni", "amit", "tara", "reni", "sam"]
unique_attendees = set(attendance)
print(f"No. of people attended : {len(unique_attendees)}") 
```

```python
#Exercise 9: Common Interests
cohort_a = {"python", "sql", "excel"} 
cohort_b = {"python", "tableau", "excel"}
print(f"Skills common to both cohorts : {cohort_a&cohort_b}")
print(f"Skills Unique to cohort_a : {cohort_a-cohort_b}")
print(f"Skills across both cohorts combined : {cohort_a|cohort_b}")
```

```python
#Exercise 10: Fast Membership Check
seen = set(range(1000))
if 999 in seen:
    print("1000 is in seen")
```

```python
#Part D — Dictionaries & Nested Structures
#Exercise 11: Contact Card
contact = {'name':'Vinaya', 'email': 'vinaya@gmail.com', 'phone':9346019595}
print(contact['name'])
print(contact['email'])
print(contact['phone'])
```

```python
#Exercise 12: Word Frequency Counter
text =  "the quick brown fox jumps over the lazy dog the fox runs"
words = text.split()
word_counts ={}
for word in words:
    word_counts[word] = word_counts.get(word,0)+1
print(word_counts)
    
```
