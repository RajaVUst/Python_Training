```python
#Part F — Cumulative Challenges 
#Exercise 19: Vowel Counter by Word
def vowel_counts(sentence):
    #sentence = sentence.lower()
    #print(sentence)
    words = sentence.split()
    vowel_counter ={}
    for word in words:
        vowel_counter[word]=sum(1 for char in word.lower() if char in "aeiou")
        
    print(vowel_counter)
    
vowel_counts("The Quick Brown Fox Jumps")
```

```python
#Exercise 20: Grade Book Summary
def grade_summary(roster):
    summary = []
    for student in roster:
        avg = sum(student["scores"]) / len(student["scores"])
        if avg >= 90:
            grade = "A"
        elif avg >= 80:
            grade = "B"
        elif avg >= 70:
            grade = "C"
        elif avg >= 60:
            grade = "D"
        else:
            grade = "F"
        summary.append({
            "name": student["name"],
            "average": round(avg, 2),
            "grade": grade
        })
    return summary
roster = [ {'name': 'Vinaya', 'scores': [90,92,83] }, {'name': 'Varshith', 'scores': [99,87,83]}, {'name': 'Vaishu', 'scores': [91,92,83]}]
result = grade_summary(roster)
for student in result:
    print(student)



```

```python
#Exercise 21: Unique Word Finder
def unique_words(text):
    text = text.lower()
    unique=set(text.split())
    return sorted(unique)
print(unique_words("hi Hi hi Vinaya This is Vinaya"))
```

```python
#Exercise 22: FizzBuzz, Structured
def fizzbuzz_map(n):
    dictionary = {}
    for i in range(1,n+1):
        if i%3==0:
            if i%5==0:
                dictionary[i]="FizzBuzz"
            else:
                dictionary[i]="Fizz"
        elif i%5==0:
            dictionary[i]="Buzz"
        else:
            dictionary[i]=i
    return dictionary
print(fizzbuzz_map(20))
        
```

```python
#Exercise 23: Password Strength Report (Capstone)
def password_report(passwords):
    pwd_category ={}
    for password in passwords:
        has_digit = any(char.isdigit() for char in password)
        if len(password)>=10 and has_digit:
            pwd_category[password]="Strong"
        elif len(password)>=6:
            pwd_category[password]="Medium"
        else:
            pwd_category[password]="Weak"
    return pwd_category
print(password_report(["vinaya","vinaya@123","vin123#$56","123456"]))

        
```
