```python
#Exercise 25: Catch custom exception

class InvalidScoreError(Exception):
    pass
def validate_score(score):
    if score < 0 or score > 100:
        raise InvalidScoreError("Score must be between 0 and 100.")
    return score
scores = [85, 120, 75, -10, 95]
for score in scores:
    try:
        validate_score(score)
        print(score, "is valid.")
    except InvalidScoreError as e:
        print(score, "is invalid:", e)
```

```python
#Exercise 26: Safe score loader

import csv
def load_scores(path):
    scores = {}
    try:
        with open(path, "r") as f:
            reader = csv.DictReader(f)
            for row in reader:
                try:
                    score = int(row["score"])
                    scores[row["name"]] = score
                except (ValueError, TypeError):
                    continue
    except FileNotFoundError:
        return {}
    return scores
result = load_scores("scores.csv")
print(result)
```

```python
#Exercise 27: Validate and summarise

def summarise(scores):
    if not scores:
        return {
            "count": 0,
            "average": 0,
            "highest": None,
            "lowest": None
        }
    values = scores.values()
    result = {
        "count": len(scores),
        "average": sum(values) / len(scores),
        "highest": max(values),
        "lowest": min(values)
    }
    return result
scores = {
    "Asha": 85,
    "Rahul": 92,
    "Priya": 105,
    "Kiran": 78
}
summary = summarise(scores)
print(summary)
```

```python
#Exercise 28: Word frequency counter

def word_frequencies(path):
    frequencies = {}
    with open(path, "r") as f:
        for line in f:
            words = line.lower().split()
            for word in words:
                frequencies[word] = frequencies.get(word, 0) + 1
    return frequencies
result = word_frequencies("notes.txt")
print(result)
```

```python
#Exercise 30: Mini Log Parser

#Step 1: Create access.log
with open("access.log", "w") as f:

    f.write("2026-03-23 10:02 ERROR Connection timeout\n")
    f.write("2026-03-23 10:05 INFO User logged in\n")
    f.write("2026-03-23 10:10 WARNING Disk space low\n")
    f.write("2026-03-23 ERROR\n")
    f.write("2026-03-23 10:20 INFO File uploaded successfully\n")
    f.write("2026-03-23 10:25 ERROR Database connection failed\n")
    f.write("2026-03-23 10:30 INFO User logged out\n")

print("access.log created.")

#Step 2: Create parse_log()

def parse_log(path):
parsed_logs = []
    skipped_lines = 0
    try:
        with open(path, "r") as f:
            for line in f:
                line = line.strip()
                parts = line.split(maxsplit=2)
                if len(parts) != 3:
                    skipped_lines += 1
                    continue
                timestamp = parts[0] + " " + parts[1]
                level = parts[2].split()[0]
                message = " ".join(parts[2].split()[1:])
                parsed_logs.append({
                    "timestamp": timestamp,
                    "level": level,
                    "message": message
                })
    except FileNotFoundError:
        print("Log file not found.")
        return []
    return parsed_logs, skipped_lines

#Step 3: Call the function

logs, skipped = parse_log("access.log")
print("Successfully parsed:", len(logs))
print("Skipped:", skipped)
print("\nParsed logs:")
for log in logs:
    print(log)
```
