```python
# Part F: Raising & Creating Custom Exceptions

# Exercise 23: Raise a built-in exception
# Prompt: Write a function check_age(age) that raises ValueError if age is negative, and otherwise returns the age unchanged. 
# Test it with a valid and an invalid value, catching the error where you call it.
# Hint: raise ValueError("some message") stops the function immediately.

def check_age(age):
    try:
        if age<0:
            raise ValueError('Age is negative')
        return age
    except ValueError:
        print('Age is negative')
    
check_age(-2)
check_age(30)
```

```python
# Exercise 24: Define a custom exception
# Prompt: Define class InvalidScoreError(Exception): pass, then write a function validate_score(score) that raises 
# InvalidScoreError if score is not between 0 and 100 inclusive, and returns it otherwise.
# Hint: This is purely a naming pattern for today -- no attributes or methods needed on the class.

class InvalidScoreError(Exception): pass

def validate_score(score):
    if score<0 or score>100:
        raise InvalidScoreError('Invalid Score')
    return score

validate_score(101)
```
