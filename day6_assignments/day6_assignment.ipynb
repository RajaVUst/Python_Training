```python
# 1
with open("notes.txt", "r") as f:
    text = f.read()
    c = 0
    for i in text:
        if i == ' ':
            continue
        c+=len(i)
    print(c)
```

```python
# 8
with open("squares.txt", "w") as f:
    for i in range(1,11):
        val = i**2
        f.write(f"{val} \n")

with open("squares.txt", "r") as f:
    text = f.readlines()
    sum = 0
    for i in text:
        sum+=int(i)
    print(sum)
```
