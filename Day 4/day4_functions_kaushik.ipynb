```python
# Task 18
def get_grade(marks):
    if marks>=90:
        return 'A'
    elif marks>=80:
        return 'B'
    elif marks>=70:
        return 'C'
    elif marks>=60:
        return 'D'
    else:
        return 'F'
scores = [95, 82, 61, 40, 100] 
for i in scores:
    print(f"for {i} marks we will get {get_grade(i)} grade")
```

```python
# Task 19
def count_vowels(s):
    d = 0
    for i in s:
        if i == 'a' or i=='e'or i=='i' or i=='o' or i=='u':
            d+=1
    return d
print(count_vowels("Python Bootcamp"))
print(count_vowels("kaushik"))
```

```python
# Task 20
def fizzbuzz_range(start, end):
    for i in range(start,end):
        if i%3 == 0:
            print("fizz",end="")
        if i%5 == 0:
            print("buzz",end="")
        if i%3 == 0 or i%5 == 0:
            print("")
fizzbuzz_range(1,20)
```

```python
# Task 21
def is_prime(n):
    for i in range(2,n):
        if n%i == 0:
            return False
    return True
def primes_up_to(limit):
    l = []
    for i in range(2,limit+1):
        if is_prime(i):
            l.append(i)
    return l

print(primes_up_to(30))
```

```python
# Task 22

def password_strength(password):
    d =0
    u =0
    s =0
    l =0
    if len(password) < 8:
        print("weak")
    else:
        for i in password:
            if ord(i) >=48 and ord(i)<=57:
                d+=1
            elif ord(i) >=ord('A') and ord(i) <=ord('Z'):
                u+=1
            elif ord(i) >=ord('a') and ord(i) <=ord('z'):
                l+=1
            else:
                s+=1
        if d < 1 or s<1 or l<1 or u<1:
            print("medium")
        else:
            print("strong")

password_strength("kaushik")   
password_strength("kaushikkk")   
password_strength("kaushi1@K")   
```

```python
# Task 23
def build_utilities(operation, *values):
    if operation == "sum":
        print(sum(values))
    elif operation == "max":
        print(max(values))
    elif operation == "min":
        print(min(values))
    else:
        print("invalid operation")
```
