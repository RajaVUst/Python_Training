```python
# Declare variables using Python's primitive data types.
name = "Chaila"
age = 23
height = 1.51
has_coded_before = False
```

```python
# Print the type of each variable to verify Python's type inference.
print(type(name))
print(type(age))
print(type(height))
print(type(has_coded_before))
```

```python
# Ask the user for their favourite number; input() returns a string.
fav_str = input("Enter your favourite number: ")
```

```python
# Convert the string to an integer and print its value multiplied by 2.
fav_num = int(fav_str)
print(fav_num * 2)
```

```python
# Combine multiple variables into one readable sentence using an f-string.
print(f"{name} is {age} years old and {height}m tall")
```
