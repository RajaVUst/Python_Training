```python
#Exercise 6

total_bill = float(input("Enter the total bill amount: "))
people = int(input("Enter the number of people: "))
wants_tip = (input("Do you want to add a 10% tip? (yes/no): ").lower() == "yes")
tip = total_bill * 0.10 * wants_tip
final_bill = total_bill + tip
share = final_bill / people
print(f"Each person's share: {share:.2f}")
```

```python
#Exericise 7
a = 5
b = 10

# Method 1: Swap using a temporary variable.
temp = a
a = b
b = temp

# Print the swapped values.
print("Method 1:")
print(f"a = {a}, b = {b}")

# Reset the variables for the second method.
a = 5
b = 10

# Method 2: Swap using Python's multiple-assignment shortcut.
a, b = b, a

# Print the swapped values.
print("Method 2:")
print(f"a = {a}, b = {b}")
```

```python
#Exercise 8
# Ask the user for the rectangle's length and width.
length = float(input("Enter the length: "))
width = float(input("Enter the width: "))

# Calculate the area and perimeter.
area = length * width
perimeter = 2 * (length + width)

# Print the area and perimeter on separate labelled lines.
print(f"Area: {area}")
print(f"Perimeter: {perimeter}")
```

```python
#Exercise 11
# Ask the user for weight and height.
weight = float(input("Enter your weight in kilograms: "))
height = float(input("Enter your height in metres: "))

# Calculate BMI using the given formula.
bmi = weight / (height ** 2)

# Print the BMI rounded to one decimal place.
print(f"BMI: {bmi:.1f}")
```

```python
#Exercise 12
# Store the number to check.
n = 15

# Check whether the number is divisible by 3 and 5.
div_by_3 = (n % 3 == 0)
div_by_5 = (n % 5 == 0)

# Check whether the number is divisible by both 3 and 5.
div_by_both = div_by_3 and div_by_5

# Print the results.
print(f"Divisible by 3: {div_by_3}")
print(f"Divisible by 5: {div_by_5}")
print(f"Divisible by both: {div_by_both}")
```

```python
#Exercise 13
# Ask the user for the USD amount and conversion rate.
amount_usd = float(input("Enter amount in US Dollars: "))
conversion_rate = float(input("Enter USD to INR conversion rate: "))

# Convert the amount from US Dollars to Indian Rupees.
amount_inr = amount_usd * conversion_rate

# Print the converted amount with commas and two decimal places.
print(f"₹{amount_inr:,.2f}")
```

```python
#Exercise 14
# Ask the user for three test scores.
score_1 = float(input("Enter first test score: "))
score_2 = float(input("Enter second test score: "))
score_3 = float(input("Enter third test score: "))

# Calculate the average of the three scores.
average = (score_1 + score_2 + score_3) / 3

# Check whether the average is 40 or above.
has_passed = (average >= 40)

# Print the student report with the average rounded to one decimal place.
print(f"Average: {average:.1f} | Passed: {has_passed}")
```
