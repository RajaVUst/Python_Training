```python
# Restaurant Bill Splitter
# Getting input
tot = float(input('Enter the Total Bill Amount : '))
n = int(input('Enter the Number Of People Splitting : '))
tip = input('Want to add Tip (yes/no) ? : ')
```

```python
# Splitting Bill and displaying it
decision = tip == 'yes'
if decision:
    final = tot * 1.10
else:
    final = tot
print(f'Final Bill : {final}\nEach Person Share : {final/n}')
```

```python
# Part D — Additional Practice
# Variable Swap
# variable Declarations 
a = 10
b = 20
```

```python
print(f'Before Swap a : {a}, b : {b}')

# Method 1: temporary variable
temp = a
a = b
b = temp
print(f'First Swap a : {a}, b : {b}')

# Method 2: Python shortcut
a,b = b,a
print(f'Second Swap a : {a}, b : {b}')
```

```python
# Rectangle Calculator  
# Getting Input
length = input('Enter the Length of the Rectangle : ')
width = input('Enter the width of the Rectangle : ')
```

```python
#Caluculating and Printing the result
area = float(length) * float(width)
perimeter = 2 * (float(length) + float(width))
print(f'The Area of Rectangle is : {area}\nThe Perimeter of Rectangle is : {perimeter}')
```
