```python
# Compare Printing vs Returning in an Even Number Check

def is_even_print(n):
    if n % 2 == 0:
        print("Even")
    else:
        print("Odd")

def is_even_return(n):
    return n % 2 == 0

is_even_print(7)

result = is_even_return(7)
print(result)

if result:
    print("The number is even.")
else:
    print("The number is odd.")
```

```python
# Write a Function with Multiple Default Parameters

def book_ticket(passenger, seat_type="Economy", meal="Veg"):
    print(f"Passenger: {passenger}, Seat Type: {seat_type}, Meal: {meal}")

# (a) Only the passenger name
book_ticket("Deepshika")

# (b) Override just meal using a keyword argument
book_ticket("Anjali", meal="Non-Veg")

# (c) Override both defaults using keyword arguments in a different order
book_ticket("Rahul", meal="Vegan", seat_type="Business")
```

```python
# Write a Function Using **kwargs to Print Student Information

def print_student_info(**details):
    for key, value in details.items():
        print(f"{key}: {value}")

print_student_info(
    name="Deepshika",
    age=21,
    course="Python Programming",
    city="Trivandrum"
)
```

```python
#  Demonstrate Local Variable Scope

def reset_score():
    score = 0
    print("Score has been reset.")

reset_score()

# This will cause an error because 'score' only exists inside the function
print(score)

# Explanation:
# The variable 'score' is a local variable, which means it is created
# inside the function and can only be accessed within that function.
# Trying to use it outside the function results in a NameError because
# Python cannot find a variable named 'score' in the global scope.
```

```python
# Use a Global Variable and the global Keyword

total_attempts = 0

def log_attempt():
    global total_attempts
    total_attempts += 1

log_attempt()
log_attempt()
log_attempt()

print(total_attempts)
```

```python
# Write a Recursive Countdown Function

def countdown(n):
    if n == 0:  # Base case
        print("Liftoff!")
    else:
        print(n)
        countdown(n - 1)

countdown(5)
```

```python
#  Write a Recursive Function to Find the Sum from 1 to n

# Base case:
# If n is 1, return 1.

# Recursive case:
# Return n plus the sum of all numbers from 1 to (n - 1).

def sum_upto(n):
    if n == 1:
        return 1
    return n + sum_upto(n - 1)

print(sum_upto(5))
print(sum_upto(1))
```

```python
#  Check if a Word is a Palindrome

def is_palindrome(word):
    word = word.lower()  # Convert to lowercase so case does not matter
    return word == word[::-1]

print(is_palindrome("level"))
print(is_palindrome("python"))
print(is_palindrome("Madam"))
```

```python
#  Return a Letter Grade Based on a Score

def grade_from_score(score):
    if score >= 90:
        return 'A'
    elif score >= 80:
        return 'B'
    elif score >= 70:
        return 'C'
    elif score >= 60:
        return 'D'
    else:
        return 'F'

scores = [95, 82, 61, 40, 100]

for score in scores:
    print(f"{score} -> {grade_from_score(score)}")
```

```python
#  Count the Number of Vowels in a String

def count_vowels(text):
    count = 0
    vowels = "aeiou"

    for char in text.lower():
        if char in vowels:
            count += 1

    return count

print(count_vowels("Python Bootcamp"))
print(count_vowels("Learning Python is fun!"))
```

```python
# Create a Reusable FizzBuzz Function

def fizzbuzz_range(start, end):
    for num in range(start, end + 1):
        if num % 3 == 0 and num % 5 == 0:
            print("FizzBuzz")
        elif num % 3 == 0:
            print("Fizz")
        elif num % 5 == 0:
            print("Buzz")
        else:
            print(num)

fizzbuzz_range(1, 15)
```

```python
#  Find Prime Numbers Using Functions

def is_prime(n):
    if n < 2:
        return False

    for i in range(2, n):
        if n % i == 0:
            return False

    return True

def primes_up_to(limit):
    primes = []

    for num in range(2, limit + 1):
        if is_prime(num):
            primes.append(num)

    return primes

print(primes_up_to(30))
```

```python
#  Check Password Strength Using String Methods and Conditionals

def password_strength(password):
    has_upper = False
    has_digit = False

    for char in password:
        if char.isupper():
            has_upper = True
        if char.isdigit():
            has_digit = True

    if len(password) >= 8 and has_upper and has_digit:
        return "Strong"
    elif len(password) >= 6 and (has_upper or has_digit):
        return "Medium"
    else:
        return "Weak"

print(password_strength("hello"))
print(password_strength("Hello12"))
print(password_strength("Python123"))
```

```python
#  Create a Mini Calculator Using *args and Conditionals

def build_utilities(operation, *values):
    if operation == "sum":
        return sum(values)
    elif operation == "max":
        return max(values)
    elif operation == "min":
        return min(values)
    elif operation == "average":
        return sum(values) / len(values)
    else:
        return "Invalid operation"

print(build_utilities("sum", 4, 8, 15))
print(build_utilities("average", 2, 4, 6, 8))
```
