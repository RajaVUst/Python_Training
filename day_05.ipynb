```python
#Exercise 4: Copy vs. Reference

original = [1, 2, 3]
alias = original
safe_copy = alias.copy()
alias.append(100)
safe_copy.append(200)
print(original, alias, safe_copy)
#Original only picked up one of the two changes - original and alias shares the same object references. so changes made in one will reflect in both. Changes made by soft_copy doesnt reflect on original because soft_copy created new object but shares nested object references

```

```python
#Exercise 12: Word Frequency Counter
text = "the quick brown fox jumps over the lazy dog the fox runs"
text=text.split(" ")
d={}
for i in text:
    if i not in d:
        d[i] = 1
    else:
        d[i]+=1
for k,v in d.items():
    print(k,'-->',v)
```

```python
#Exercise 19: Vowel Counter by Word
def vowel_counts(s):
    s=s.split(" ")
    d={}
    for i in s:
        c=0
        for j in i.lower():
            if j in "aeiou":
                c+=1
        d[i] = c
    return d
print(vowel_counts("The Quick Brown Fox Jumps"))
```

```python
#Exercise 20: Grade Book Summary
def grade_summary(roster):
    result = []
    for student in roster:
        name = student["name"]
        scores = student["scores"]
        average = sum(scores) / len(scores)
        if average >= 90:
            grade = "A"
        elif average >= 80:
            grade = "B"
        elif average >= 70:
            grade = "C"
        elif average >= 60:
            grade = "D"
        else:
            grade = "F"
        result.append({
            "name": name,
            "average": average,
            "grade": grade
        })
    return result
roster = [
    {"name": "Amit", "scores": [85, 90, 88]},
    {"name": "Reni", "scores": [95, 92, 98]},
    {"name": "Tara", "scores": [65, 70, 68]}
]

res = grade_summary(roster)
for i in res:
    print(i)
```

```python
#Exercise 21: Unique Word Finder
def unique_words(text):
    s=text.lower().split(" ")
    return sorted(set(s))
print(unique_words("The quick brown fox jumps over the lazy dog"))
```

```python
# Exercise 22: FizzBuzz, Structured
def fizzbuzz_map(n):
    result = {}
    for i in range(1, n + 1):
        if i % 3 == 0 and i % 5 == 0:
            result[i] = "FizzBuzz"
        elif i % 3 == 0:
            result[i] = "Fizz"
        elif i % 5 == 0:
            result[i] = "Buzz"
        else:
            result[i] = str(i)
    return result
print(fizzbuzz_map(20))
```

```python
#Exercise 23: Password Strength Report (Capstone)
def password_report(passwords):
    report = {
        "weak": [],
        "medium": [],
        "strong": []
    }
    for password in passwords:
        has_digit = any(char.isdigit() for char in password)
        if len(password) >= 10 and has_digit:
            report["strong"].append(password)
        elif len(password) >= 6:
            report["medium"].append(password)
        else:
            report["weak"].append(password)
    return report
passwords = ["cat","hello","python","welcome","hello123","SuperPassword1"]
print(password_report(passwords))
```
