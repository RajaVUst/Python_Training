# Day 2 — Strings

## Part A — Guided String Walkthrough

```python
# Create a sentence with leading and trailing spaces

sentence = "   Python IS Fun to Learn!!   "
```

```python
# Remove leading and trailing whitespace

clean = sentence.strip()
print(clean)
```

```python
# Print lowercase and uppercase versions

print(clean.lower())
print(clean.upper())
```

```python
# Print the first character, last character, and first five characters

print(clean[0])
print(clean[-1])
print(clean[:5])
```

```python
# Split the sentence into words and count them

words = clean.split()

print(words)
print(len(words))
```

```python
# Replace Fun with Powerful

print(clean.replace("Fun", "Powerful"))
```

```python
# Join the words using a dash

print("-".join(words))
```

```python
# Print a summary containing the word count and reversed sentence

print(f"Words: {len(words)} | Reversed: {clean[::-1]}")
```

## Part C — Mini-Exercises

```python
# Exercise 1: Name Tag Formatter

first_name = input("Enter your first name: ")
last_name = input("Enter your last name: ")

first_name = first_name.title()
last_name = last_name.title()

print(f"Hello, {first_name} {last_name}!")
```

```python
# Exercise 2: Coupon Code Slicer

code = "PYTHON2026SALE"

print(f"First 6 characters: {code[:6]}")
print(f"Last 4 characters: {code[-4:]}")
print(f"Reversed code: {code[::-1]}")
```

```python
# Exercise 3: Palindrome Checker

word = "level"

is_palindrome = (word == word[::-1])

print(is_palindrome)
```

```python
# Exercise 4: Email Splitter

email = "reni.k@company.com"

username, domain = email.split("@")

print(f"Username: {username}")
print(f"Domain: {domain}")
```

```python
# Exercise 5: Slug Normalizer

title = "  My First   Python Project!!  "

clean = title.strip()
print(clean)

clean = clean.replace("!!", "")
print(clean)

clean = clean.lower()
print(clean)

clean = clean.replace(" ", "_")
print(clean)
```

```python
# Exercise 6: Sentence Dashboard

sentence = input("Enter a sentence: ")

character_count = len(sentence)
word_count = len(sentence.split())

vowel_count = (sentence.count("a") + sentence.count("e") + sentence.count("i") + sentence.count("o") + sentence.count("u"))

longer_than_30 = len(sentence) > 30

print(f"Character count: {character_count}")
print(f"Word count: {word_count}")
print(f"Vowel count: {vowel_count}")
print(f"Longer than 30 characters: {longer_than_30}")
```
