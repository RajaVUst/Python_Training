```python
#Exercise 12: Sum of digits
n=int(input("Enter the number: "))#1234

sum=0
while n> 0:
    digit = n % 10
    sum = sum + digit
    n = n // 10

print("Sum of digits:", sum)

```

```python
#Exercise 17: Reverse an integer
n=int(input("Enter the number: "))#1234
rev=0
while(n>0):
    digit=n%10#1
    rev=rev*10+digit
    n//=10#
print(rev)
```

```python
#Exercise 18: Palindrome check for an integer
n=int(input("Enter the number: "))#1234
rev=0
dup=n
while(n>0):
    digit=n%10#1
    rev=rev*10+digit
    n//=10#
    
if dup==rev:
             print(f"{dup} is a Palindrome")
else:
        print(f"{dup} is Not a  Palindrome")
```

```python
#Exercise 19: Prime number check
n = int(input("Enter a number: "))

is_prime = True

for i in range(2, n):
    if n % i == 0:
        is_prime = False
        break

if is_prime:
    print("Prime number")
else:
    print("Not a prime number")
```

```python
#Exercise 20: Print primes in a range
for num in range(2, 51):
    is_prime = True

    for i in range(2, num):
        if num % i == 0:
            is_prime = False
            break

    if is_prime:
        print(num)
```

```python
#Exercise 23: Simple currency converter
amount = float(input("Enter amount in INR: "))
currency = input("Enter target currency (USD/EUR/GBP): ").upper()

if currency == "USD":
    converted = amount * 0.012
    print("Converted amount:", round(converted, 2), "USD")

elif currency == "EUR":
    converted = amount * 0.011
    print("Converted amount:", round(converted, 2), "EUR")

elif currency == "GBP":
    converted = amount * 0.009
    print("Converted amount:", round(converted, 2), "GBP")

else:
    print("Invalid currency")
```

```python
#Exercise 24: Number guessing game (fixed target)
secret_number = 25
guesses = [10, 30, 20, 25, 40]

for guess in guesses:
    if guess > secret_number:
        print("Too high")
    elif guess < secret_number:
        print("Too low")
    else:
        print("Correct!")
        break
```
