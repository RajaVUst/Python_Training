```python
n = int(input("Enter a positive whole number: "))

total = 0

while n > 0:
    digit = n % 10
    total += digit
    n = n // 10

print(f"Sum of digits: {total}")
```

```python
n = int(input("Enter a positive whole number: "))

reversed_n = 0

while n > 0:
    digit = n % 10
    reversed_n = reversed_n * 10 + digit
    n = n // 10

print(f"Reversed number: {reversed_n}")
```

```python
n = int(input("Enter a positive whole number: "))

original = n
reversed_n = 0

while n > 0:
    digit = n % 10
    reversed_n = reversed_n * 10 + digit
    n = n // 10

if original == reversed_n:
    print("Palindrome")
else:
    print("Not a palindrome")
```

```python
n = int(input("Enter a whole number greater than 1: "))

is_prime = True

for i in range(2, n):
    if n % i == 0:
        is_prime = False
        break

if is_prime:
    print("Prime")
else:
    print("Not prime")
```

```python
for n in range(2, 51):
    is_prime = True

    for i in range(2, n):
        if n % i == 0:
            is_prime = False
            break

    if is_prime:
        print(n)
```

```python
rupees = float(input("Enter amount in INR: "))
currency = input("Enter target currency (USD/EUR/GBP): ").upper()

if currency == "USD":
    converted = rupees / 83
    print(f"₹{rupees:.2f} = ${round(converted, 2)}")

elif currency == "EUR":
    converted = rupees / 90
    print(f"₹{rupees:.2f} = €{round(converted, 2)}")

elif currency == "GBP":
    converted = rupees / 105
    print(f"₹{rupees:.2f} = £{round(converted, 2)}")

else:
    print("Invalid currency. Please choose USD, EUR, or GBP.")
```
