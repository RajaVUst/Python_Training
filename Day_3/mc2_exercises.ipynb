```python
#Exercise 17

n = int(input("Enter a number: "))

reversed_n = 0

while n > 0:
    digit = n % 10
    reversed_n = reversed_n * 10 + digit
    n = n // 10

print(reversed_n)
```

```python
#Exercise 18

n = int(input("Enter a number: "))

original = n
reversed_n = 0

while n > 0:
    digit = n % 10
    reversed_n = reversed_n * 10 + digit
    n = n // 10

if original == reversed_n:
    print("Palindrome")
else:
    print("Not a palindrome")
```

```python
#Exercise 19

n = int(input("Enter a number: "))

is_prime = True

if n <= 1:
    is_prime = False
else:
    for i in range(2, n):
        if n % i == 0:
            is_prime = False
            break

if is_prime:
    print("Prime")
else:
    print("Not prime")
```

```python
#Exercise 20

for n in range(2, 51):

    is_prime = True

    for i in range(2, n):
        if n % i == 0:
            is_prime = False
            break

    if is_prime:
        print(n)
```

```python
#Exercise 23

amount = float(input("Enter amount in INR: "))
currency = input("Enter target currency (USD/EUR/GBP): ")

if currency == "USD":
    converted = amount / 90
    print(round(converted, 2))

elif currency == "EUR":
    converted = amount / 105
    print(round(converted, 2))

elif currency == "GBP":
    converted = amount / 120
    print(round(converted, 2))

else:
    print("Invalid currency")
```

```python
#Exercise 24

secret_number = 25

guesses = [10, 30, 15, 25, 40]

for guess in guesses:

    if guess > secret_number:
        print("Too high")

    elif guess < secret_number:
        print("Too low")

    else:
        print("Correct!")
        break
```
