```python
# Exercise 4: Copy vs. Reference

original = [1, 2, 3]
alias = original
safe_copy = original.copy()
alias.append(100)
safe_copy.append(200)
print("Original:", original)
print("Alias:", alias)
print("Safe copy:", safe_copy)
# Alias it points to the same list as original, but safe_copy is a separate list
```

```python

# Exercise 7: Multiple Return Values

def stats(numbers):
    low = min(numbers)
    high = max(numbers)
    avg = sum(numbers) / len(numbers)
    return low, high, avg
low, high, avg = stats([4, 9, 1, 7, 15])
print("Minimum:", low)
print("Maximum:", high)
print("Average:", avg)
```

```python
# Exercise 12: Word Frequency Counter

text = "the quick brown fox jumps over the lazy dog the fox runs"
words = text.split()
word_counts = {}
for word in words:
    word_counts[word] = word_counts.get(word, 0) + 1
print("Word counts:", word_counts)
```

```python
# Exercise 14: Lookup Table

grade_lookup = {
    "A": 90,
    "B": 80,
    "C": 70,
    "D": 60,
    "F": 0
}
score = 84
if score >= grade_lookup["A"]:
    grade = "A"
elif score >= grade_lookup["B"]:
    grade = "B"
elif score >= grade_lookup["C"]:
    grade = "C"
elif score >= grade_lookup["D"]:
    grade = "D"
else:
    grade = "F"
print("Grade:", grade)
```

```python
# Exercise 17: Dictionary From Two Lists

names = ["Amit", "Reni", "Tara"]
scores = [78, 91, 65]
scores_dict = {
    name: score
    for name, score in zip(names, scores)
}
print("Scores:", scores_dict)
```

```python
# Exercise 18: Filtering With a Dict Comprehension

passing_scores = {
    name: score
    for name, score in scores_dict.items()
    if score >= 70
}
print("Scores 70 or above:", passing_scores)
```

```python
# Exercise 19: Vowel Counter by Word

def vowel_counts(sentence):
    vowels = "aeiou"
    result = {}
    for word in sentence.lower().split():
        count = 0
        for char in word:
            if char in vowels:
                count += 1
        result[word] = count
    return result
print(vowel_counts("The Quick Brown Fox Jumps"))
```

```python
# Exercise 20: Grade Book Summary

def grade_summary(roster):
    result = []
    for student in roster:
        average = sum(student["scores"]) / len(student["scores"])
        if average >= 90:
            grade = "A"
        elif average >= 80:
            grade = "B"
        elif average >= 70:
            grade = "C"
        elif average >= 60:
            grade = "D"
        else:
            grade = "F"
        result.append({"name": student["name"],
                       "average": round(average, 1),
                       "grade": grade})
    return result
roster = [
    {"name": "Amit", "scores": [80, 75, 90]},
    {"name": "Reni", "scores": [91, 88, 95]},
    {"name": "Tara", "scores": [60, 65, 70]}
]
print(grade_summary(roster))
```

```python
# Exercise 21: Unique Word Finder

def unique_words(text):
    words = text.lower().split()
    unique = set(words)
    return sorted(unique)
print(unique_words("Python is easy and Python is useful"))
```

```python
# Exercise 22: FizzBuzz, Structured

def fizzbuzz_map(n):
    result = {}
    for i in range(1, n + 1):
        if i % 3 == 0 and i % 5 == 0:
            result[i] = "FizzBuzz"
        elif i % 3 == 0:
            result[i] = "Fizz"
        elif i % 5 == 0:
            result[i] = "Buzz"
        else:
            result[i] = str(i)
    return result
print(fizzbuzz_map(20))
```

```python
# Exercise 23: Password Strength Report

def password_report(passwords):
    result = {
        "weak": [],
        "medium": [],
        "strong": []
    }
    for pw in passwords:
        has_digit = any(ch.isdigit() for ch in pw)
        if len(pw) >= 10 and has_digit:
            result["strong"].append(pw)
        elif len(pw) >= 6:
            result["medium"].append(pw)
        else:
            result["weak"].append(pw)
    return result
passwords = [
    "cat",
    "hello1",
    "python",
    "welcome",
    "python1234",
    "MyPassword9"
]
print(password_report(passwords))
```
