```python
# Task 15: Recursive countdown(n)

def countdown(n):
    if n <= 0:
        print("Liftoff!")
        return
    print(n)
    countdown(n - 1)
countdown(5)
```

```python
# Task 16: Recursive sum_upto(n)

def sum_upto(n):
    if n == 1:
        return 1
    return n + sum_upto(n - 1)
print("sum_upto(5):", sum_upto(5))
print("sum_upto(1):", sum_upto(1))
```

```python
# PART E: CUMULATIVE CHALLENGES

# Task 17: is_palindrome(word)

def is_palindrome(word):
    word = word.lower()
    return word == word[::-1]
print("level:", is_palindrome("level"))
print("python:", is_palindrome("python"))
print("Madam:", is_palindrome("Madam"))
```

```python
# Task 18: grade_from_score(score)

def grade_from_score(score):
    if score >= 90:
        return "A"
    elif score >= 80:
        return "B"
    elif score >= 70:
        return "C"
    elif score >= 60:
        return "D"
    else:
        return "F"
scores = [95, 82, 61, 40, 100]
for score in scores:
    grade = grade_from_score(score)
    print(f"{score} -> {grade}")
```

```python
# Task 19: count_vowels(text)

def count_vowels(text):
    vowels = "aeiou"
    count = 0
    for character in text.lower():
        if character in vowels:
            count += 1
        return count
print(
    "Vowels in 'Python Bootcamp':",
    count_vowels("Python Bootcamp")
)
print(
    "Vowels in sentence:",
    count_vowels("Python is easy to learn")
)

```

```python
# Task 20: fizzbuzz_range(start, end)

def fizzbuzz_range(start, end):
    for number in range(start, end + 1):
        if number % 3 == 0 and number % 5 == 0:
            print("FizzBuzz")
        elif number % 3 == 0:
            print("Fizz")
        elif number % 5 == 0:
            print("Buzz")
        else:
            print(number)
fizzbuzz_range(1, 15)
```

```python
# Task 21: is_prime(n) and primes_up_to(limit)

def is_prime(n):
    if n < 2:
        return False
    for number in range(2, n):
        if n % number == 0:
            return False
    return True

def primes_up_to(limit):
    primes = []
    for number in range(2, limit + 1):
        if is_prime(number):
            primes.append(number)
    return primes
print("Primes up to 30:")
print(primes_up_to(30))
```

```python
# Task 22: password_strength(password)

def password_strength(password):
    has_digit = False
    has_uppercase = False
    for character in password:
        if character.isdigit():
            has_digit = True
        if character.isupper():
            has_uppercase = True
    if len(password) >= 8 and has_digit and has_uppercase:
        return "Strong"
    elif len(password) >= 6 and (has_digit or has_uppercase):
        return "Medium"
    else:
        return "Weak"
print("Password 1:", password_strength("abc"))
print("Password 2:", password_strength("python1"))
print("Password 3:", password_strength("Python123"))
```

```python
# Task 23: CAPSTONE - build_utilities()

def build_utilities(operation, *values):
    if operation == "sum":
        return sum(values)
    elif operation == "max":
        return max(values)
    elif operation == "min":
        return min(values)
    elif operation == "average":
        return sum(values) / len(values)
    else:
        return "Invalid operation"
print(
    'build_utilities("sum", 4, 8, 15):',
    build_utilities("sum", 4, 8, 15)
)
print(
    'build_utilities("max", 4, 8, 15):',
    build_utilities("max", 4, 8, 15)
)
print(
    'build_utilities("min", 4, 8, 15):',
    build_utilities("min", 4, 8, 15)
)
print(
    'build_utilities("average", 2, 4, 6, 8):',
    build_utilities("average", 2, 4, 6, 8)
)
```
