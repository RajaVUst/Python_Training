```python
### Task 1  —  Design a Class
```

```python
class TrainingRun:

    # Class attribute
    run_count = 0

    def __init__(self, model_name, learning_rate):
        # Instance attributes
        self.model_name = model_name
        self.learning_rate = learning_rate
        self.status = "Not Started"

        # Increase count whenever an object is created
        TrainingRun.run_count += 1

    # Instance method
    def start(self):
        self.status = "Running"
        print(f"{self.model_name} training started.")

    # Instance method
    def summary(self):
        print(f"Model: {self.model_name}")
        print(f"Learning Rate: {self.learning_rate}")
        print(f"Status: {self.status}")

    # Class method
    @classmethod
    def from_config(cls, config_dict):
        return cls(
            config_dict["model_name"],
            config_dict["learning_rate"]
        )


# Normal object creation
run1 = TrainingRun("ResNet", 0.001)

# Another object
run2 = TrainingRun("BERT", 0.0001)

# Start a run
run1.start()

# Show status
run1.summary()

print("Total runs:", TrainingRun.run_count)


# Creating an object using the class method
config = {
    "model_name": "GPT",
    "learning_rate": 0.0005
}

run3 = TrainingRun.from_config(config)

run3.start()
run3.summary()

print("Total runs:", TrainingRun.run_count)
```

```python
### Task 2  —  Encapsulation
```

```python
class TrainingRun:

    run_count = 0

    def __init__(self, model_name, learning_rate):
        self.model_name = model_name
        self.learning_rate = learning_rate

        # Protected attribute
        self._status = "pending"

        # Private/name-mangled attribute
        self.__api_key = "fake-api-key-12345"

        TrainingRun.run_count += 1

    def start(self):
        self.status = "running"
        print(f"{self.model_name} training started.")

    def summary(self):
        print(f"Model: {self.model_name}")
        print(f"Learning Rate: {self.learning_rate}")
        print(f"Status: {self.status}")

    @property
    def status(self):
        return self._status

    @status.setter
    def status(self, value):
        if value not in ("pending", "running", "done"):
            raise ValueError(
                "Status must be 'pending', 'running', or 'done'"
            )

        self._status = value

    @classmethod
    def from_config(cls, config_dict):
        return cls(
            config_dict["model_name"],
            config_dict["learning_rate"]
        )


# -------------------------
# Testing
# -------------------------

run = TrainingRun("ResNet", 0.001)

print(run.status)
# pending

run.status = "running"
print(run.status)
# running

run.status = "done"
print(run.status)
# done


# Invalid status
try:
    run.status = "paused"
except ValueError as e:
    print(e)
# Status must be 'pending', 'running', or 'done'


# --------------------------------
# Private attribute demonstration
# --------------------------------

# This raises AttributeError:
# print(run.__api_key)

# This works because of name mangling:
print(run._TrainingRun__api_key)
# fake-api-key-12345
```

```python
### Task 3  —  Inheritance
```

```python
class TrainingRun:

    run_count = 0

    def __init__(self, model_name, learning_rate):
        self.model_name = model_name
        self.learning_rate = learning_rate
        self._status = "pending"

        TrainingRun.run_count += 1

    def start(self):
        self.status = "running"
        print(f"{self.model_name} training started.")

    def summary(self):
        print(f"Model: {self.model_name}")
        print(f"Learning Rate: {self.learning_rate}")
        print(f"Status: {self.status}")

    @property
    def status(self):
        return self._status

    @status.setter
    def status(self, value):
        if value not in ("pending", "running", "done"):
            raise ValueError(
                "Status must be 'pending', 'running', or 'done'"
            )

        self._status = value


# --------------------------------
# Child class
# --------------------------------

class LRSchedulerRun(TrainingRun):

    def __init__(self, model_name, learning_rate, schedule):
        # Call parent constructor
        super().__init__(model_name, learning_rate)

        # Child-specific attribute
        self.schedule = schedule

    # Override parent summary()
    def summary(self):
        # Reuse parent's summary
        super().summary()

        # Add child-specific information
        print(f"Learning Rate Schedule: {self.schedule}")


# --------------------------------
# Testing
# --------------------------------

run = LRSchedulerRun(
    "ResNet",
    0.001,
    [0.001, 0.0005, 0.0001]
)

run.summary()

print("Total training runs:", TrainingRun.run_count)
```

```python
### Task 4  —  Polymorphism
```

```python
class TrainingRun:

    run_count = 0

    def __init__(self, model_name, learning_rate):
        self.model_name = model_name
        self.learning_rate = learning_rate
        self._status = "pending"

        TrainingRun.run_count += 1

    def start(self):
        self.status = "running"
        print(f"{self.model_name} training started.")

    def summary(self):
        print(f"Model: {self.model_name}")
        print(f"Learning Rate: {self.learning_rate}")
        print(f"Status: {self.status}")

    @property
    def status(self):
        return self._status

    @status.setter
    def status(self, value):
        if value not in ("pending", "running", "done"):
            raise ValueError(
                "Status must be 'pending', 'running', or 'done'"
            )

        self._status = value


# --------------------------------
# First subclass
# --------------------------------

class LRSchedulerRun(TrainingRun):

    def __init__(self, model_name, learning_rate, schedule):
        super().__init__(model_name, learning_rate)
        self.schedule = schedule

    def summary(self):
        super().summary()
        print(f"Learning Rate Schedule: {self.schedule}")


# --------------------------------
# Second subclass
# --------------------------------

class EarlyStoppingRun(TrainingRun):

    def __init__(self, model_name, learning_rate, patience):
        super().__init__(model_name, learning_rate)
        self.patience = patience

    def summary(self):
        super().summary()
        print(f"Early Stopping Patience: {self.patience} epochs")


# --------------------------------
# Polymorphic function
# --------------------------------

def print_all_summaries(runs):

    for run in runs:
        run.summary()


# --------------------------------
# Create different objects
# --------------------------------

run1 = TrainingRun(
    "BasicModel",
    0.001
)

run2 = LRSchedulerRun(
    "ResNet",
    0.001,
    [0.001, 0.0005, 0.0001]
)

run3 = EarlyStoppingRun(
    "BERT",
    0.0001,
    5
)


# --------------------------------
# Mixed list
# --------------------------------

runs = [run1, run2, run3]

print_all_summaries(runs)
```

```python
from abc import ABC, abstractmethod


# -----------------------------
# Abstract Base Class
# -----------------------------

class Step(ABC):

    @abstractmethod
    def process(self, data):
        pass


# -----------------------------
# Concrete Steps
# -----------------------------

class Cleaner(Step):

    def process(self, data):
        return data.strip()


class Tokenizer(Step):

    def process(self, data):
        return data.split()


class Normalizer(Step):

    def process(self, data):
        return [word.lower() for word in data]


# -----------------------------
# Pipeline
# -----------------------------

def run_pipeline(steps, data):

    for step in steps:
        data = step.process(data)

    return data


# -----------------------------
# Test
# -----------------------------

steps = [
    Cleaner(),
    Tokenizer(),
    Normalizer()
]

data = "  HELLO WORLD PYTHON  "

result = run_pipeline(steps, data)

print(result)


# -----------------------------
# Test that Step cannot be
# instantiated directly
# -----------------------------

try:
    step = Step()
except TypeError as e:
    print("Cannot instantiate Step:", e)
```
