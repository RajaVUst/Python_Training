```python
# Exercise 6 — Restaurant Bill Splitter
total_bill = float(input("Enter the total bill amount: "))
num_people = int(input("Enter the number of people: "))

wants_tip = input("Add a 10% tip? (yes/no): ").strip().lower() == "yes"

if wants_tip:
    total_bill *= 1.10

share_per_person = total_bill / num_people

print(f"Each person's share: ${share_per_person:.2f}")
```

```python
# Exercise 7 — Variable Swap
a = 5
b = 10

temp = a
a = b
b = temp

print(f"a = {a}")
print(f"b = {b}")
```

```python
a, b = b, a

print(f"a = {a}")
print(f"b = {b}")
```

```python
# Exercise 8 — Rectangle Calculator
length = float(input("Enter the rectangle's length: "))
width = float(input("Enter the rectangle's width: "))

area = length * width
perimeter = 2 * (length + width)

print(f"Area: {area}")
print(f"Perimeter: {perimeter}")
```

```python
# Exercise 11 — BMI Calculator
weight = float(input("Enter your weight in kilograms: "))
height = float(input("Enter your height in metres: "))

bmi = weight / (height ** 2)

print(f"Your BMI is: {bmi:.1f}")
```

```python
# Exercise 12 — Divisibility Checker
n=15
div_by_3=(n%3==0)
div_by_5=(n%5==0)
div_by_both=div_by_3 and div_by_5

print(f"Divisible by 3: {div_by_3}")
print(f"Divisible by 5: {div_by_5}")
print(f"Divisible by both: {div_by_both}")
```

```python
# Exercise 13 — Currency Converter
usd_amount = float(input("Enter the amount in US Dollars: "))
conversion_rate = float(input("Enter the USD to INR conversion rate: "))

inr_amount = usd_amount * conversion_rate

print(f"${usd_amount:,.2f} = ₹{inr_amount:,.2f}")
```

```python
# Exercise 14 — Student Report Line
score1 = float(input("Enter the first test score: "))
score2 = float(input("Enter the second test score: "))
score3 = float(input("Enter the third test score: "))

average = (score1 + score2 + score3) / 3
has_passed = average >= 40

print(f"Average: {average:.1f} | Passed: {has_passed}")
```
