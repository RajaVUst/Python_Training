```python
# Checks whether a given year is a leap year.
 
year = 2024
 
if (year % 4 == 0 and year % 100 != 0) or (year % 400 == 0):
    print("Leap Year")
else:
    print("Not a Leap Year")
```

## Part D: Loops Practice 

```python
# Generates the multiplication table of a given number from 1 to 10.
 
n = 5
 
for i in range(1, 11):
    print(f"{n} x {i} = {n * i}")
```

```python
# Calculates the sum of the first N natural numbers using a loop.
 
n = 10
total = 0
 
for i in range(1, n + 1):
    total += i
 
print("Sum =", total)
```

```python
# Calculates the sum of all digits in a positive whole number.
 
number = 1234
digit_sum = 0
 
while number > 0:
    digit_sum += number % 10
    number //= 10
 
print("Sum of digits =", digit_sum)
```

```python
# Prints a countdown from a starting number and finishes with a liftoff message.
 
start = 5
 
while start >= 1:
    print(start)
    start -= 1
 
print("Liftoff!")
```

```python
# Prints a right-angled triangle pattern using nested loops.
 
n = 5
 
for i in range(1, n + 1):
    for j in range(i):
        print("*", end="")
    print()
```

```python
# Counts the number of vowels in a word or sentence.
 
text = "Python Readiness Training"
vowel_count = 0
 
for ch in text:
    if ch.lower() in "aeiou":
        vowel_count += 1
 
print("Number of vowels =", vowel_count)
```

## Part E: Combined Logic — Classic Problems 

```python
# Prints numbers from 1 to 50 using the FizzBuzz rules.
 
for i in range(1, 51):
    if i % 15 == 0:
        print("FizzBuzz")
    elif i % 3 == 0:
        print("Fizz")
    elif i % 5 == 0:
        print("Buzz")
    else:
        print(i)
```

```python
# Reverses a positive whole number without using string slicing.
 
number = 1234
reversed_number = 0
 
while number > 0:
    reversed_number = reversed_number * 10 + (number % 10)
    number //= 10
 
print("Reversed number =", reversed_number)
```

```python
# Checks whether a positive whole number is a palindrome.
 
number = 12321
original_number = number
reversed_number = 0
 
while number > 0:
    reversed_number = reversed_number * 10 + (number % 10)
    number //= 10
 
if original_number == reversed_number:
    print("Palindrome")
else:
    print("Not a palindrome")
```

```python
# Checks whether a whole number greater than 1 is prime.
 
n = 17
is_prime = True
 
for i in range(2, n):
    if n % i == 0:
        is_prime = False
        break
 
if is_prime:
    print("Prime")
else:
    print("Not Prime")
```

```python
# Prints all prime numbers between 2 and 50.
 
for n in range(2, 51):
    is_prime = True
 
    for i in range(2, n):
        if n % i == 0:
            is_prime = False
            break
 
    if is_prime:
        print(n)
```

```python
# Calculates the factorial of a whole number using a loop.
 
n = 5

factorial = 1
 
for i in range(1, n + 1):

    factorial *= i
 
print("Factorial =", factorial)
 
# Generates the first 15 numbers in the Fibonacci series.
 
a = 0

b = 1
 
for i in range(15):

    print(a, end=" ")

    a, b = b, a + b
 
```

```python
# Converts an amount from Indian Rupees to a selected foreign currency.
 
amount_inr = 5000
currency = "USD"
 
if currency == "USD":
    converted_amount = round(amount_inr * 0.012, 2)
elif currency == "EUR":
    converted_amount = round(amount_inr * 0.011, 2)
elif currency == "GBP":
    converted_amount = round(amount_inr * 0.0095, 2)
else:
    print("Invalid currency")
    converted_amount = None
 
if converted_amount is not None:
    print(f"{amount_inr} INR = {converted_amount} {currency}")
```

```python
# Checks a list of guesses against a secret number and stops when the correct guess is found.
 
secret_number = 25
guesses = [10, 30, 20, 25, 40]
 
for guess in guesses:
    if guess > secret_number:
        print("Too high")
    elif guess < secret_number:
        print("Too low")
    else:
        print("Correct!")
        break
```

```python
# Prints a numeric pyramid pattern using nested loops.
 
n = 5
 
for row in range(1, n + 1):
    for num in range(1, row + 1):
        print(num, end=" ")
    print()
```
