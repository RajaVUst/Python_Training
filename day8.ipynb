```python
#Task1
# inventory_tracker.py

# Items received today (duplicates included because items may arrive in multiple batches)
items = ["pen", "notebook", "pen", "eraser", "notebook", "pen", "stapler"]

# List: preserves the arrival order of all received items
inventory_list = items

# Tuple: stores the first 3 received items as an immutable first-batch record
first_batch = tuple(items[:3])

# Set: keeps only unique item names, removing duplicates
unique_items = set(items)

# Dictionary: counts how many times each item was received
item_counts = {}

for item in items:
    item_counts[item] = item_counts.get(item, 0) + 1

# Print results
print("List (arrival order):", inventory_list)
print("Tuple (first batch):", first_batch)
print("Set (unique items):", unique_items)
print("Dictionary (item counts):", item_counts)
```

```python
# comprehensions_rewrite.py

# List comprehension
squares = [n * n for n in range(1, 21) if n % 2 == 0]

# Dictionary comprehension
word_lengths = {word: len(word) for word in ["python", "java", "c", "kotlin"]}

# Set comprehension
unique_vowels = {ch for ch in "the quick brown fox jumps over the lazy dog" if ch in "aeiou"}

# Print results
print("Squares of even numbers:", squares)
print("Word lengths:", word_lengths)
print("Unique vowels:", unique_vowels)
```

```python
# nested_comprehension_challenge.py

sentences = [
    "the sky is blue",
    "python is fun",
    "list comprehensions are handy"
]

# Flat list of all words longer than 2 characters across all sentences
long_words = [word for sentence in sentences for word in sentence.split() if len(word) > 2]

# Set of all unique words across all sentences (case-insensitive)
unique_words = {
    word.lower()
    for sentence in sentences
    for word in sentence.split()
}

print("Words longer than 2 characters:")
print(long_words)

print("\nUnique words (case-insensitive):")
print(unique_words)

# Checkpoint Answer:
# A nested comprehension becomes less readable when it contains multiple
# loops and complex conditions that require careful mental parsing.
# Example:
# [(x, y, z) for x in a for y in b if x > y for z in c if z % 2 == 0]
# In cases like this, a regular nested loop is often easier to understand,
# debug, and maintain.
```

```python
# student_grade_book.py

students = {
    "alice": [88, 92, 79],
    "bob": [65, 70, 68],
    "carol": [95, 91, 89],
    "david": []  # Example student with no grades
}

# Dict comprehension: student -> average grade
# Avoid ZeroDivisionError by checking whether grades exist
averages = {
    student: (sum(grades) / len(grades) if grades else 0)
    for student, grades in students.items()
}

# Set of students whose every grade is above 80
# Students with empty grade lists are excluded
high_performers = {
    student
    for student, grades in students.items()
    if grades and all(grade > 80 for grade in grades)
}

# Sort students by average grade descending
sorted_students = sorted(
    averages.items(),
    key=lambda x: x[1],
    reverse=True
)

print("Average Grades:")
print(averages)

print("\nStudents with all grades above 80:")
print(high_performers)

print("\nStudents sorted by average grade (highest first):")
print(sorted_students)
```

```python
# access_control.py

team_a = {"figma", "jira", "slack", "github"}
team_b = {"jira", "slack", "notion", "vscode"}

# Set operations
common_tools = team_a & team_b
unique_to_a = team_a - team_b
combined_tools = team_a | team_b
symmetric_difference = team_a ^ team_b

print("Common tools:", common_tools)
print("Unique to Team A:", unique_to_a)
print("Combined toolset:", combined_tools)
print("Symmetric difference:", symmetric_difference)


def access_report(team_a, team_b):
    return {
        "common": team_a & team_b,
        "unique_to_a": team_a - team_b,
        "combined": team_a | team_b,
        "symmetric_difference": team_a ^ team_b
    }


print("\nAccess Report:")
print(access_report(team_a, team_b))
```

```python
# basic_file_handling.py

logs = [
    "INFO: Application started",
    "INFO: User logged in",
    "WARNING: Low disk space",
    "ERROR: Failed to connect to database",
    "INFO: Application closed"
]

# ----------------------------------
# Write using a with block
# ----------------------------------
with open("log.txt", "w") as file:
    for log in logs:
        file.write(log + "\n")

print("Log file written successfully.\n")

# ----------------------------------
# Read back with line numbers
# ----------------------------------
with open("log.txt", "r") as file:
    for line_number, line in enumerate(file, start=1):
        print(f"{line_number}: {line.strip()}")

# ----------------------------------
# Append a new log entry
# ----------------------------------
with open("log.txt", "a") as file:
    file.write("INFO: New session started\n")

print("\nNew log entry appended.")
```

```python
# employee_records.py

import csv

employees = [
    ["Alice", "Engineering", 75000],
    ["Bob", "HR", 55000],
    ["Carol", "Engineering", 80000],
    ["David", "Sales", 60000],
    ["Eva", "Engineering", 90000]
]

# ----------------------------------
# Create employees.csv
# ----------------------------------
with open("employees.csv", "w", newline="") as file:
    writer = csv.writer(file)

    writer.writerow(["name", "department", "salary"])

    for employee in employees:
        writer.writerow(employee)

print("employees.csv created.\n")

# ----------------------------------
# Read with DictReader
# ----------------------------------
department_totals = {}
employee_records = []

print("Employee Records:")

with open("employees.csv", "r") as file:
    reader = csv.DictReader(file)

    for row in reader:
        print(row)

        salary = float(row["salary"])

        department = row["department"]

        department_totals[department] = (
            department_totals.get(department, 0) + salary
        )

        employee_records.append(row)

# ----------------------------------
# Total salary per department
# ----------------------------------
print("\nTotal Salary by Department:")

for department, total in department_totals.items():
    print(f"{department}: {total}")

# ----------------------------------
# Give Engineering employees a 10% raise
# ----------------------------------
for employee in employee_records:

    if employee["department"] == "Engineering":
        salary = float(employee["salary"])
        salary *= 1.10
        employee["salary"] = round(salary, 2)

# ----------------------------------
# Write updated CSV
# ----------------------------------
with open("employees_updated.csv", "w", newline="") as file:

    fieldnames = ["name", "department", "salary"]

    writer = csv.DictWriter(file, fieldnames=fieldnames)

    writer.writeheader()

    writer.writerows(employee_records)

print("\nemployees_updated.csv created with 10% Engineering raises.")
```

```python
# safe_divide.py

def safe_divide(a, b):
    try:
        return a / b

    except ZeroDivisionError:
        print("Warning: Cannot divide by zero.")
        return None

    except TypeError:
        print("Warning: Invalid data type supplied.")
        return None

    finally:
        print("Operation attempted")


# Test cases
print(safe_divide(10, 2))
print()

print(safe_divide(10, 0))
print()

print(safe_divide("10", 2))
```

```python
# Exception Handling Concept Check

## 1. Specific Exception vs Bare except

Catching a specific exception (such as ZeroDivisionError) means the program
handles only the error it expects.

Example:

    try:
        result = a / b
    except ZeroDivisionError:
        print("Cannot divide by zero")

A bare except:

    try:
        result = a / b
    except:
        print("Something went wrong")

catches every exception, including unexpected programming bugs such as
NameError, TypeError, KeyboardInterrupt, and SystemExit.

Why is bare except considered bad practice?

- It hides real bugs.
- It makes debugging difficult.
- It can catch exceptions that should not be ignored.
- It provides little information about what actually failed.

Specific exceptions make programs safer and easier to maintain.

---

## 2. Purpose of finally

The finally block is used to execute cleanup code that must run regardless of
whether an exception occurs.

Example:

You manually open a file:

    file = open("data.txt")

If an exception occurs before file.close(), the file may remain open.

Using finally:

    file = open("data.txt")

    try:
        process_data(file)

    finally:
        file.close()

ensures the file is always closed.

Other real-world examples include:

- Closing database connections
- Releasing network sockets
- Unlocking files
- Cleaning temporary resources

---

## 3. Why Catch File Exceptions Separately?

Python provides different exception types because different problems require
different responses.

Example:

- FileNotFoundError
    The file does not exist.

- PermissionError
    The file exists but access is denied.

- IsADirectoryError
    A directory was supplied where a file was expected.

Handling them separately allows the program to provide accurate messages and
take appropriate corrective action.

Example:

    except FileNotFoundError:
        print("Please check the filename.")

    except PermissionError:
        print("You do not have access rights.")

    except IsADirectoryError:
        print("A file was expected, not a directory.")

If all were caught with generic Exception, the user would receive the same
message for very different problems, making troubleshooting harder.
```

```python
%%writefile convert_orders.py

import csv
import json

customer_totals = {}
statuses = set()

try:
    # Read CSV using DictReader inside a with block
    with open("orders.csv", "r", newline="") as file:

        reader = csv.DictReader(file)

        for row_number, row in enumerate(reader, start=2):

            try:
                amount = float(row["amount"])

            except (ValueError, TypeError):
                print(
                    f"Warning: Invalid amount on row {row_number} "
                    f"for customer {row['customer']}. Skipping record."
                )
                continue

            customer = row["customer"]

            # Group total amount per customer
            customer_totals[customer] = (
                customer_totals.get(customer, 0) + amount
            )

            # Collect distinct statuses
            statuses.add(row["status"])

    # Create summary dictionary
    summary = {
        "customer_totals": customer_totals,
        "statuses": sorted(statuses)
    }

    # Write JSON output
    with open("orders_summary.json", "w") as file:
        json.dump(summary, file, indent=2)

    print("\nSummary written to orders_summary.json")

except FileNotFoundError:
    print("Error: orders.csv was not found.")

except Exception as e:
    print(f"Unexpected error: {e}")

finally:
    print("Conversion attempt finished")
```
