```python
#Exercise 12: Sum of digits 
num =int(input("Enter a number:"))
org_num=num
sum_of_digits = 0
while num!=0:
    sum_of_digits=sum_of_digits+num%10
    num = num//10
print(f"Sum of digits of {org_num} is {sum_of_digits}")
    
```

```python
#Exercise 17: Reverse an integer 
num = int(input("Enter a number:"))
reversed_num=0
while num!=0:
    reversed_num=reversed_num*10+(num % 10)
    num = num//10
print(reversed_num)
```

```python
#Exercise 18: Palindrome check for an integer 
num = int(input("Enter a number:"))
org_num = num
reversed_num=0
while num!=0:
    reversed_num=reversed_num*10+(num % 10)
    num = num//10
if reversed_num==org_num:
    print(f"{org_num} is a palindrome")
else:
    print(f"{org_num} is  not a palindrome")
    
```

```python
#Exercise 19: Prime number check 
num = int(input("Enter a number:"))
if num<2:
    print(f"{num} is not a prime number")
else:
    for i in range(2,num):
        if num%i==0:
            print(f"{num} is not a prime number")
            break
    else:
        print(f"{num} is a prime number")
```

```python
#Exercise 20: Print primes in a range 
for num in range(2,51):
    for i in range(2,num):
        if num%i==0:
            print(f"{num} is not a prime number")
            break
    else:
        print(f"{num} is a prime number")
    
```

```python
#Exercise 21: Factorial of a number 
n = int(input("Enter the number:"))
fact = 1
while(n>0):
    fact=fact*n
    n=n-1
print(f"Factorial of {n} is {fact}")
```

```python
#Exercise 22: Fibonacci series 
num1 = 0
num2 = 1
print("The fibonnaci series of first 15 numbers is ")
print(num1,end =",")
print(num2,end =",")
for i in range(15):
    temp = num1+num2
    num1= num2
    num2=temp
    print(temp,end=",")
print()

    
```

```python
#Exercise 23: Simple currency converter 
amt = int(input("Enter an amount in Indian Ruppes: "))
currency = input("Enter type of currency: ")
if currency == "USD":
    print(f"USD: {(amt/90):.2f}")
elif currency == "EUR":
    print(f"EUR: {(amt/120):.2f}")
elif currency == "GBP":
    print(f"GBP: {(amt/100):.2f}")
```

```python
# Exercise 24: Number guessing game (fixed target)
n = 5
l = [1,2,6,5,6]
 
for i in l:
    if n == i:
        print("Correct")
        break
    elif n-i > 0:
        print("Too low")
    elif i-n > 0:
        print("Too High")
```
