```python
# Exercise 12: Sum of digits

n = 1234

total = 0

while n > 0:
    digit = n % 10
    total = total + digit
    n = n // 10

print("Sum of digits:", total)
```

```python
# Exercise 17: Reverse an integer

n = 1234

reversed_n = 0

while n > 0:
    digit = n % 10
    reversed_n = reversed_n * 10 + digit
    n = n // 10

print("Reversed number:", reversed_n)
```

```python
# Exercise 18: Palindrome check

n = 12321

original = n
reversed_n = 0

while n > 0:
    digit = n % 10
    reversed_n = reversed_n * 10 + digit
    n = n // 10

if original == reversed_n:
    print("Palindrome")
else:
    print("Not a palindrome")
```

```python
# Exercise 19: Prime number check

n = 17

is_prime = True

for i in range(2, n):
    if n % i == 0:
        is_prime = False
        break

if is_prime:
    print("Prime number")
else:
    print("Not a prime number")
```

```python
# Exercise 20: Print primes from 2 to 50

for n in range(2, 51):

    is_prime = True

    for i in range(2, n):
        if n % i == 0:
            is_prime = False
            break

    if is_prime:
        print(n)
```

```python
# Exercise 21: Factorial

n = 5

factorial = 1

for i in range(1, n + 1):
    factorial = factorial * i

print("Factorial:", factorial)
```

```python
# Exercise 22: Fibonacci series

a = 0
b = 1

for i in range(15):
    print(a, end=" ")

    next_number = a + b
    a = b
    b = next_number
```

```python
# Exercise 23: Simple currency converter

amount = 8500
currency = "USD"

if currency == "USD":
    converted = amount / 85
elif currency == "EUR":
    converted = amount / 100
elif currency == "GBP":
    converted = amount / 115
else:
    print("Invalid currency")
    converted = None

if converted is not None:
    print("Converted amount:", round(converted, 2), currency)
```
