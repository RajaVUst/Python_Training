```python
#  Copy vs. Reference

original = [1, 2, 3]
alias = original
safe_copy = original.copy()

alias.append(100)
safe_copy.append(200)

# original changed only with alias because alias references the same list object, while safe_copy is a separate copy.
print(original)
print(alias)
print(safe_copy)
```

```python
#  Multiple Return Values

def stats(numbers):
    return min(numbers), max(numbers), sum(numbers) / len(numbers)

low, high, avg = stats([4, 9, 1, 7, 15])

print(low, high, avg)
```

```python
#  Word Frequency Counter

text = "the quick brown fox jumps over the lazy dog the fox runs"

word_counts = {}

for word in text.split():
    word_counts[word] = word_counts.get(word, 0) + 1

print(word_counts)
```

```python
# Class Roster (Nested Structure)

roster = [
    {"name": "Amit", "scores": [85, 90, 88]},
    {"name": "Reni", "scores": [92, 95, 89]},
    {"name": "Tara", "scores": [78, 82, 80]}
]

for student in roster:
    average = round(sum(student["scores"]) / len(student["scores"]), 1)
    print(f"{student['name']}: {average}")
```

```python
#  Lookup Table

grade_lookup = {
    "A": 90,
    "B": 80,
    "C": 70,
    "D": 60,
    "F": 0
}

score = 84

for grade, threshold in grade_lookup.items():
    if score >= threshold:
        print(grade)
        break
```

```python
#  Dictionary From Two Lists

names = ["Amit", "Reni", "Tara"]
scores = [78, 91, 65]

score_dict = {name: score for name, score in zip(names, scores)}

print(score_dict)
```

```python
#  Filtering With a Dict Comprehension

scores_dict = {"Amit": 78, "Reni": 91, "Tara": 65}

filtered_scores = {name: score for name, score in scores_dict.items() if score >= 70}

print(filtered_scores)
```

```python
#  Vowel Counter by Word

def vowel_counts(sentence):
    vowels = "aeiou"
    return {
        word: sum(1 for char in word.lower() if char in vowels)
        for word in sentence.split()
    }

print(vowel_counts("The Quick Brown Fox Jumps"))
```

```python
#  Grade Book Summary

def grade_summary(roster):
    summary = []

    for student in roster:
        average = sum(student["scores"]) / len(student["scores"])

        if average >= 90:
            grade = "A"
        elif average >= 80:
            grade = "B"
        elif average >= 70:
            grade = "C"
        elif average >= 60:
            grade = "D"
        else:
            grade = "F"

        summary.append({
            "name": student["name"],
            "average": round(average, 1),
            "grade": grade
        })

    return summary

roster = [
    {"name": "Amit", "scores": [78, 82, 80]},
    {"name": "Reni", "scores": [92, 95, 89]},
    {"name": "Tara", "scores": [65, 70, 68]}
]

print(grade_summary(roster))
```

```python
#  Unique Word Finder

def unique_words(text):
    return sorted(set(word.lower() for word in text.split()))

print(unique_words("Python is fun and Python is easy"))
```

```python
#  FizzBuzz, Structured

def fizzbuzz_map(n):
    result = {}

    for i in range(1, n + 1):
        if i % 3 == 0 and i % 5 == 0:
            result[i] = "FizzBuzz"
        elif i % 3 == 0:
            result[i] = "Fizz"
        elif i % 5 == 0:
            result[i] = "Buzz"
        else:
            result[i] = str(i)

    return result

print(fizzbuzz_map(20))
```

```python
#  Password Strength Report

def password_report(passwords):
    report = {
        "weak": [],
        "medium": [],
        "strong": []
    }

    for pw in passwords:
        has_digit = any(ch.isdigit() for ch in pw)

        if len(pw) >= 10 and has_digit:
            report["strong"].append(pw)
        elif len(pw) >= 6:
            report["medium"].append(pw)
        else:
            report["weak"].append(pw)

    return report

passwords = [
    "abc",
    "secret",
    "pass123",
    "mypassword1",
    "StrongPass99"
]

print(password_report(passwords))
```
