```python
# Task 1.2 — Comprehensions Rewrite
# Take this loop-based code and rewrite each block as a comprehension:
# squares = []
# for n in range(1, 21):
#     if n % 2 == 0:
#         squares.append(n * n)

# word_lengths = {}
# for word in ["python", "java", "c", "kotlin"]:
#     word_lengths[word] = len(word)

# unique_vowels = set()
# for ch in "the quick brown fox jumps over the lazy dog":
#     if ch in "aeiou":
#         unique_vowels.add(ch)
# •	Produce: a list comprehension, a dict comprehension, a set comprehension.
# •	Time yourself doing it with a loop vs a comprehension — note which felt faster to write and which felt easier to read back.


squares = [i*i for i in range(1, 21) if i%2 == 0]
print(squares)

word_lengths = { word:len(word) for word in ["python", "java", "c", "kotlin"] }
print(word_lengths)

unique_vowels = set([ ch for ch in "the quick brown fox jumps over the lazy dog" if ch in "aeiou"])
print(unique_vowels)
```

```python
# Task 1.3 — Nested Comprehension Challenge
# Scenario:  Given a list of sentences:
# sentences = ["the sky is blue", "python is fun", "list comprehensions are handy"]
# •	In one line, produce a flat list of all words longer than 2 characters across all sentences (nested comprehension).
# •	Also produce a set of all unique words across all sentences (case-insensitive).
# Checkpoint question (answer in a comment): At what point does a nested comprehension become less readable than a plain loop?
# Give your own opinion with an example.

sentences = ["the sky is blue", "python is fun", "list comprehensions are handy"]
l = [j for i in sentences for j in i.split() if len(j)>2]
print(l)

s = {j.lower() for i in sentences for j in i.split()}
print(s)

```

```python
# Collections in Practice
# Task 2.1 — Student Grade Book
# Scenario:  Build a small grade book system.
# students = {
#     "alice": [88, 92, 79],
#     "bob":   [65, 70, 68],
#     "carol": [95, 91, 89],
# }
# •	Compute a new dict mapping student → average grade (dict comprehension).
# •	Use a set to find students who have every grade above 80.
# •	Sort students by average grade descending and store as a list of (name, avg) tuples.
# •	Handle the case where a student has an empty grade list without crashing (avoid ZeroDivisionError).

students = {
    "alice": [88, 92, 79],
    "bob":   [65, 70, 68],
    "carol": [95, 91, 89],
}

res_dict = {i:sum(j)/len(j) if j else 0 for i,j in students.items() }
print(res_dict)

res_set = set()
for i,j in students.items():
    f = True
    for s in j:
        if s<=80:
            f = False
            break
    if f:
        res_set.add(i)

print(res_set)

l = sorted(res_dict.items(), key=lambda x: x[1], reverse=True)
print(l)

```

```python
# Task 3.2 — CSV: Employee Records
# Scenario:  Create employees.csv with columns name,department,salary for 5 employees using the csv module (csv.writer).
# •	Read it back using csv.DictReader and print each employee as a dict.
# •	Compute total salary per department (combine with Hour 1/2 dict skills!).
# •	Add a raise of 10% to everyone in "Engineering" and write out employees_updated.csv.

import csv 

employees = [ ["Varshith", "Engineering", 70000], ["Vinaya", "HR", 50000], ["Aish", "Engineering", 80000], ["Prasanna", "Sales", 60000], ["Ajay", "HR", 55000]]
with open('employees.csv', 'w', newline="") as f:
    writer = csv.writer(f)
    writer.writerow(["name", "department", "salary"])
    writer.writerows(employees)

with open("employees.csv", "r", newline="") as f:
    reader = csv.DictReader(f)
    for row in reader:
        print(row)
print()
dept_salary = {}
with open("employees.csv", "r", newline="") as f:
    reader = csv.DictReader(f)
    for row in reader:
        dept = row["department"]
        salary = float(row["salary"])
        dept_salary[dept] = dept_salary.get(dept, 0) + salary
print(dept_salary)

updated_rows = []
with open("employees.csv", "r", newline="") as f:
    reader = csv.DictReader(f)
    for row in reader:
        if row["department"] == "Engineering":
            row["salary"] = str(float(row["salary"]) * 1.10)
        updated_rows.append(row)

with open("employees_updated.csv", "w", newline="") as f:
    fieldnames = ["name", "department", "salary"]
    writer = csv.DictWriter(f, fieldnames=fieldnames)
    writer.writeheader()
    writer.writerows(updated_rows)
```

```python
# Exceptions + Capstone
# Task 4.1 — try/except/finally Basics
# Write a function safe_divide(a, b) that:
# •	Returns a / b.
# •	Catches ZeroDivisionError and returns None with a printed warning.
# •	Uses finally to print "Operation attempted" no matter what.
# •	Also catches TypeError (e.g., calling safe_divide("10", 2)).


def safe_divide(a, b):
    try:
        return a / b

    except ZeroDivisionError:
        print("Warning: Cannot divide by zero.")
        return None

    except TypeError:
        print("Warning: Invalid operand type.")
        return None

    finally:
        print("Operation attempted")

print(safe_divide(10, 2))
```

```python
# Task 4.3 — Capstone: Robust CSV → JSON Converter
# Scenario:  Combine everything from today into one script, convert_orders.py.
# Input: an orders.csv file with columns order_id,customer,amount,status (create sample data with at least 8 rows, include 1-2 rows with a missing or non-numeric amount to force error handling).
# Requirements:
# •	Read the CSV using csv.DictReader inside a with block.
# •	For each row, try to convert amount to float; if it fails, catch the exception, log a warning (which row/customer), and skip that row instead of crashing the whole program.
# •	Use a comprehension or dict-building loop to group total amount per customer (dict) and a set of all distinct statuses seen.
# •	Write the final summary out to orders_summary.json using json.dump.
# •	Wrap the whole file-handling section in try/except/finally, catching FileNotFoundError specifically (in case orders.csv is missing) and a generic Exception as a fallback, with finally printing "Conversion attempt finished".


import csv
import json

rows = [
    ["order_id", "customer", "amount", "status"],
    ["1", "Varshith", "1000", "Completed"],
    ["2", "Vinaya", "2000", "Pending"],
    ["3", "Aishu", "1500", "Completed"],
    ["4", "Prasanna", "abc", "Cancelled"],
    ["5", "Ajay", "", "Pending"],
    ["6", "Varshith", "2500", "Completed"],
    ["7", "Vinaya", "3000", "Shipped"],
    ["8", "Ajay", "1800", "Completed"]
]

with open("orders.csv", "w", newline="") as f:
    csv.writer(f).writerows(rows)

try:
    totals = {}
    statuses = set()

    with open("orders.csv", "r") as f:
        reader = csv.DictReader(f)

        for row in reader:
            try:
                amount = float(row["amount"])
                customer = row["customer"]

                totals[customer] = totals.get(customer, 0) + amount
                statuses.add(row["status"])

            except (ValueError, TypeError):
                print(f"Warning: Invalid amount for {row['customer']}")

    summary = {
        "customer_totals": totals,
        "statuses": list(statuses)
    }

    with open("orders_summary.json", "w") as f:
        json.dump(summary, f, indent=2)

except FileNotFoundError:
    print("orders.csv not found")

except Exception as e:
    print(e)

finally:
    print("Conversion attempt finished")
```
